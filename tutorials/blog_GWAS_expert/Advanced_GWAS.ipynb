## Advanced GWAS Theory and Mixed Model Interpretation

This article is designed for readers who already understand basic GWAS workflows and want to explore the deeper statistical genetics principles behind modern association studies.

### What this article covers

- Genetic architecture and the infinitesimal model
- Random-effects linear mixed models and SNP heritability
- Interpretation of the genetic relationship matrix (GRM)
- Control for population structure and cryptic relatedness
- Differences between fixed-effect and mixed-model GWAS
- Why genomic inflation is not always confounding
- Advanced diagnostics: LD score regression, QQ deviations, and heritability estimates

### Why this matters

Expert-level GWAS practice requires more than running PLINK or GCTA commands. It requires understanding the assumptions that underlie the test statistics, how population structure and relatedness distort those statistics, and how mixed models partition variance to recover unbiased SNP associations.

### Core concepts we will explore

#### 1. The mixed model and polygenic covariance

A standard GWAS linear mixed model can be written as:

$$
y = X\beta + g + \epsilon
$$

where:

- $y$ is the phenotype vector
- $X$ contains fixed covariates such as age, sex, and PCs
- $\beta$ are the fixed-effect coefficients
- $g$ is the polygenic random effect
- $\epsilon$ is the residual error

The random effect is modeled as:

$$
g \sim \mathcal{N}(0, \sigma_g^2 K)
$$

and the residual error as:

$$
\epsilon \sim \mathcal{N}(0, \sigma_e^2 I)
$$

Here, $K$ is the genetic relationship matrix (GRM). The GRM acts as a covariance matrix for the polygenic background and is the key to interpreting mixed-model GWAS.

#### 2. Constructing the GRM and scaling issues

The GRM is typically constructed from standardized genotype data. If $G$ is the genotype matrix coded as 0/1/2, and $p_j$ is the allele frequency of SNP $j$, then the standardized genotype matrix is:

$$
Z_{ij} = \frac{G_{ij} - 2p_j}{\sqrt{2p_j(1-p_j)}}
$$

A common GRM estimator is:

$$
K = \frac{1}{M} Z Z^T
$$

where $M$ is the number of SNPs. This scaling makes the diagonal entries approximately 1 and the off-diagonals interpretable as realized genomic similarity.

**Expert note:** Different software may use alternative scalings. If $K$ is not scaled correctly, estimates of $\sigma_g^2$ and SNP heritability will change.

#### 3. Heritability on observed and liability scales

The narrow-sense SNP heritability is:

$$
h^2 = \frac{\sigma_g^2}{\sigma_g^2 + \sigma_e^2}
$$

For binary traits, GWAS results are reported on the observed scale, but the underlying genetic architecture is often interpreted on the liability scale. The liability-scale heritability is:

$$
h^2_{L} = \frac{h^2_{O} K^2 (1-K)^2}{z^2 P (1-P)}
$$

where:

- $h^2_{O}$ is observed-scale heritability
- $K$ is population prevalence
- $P$ is sample prevalence
- $z$ is the standard normal density at the liability threshold

This transformation is a critical expert-level step for comparing heritability across studies with different case-control ratios.

#### 4. Fixed-effect GWAS versus mixed-model GWAS

A fixed-effect GWAS typically models:

$$
y = X\beta + G_j \alpha_j + \epsilon
$$

with $G_j$ representing the genotype for SNP $j$ and covariates such as PCs included explicitly. Mixed-model GWAS instead uses a random effect for background genetics:

$$
y = X\beta + G_j \alpha_j + u + \epsilon$$

The random effect $u$ absorbs polygenic signal and relatedness. This is especially important when:

- there is cryptic relatedness among samples
- population structure is subtle or not fully captured by PCs
- polygenic signal is strong enough to inflate test statistics

#### 5. Interpreting genomic inflation and LD score regression

The genomic inflation factor $\lambda_{GC}$ is not a binary indicator of confounding. It can reflect:

- population stratification
- cryptic relatedness
- polygenic architecture
- sample ascertainment

Advanced practice uses LD score regression to separate confounding from polygenicity. The LD score regression intercept estimates inflation from confounding, while the slope estimates true polygenic signal.

### Practical expert-level workflow

1. Build and inspect the GRM.
2. Compare PCA-based adjustment and mixed-model results.
3. Compute SNP heritability and check its scale.
4. Evaluate LD score regression intercept and slope.
5. Inspect QQ plots, effect size distributions, and mixed-model diagnostics.

### Example: constructing the GRM in Python

```python
import numpy as np
import pandas as pd

# G is an N x M genotype matrix coded as 0/1/2
# p is the vector of allele frequencies for each SNP
p = G.mean(axis=0) / 2.0

# standardize genotypes
Z = (G - 2 * p) / np.sqrt(2 * p * (1 - p))

# construct GRM
M = G.shape[1]
K = Z @ Z.T / M

# make sure the diagonal is approximately 1
diag = np.diag(K)
print('GRM diag mean:', diag.mean())
```

This matrix $K$ is the foundation of mixed-model GWAS. In practice, tools such as GCTA, BOLT-LMM, or FastGWA build and use the GRM internally.

### Example: interpreting a random effect model

In a mixed-model GWAS, the association statistic for SNP $j$ can be seen as testing whether $\alpha_j \neq 0$ after conditioning on the random effect. The random effect effectively reduces inflation caused by background genetic similarity.

A simplified view of the mixed-model likelihood is:

$$
\ell(\theta) = -\frac{1}{2} \left[ \log |V| + (y - X\beta)^T V^{-1} (y - X\beta) \right]
$$

where $V = \sigma_g^2 K + \sigma_e^2 I$. Estimating $\sigma_g^2$ and $\sigma_e^2$ by REML is a core expert step.
