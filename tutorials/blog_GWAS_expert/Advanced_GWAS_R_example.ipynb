This R notebook demonstrates similar steps: simulate genotype data, compute PCs from a GRM, and run linear models with and without PC covariates. It uses base R and `stats` functions so it runs without extra packages.

```python
set.seed(1)
N <- 800
M <- 2000
p <- runif(M, 0.05, 0.5)
G <- matrix(rbinom(N * M, 2, rep(p, each = N)), nrow = N, ncol = M)
# causal SNP at column 1
beta <- rep(0, M)
beta[1] <- 0.6
polygenic <- G %*% rnorm(M, 0, 0.01)
pheno <- G[,1] * beta[1] + polygenic + rnorm(N, 0, 1)
```
