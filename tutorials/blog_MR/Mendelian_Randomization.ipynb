One of the fundamental goals of science is to determine whether one variable causes another. In medicine, epidemiology, genetics, economics, and social sciences, researchers often observe relationships between variables. However, observing a relationship is not the same as proving a causal effect.

For example:

* Individuals with higher body mass index (BMI) tend to have a higher risk of coronary heart disease.
* Smokers are more likely to develop lung cancer.
* Elevated C-reactive protein (CRP) levels are associated with hypertension.
* Individuals who exercise regularly tend to live longer.

These observations are informative, but they do not automatically establish causality.

Understanding the distinction between correlation and causation is one of the most important concepts in statistics and epidemiology. It is also the motivation behind Mendelian Randomization, a powerful approach that uses genetic variants to investigate causal relationships.

### What Is Correlation?

Correlation describes the extent to which two variables vary together.

When one variable increases and another tends to increase as well, the variables have a positive correlation.

When one variable increases while another decreases, they have a negative correlation.

Mathematically, correlation is often measured using the Pearson correlation coefficient: r


where:$
-1 \le r \le 1
$

Interpretation:

| Correlation | Interpretation                |
| ----------- | ----------------------------- |
| (r = 1)     | Perfect positive relationship |
| (r = 0)     | No linear relationship        |
| (r = -1)    | Perfect negative relationship |

For example, height and weight are usually positively correlated because taller individuals generally weigh more than shorter individuals.

However, a correlation does not tell us whether one variable causes the other.


### The Central Question

Suppose researchers observe that individuals with elevated CRP levels tend to have higher blood pressure.

This observation raises an important question:

> Does CRP cause high blood pressure?

Several alternative explanations are possible:

1. CRP causes high blood pressure.
2. High blood pressure causes elevated CRP.
3. Another factor influences both CRP and blood pressure.
4. The relationship is partly causal and partly confounded.

Observational data alone often cannot distinguish between these possibilities.

#### Correlation Does Not Imply Causation

One of the most widely quoted principles in statistics is:

> Correlation does not imply causation.

Two variables may be strongly correlated even when neither directly causes the other.

There are several reasons why this can occur.

#### Reverse Causation

Reverse causation occurs when the true direction of the relationship is opposite to what we initially assume.

Researchers may observe

Biomarker → Disease and conclude 
Biomarker → Disease,

when the true relationship is: Disease → Biomarker


For example, elevated inflammatory markers may be observed in individuals with disease.

This does not necessarily mean the inflammatory marker caused the disease. The disease itself may have increased the biomarker level.

Reverse causation is particularly problematic in observational studies because measurements are often collected after disease processes have already begun.

#### Confounding

Confounding occurs when a third variable influences both the exposure and the outcome.

Consider the following example:

Health → Consciousness

influences both: Exercise and Healthy Diet.



Individuals who exercise frequently may also eat healthier diets.

A researcher might observe a strong association between exercise and diet quality, but this relationship is partly explained by the underlying tendency to engage in healthy behaviors.

The confounding variable creates an association that may not represent a direct causal effect.

#### A Classic Example: Ice Cream and Drowning

Imagine collecting data across an entire year and observing the following:

* Ice cream sales increase during summer.
* Drowning incidents also increase during summer.

The data may show a strong positive correlation between ice cream consumption and drowning.

At first glance, one might incorrectly conclude:

$
Ice\ Cream → Drowning
$

Clearly this is not the true explanation.

Instead:

$
Temperature →Ice\ Cream\ Sales
$

and

$
Temperature → Drowning
$

Temperature is a confounder that influences both variables.

This simple example illustrates why associations alone are insufficient for establishing causality.

#### Why Observational Studies Are Challenging

Most epidemiological studies are observational.

Researchers observe naturally occurring variation in exposures and outcomes without controlling who receives the exposure.

Examples include:

* Smoking behavior
* Alcohol consumption
* Obesity
* Physical activity
* Diet
* Blood biomarkers

Observational studies are valuable because they are often inexpensive and can involve very large populations.

However, observational estimates are frequently affected by:

* Confounding
* Reverse causation
* Measurement error
* Selection bias

As a result, observed associations may not reflect true causal effects.

#### The Counterfactual Problem

A fundamental challenge in causal inference is that we cannot observe alternative realities.

Consider an individual with obesity.

We observe their actual outcome:

$
Obesity → Outcome
$

But we cannot simultaneously observe what would have happened had the same person not been obese.

This unobservable alternative outcome is called the counterfactual.

Because counterfactual outcomes cannot be directly observed, causal inference is inherently difficult.

### Randomized Controlled Trials

The gold standard for causal inference is the randomized controlled trial (RCT).

In an RCT, participants are randomly assigned to:

* Treatment group
* Control group

Randomization helps balance both measured and unmeasured confounders across groups.

Conceptually, Randomization leads to Balanced Confounders which allows us to estimate Treatment → Outcome


with reduced bias.

If the treatment and control groups differ in outcomes after randomization, the difference is more likely to represent a causal effect.

#### Why Not Use RCTs for Everything?

Although randomized trials are powerful, they are often impossible, unethical, or impractical.

Examples include:

* Randomly assigning individuals to smoke cigarettes.
* Randomly assigning obesity.
* Randomly assigning alcohol addiction.
* Randomly assigning high blood pressure.

For many important public health questions, randomized experiments cannot be performed.

Scientists therefore need alternative methods to investigate causality.



#### Nature's Randomized Experiment

Remarkably, genetics provides a natural form of randomization.

During reproduction, genetic variants are transmitted from parents to offspring according to Mendel's laws.

Because this transmission is largely random, genetic variants are generally assigned before:

* Birth
* Disease onset
* Lifestyle choices
* Environmental exposures

This natural randomization creates an opportunity to study causal relationships using genetic information.

The central idea is simple:

> If genetic variants influence an exposure, and that exposure truly causes an outcome, then the genetic variants should also be associated with the outcome.

This insight forms the foundation of Mendelian Randomization.

* Genetic variants provide a natural source of randomization.

### What Is Mendelian Randomization?

In Part 1, we discussed why establishing causality is difficult. Observational studies are often affected by confounding, reverse causation, and other forms of bias. While randomized controlled trials (RCTs) are considered the gold standard for causal inference, they are not always feasible or ethical.

This raises an important question:

> Can we obtain evidence about causality without performing a randomized experiment?

Mendelian Randomization (MR) is one approach that attempts to answer this question using genetic variation.

After completing this tutorial, you should be able to:

* Explain the basic idea behind Mendelian Randomization.
* Understand the analogy between randomized controlled trials and genetics.

### The Central Idea Behind Mendelian Randomization

Mendelian Randomization is a method that uses genetic variants as proxies for modifiable exposures.

The key insight is that genetic variants are assigned at conception according to Mendel's laws of inheritance.

Because genetic variants are determined before birth:

* They cannot be influenced by later disease processes.
* They are generally unaffected by lifestyle choices.
* They are usually independent of many environmental confounders.

This makes genetic variants useful instruments for studying causality.

#### A Motivating Example

Suppose researchers observe that individuals with elevated C-reactive protein (CRP) levels tend to have higher blood pressure.

The observational association may be written as:


CRP → Blood\ Pressure


However, observational studies cannot easily determine whether:

$
CRP → Blood\ Pressure
$

or

$
Blood\ Pressure → CRP
$

or whether both are influenced by confounding variables.

Mendelian Randomization attempts to answer this question using genetic variants associated with CRP.

### Mendelian Randomization and Randomized Controlled Trials

The logic of Mendelian Randomization closely resembles that of a randomized controlled trial.

#### Randomized Controlled Trial

In a randomized trial:

1. Participants are randomly assigned to treatment groups.
2. Randomization balances confounding factors.
3. Differences in outcomes can be attributed to the treatment.

Conceptually:

$
Randomization → Exposure → Outcome
$

Because treatment assignment is random, confounding is minimized.

### Mendelian Randomization as Nature's Experiment

In Mendelian Randomization, we replace treatment assignment with genetic inheritance.

Instead of randomly assigning individuals to receive a treatment, nature randomly assigns genetic variants during reproduction.

Conceptually:

$
Genetic\ Variant → Exposure → Outcome
$

The random segregation of alleles serves as a natural experiment.

This idea is the foundation of Mendelian Randomization.


#### Example: Smoking and Lung Cancer

Consider a genetic variant that influences smoking behavior.

Suppose:

* Individuals carrying one genotype tend to smoke heavily.
* Individuals carrying another genotype tend to smoke less.

If smoking truly causes lung cancer, then individuals carrying the smoking-promoting genotype should also have a higher risk of lung cancer.

The genetic variant acts as a proxy for smoking behavior.

Instead of directly comparing smokers and non-smokers, we compare groups defined by genotype.

This comparison is less vulnerable to many traditional confounders.

### What Is an Instrumental Variable?

An instrumental variable is a variable that helps us estimate a causal effect when confounding is present.

In Mendelian Randomization:

$
Genetic\ Variant
$

acts as the instrument.

The exposure might be:

* BMI
* Smoking
* Cholesterol
* Blood pressure
* CRP
* Alcohol consumption

The outcome might be:

* Coronary heart disease
* Stroke
* Diabetes
* Cancer
* Depression

The causal framework becomes:

$
Genetic\ Variant
\rightarrow
Exposure
\rightarrow
Outcome
$

The genetic variant should influence the outcome only through the exposure.


### Instrumental Variable Framework

The basic MR model consists of three components.

#### Exposure

A modifiable factor of interest.

Examples:

* BMI
* Smoking
* LDL cholesterol
* CRP
* Physical activity

#### Outcome

A disease or trait.

Examples:

* Coronary heart disease
* Stroke
* Type 2 diabetes
* Alzheimer's disease

#### Instrument

A genetic variant associated with the exposure.

Examples:

* SNPs associated with BMI
* SNPs associated with cholesterol
* SNPs associated with smoking behavior

### The Three Core Assumptions of Mendelian Randomization

Every Mendelian Randomization study depends on three fundamental assumptions.

Violation of these assumptions can lead to biased causal estimates.

#### Assumption 1: Relevance

The genetic variant must be associated with the exposure.

$
SNP → Exposure
$

This is known as the relevance assumption.

For example:

$
SNP → BMI
$

If the SNP has no relationship with BMI, it cannot provide information about whether BMI affects disease.

Strong instruments produce more reliable MR estimates.

Weak instruments can lead to biased results.

#### Assumption 2: Independence

The genetic variant should not be associated with confounding variables.

$
SNP \not\leftrightarrow Confounders
$

Examples of potential confounders:

* Income
* Education
* Diet
* Exercise
* Socioeconomic status

If the genetic variant is correlated with confounders, the instrument may no longer behave like a randomized experiment.

This violates the independence assumption.

#### Assumption 3: Exclusion Restriction

The genetic variant must influence the outcome only through the exposure.

The desired pathway is:

$
SNP
→Exposure
→Outcome
$

The following pathway should not exist:

$
SNP
→Outcome
$

without going through the exposure.

This assumption is often the most difficult to verify.



### Understanding Pleiotropy

A major threat to Mendelian Randomization is pleiotropy.

Pleiotropy occurs when a genetic variant influences multiple traits.

For example:

$
SNP
→
BMI
$

and

$
SNP
→
Blood\ Pressure
$

If we are studying:

$
BMI → Blood\ Pressure
$

the direct SNP effect on blood pressure creates problems.

This violates the exclusion restriction assumption.

Later tutorials will discuss methods such as MR-Egger that attempt to detect and correct for pleiotropy.

### Why Genetic Variants Are Useful Instruments

Genetic variants have several attractive properties.

#### They Are Fixed at Conception

Genotypes are established before birth.

Disease cannot alter an individual's genotype.

This greatly reduces reverse causation.

#### They Usually Precede Disease

Genetic variants exist long before disease develops.

This provides a natural temporal ordering:

$
Genotype
→Exposure
→Outcome
$

#### They Are Less Influenced by Lifestyle Factors

Unlike many environmental exposures, genotypes are generally not modified by:

* Diet
* Exercise
* Smoking
* Medication use

This helps reduce confounding.

#### A Conceptual Example

Suppose a SNP is associated with higher CRP levels.

Researchers estimate:

$
SNP → CRP
$

They then test whether the same SNP is associated with blood pressure:

$
SNP → Blood\ Pressure
$

If:

1. The SNP influences CRP.
2. The SNP influences blood pressure.
3. The three MR assumptions hold.

then this provides evidence that CRP may causally influence blood pressure.

This is the central logic of Mendelian Randomization.


### Strengths of Mendelian Randomization

Compared with traditional observational studies, MR offers several advantages.

#### Reduced Confounding

Genotypes are generally independent of many environmental factors.

#### Reduced Reverse Causation

Disease processes cannot change inherited genetic variants.

#### Use of Existing GWAS Data

Modern MR studies can leverage large publicly available GWAS summary statistics.

#### Ethical Feasibility

MR can investigate questions that would be impossible or unethical to study through randomized trials.

### Limitations of Mendelian Randomization

MR is not a perfect substitute for randomized controlled trials.

Important challenges include:

* Weak instruments
* Population stratification
* Horizontal pleiotropy
* Selection bias
* Measurement error

For this reason, MR studies typically perform multiple sensitivity analyses to assess robustness.

### Looking Ahead

Now that we understand the basic framework of Mendelian Randomization, we are ready to estimate causal effects.

In the next tutorial, we will introduce:

* The Wald Ratio
* Single-SNP Mendelian Randomization
* Estimating causal effects mathematically
* The relationship between SNP-exposure and SNP-outcome effects

This will provide the first quantitative Mendelian Randomization estimator.

### Key Takeaways

* Mendelian Randomization uses genetic variants as instrumental variables.
* Genetic variants act as naturally randomized proxies for exposures.
* Mendelian Randomization is conceptually similar to a randomized controlled trial.
* Every MR study depends on three assumptions:

  * Relevance
  * Independence
  * Exclusion Restriction
* Pleiotropy is a major threat to causal inference in MR.
* Genetic variants can provide evidence about causal relationships when randomized experiments are impossible.


### The Wald Ratio and Single-SNP Mendelian Randomization

In Part 2, we introduced Mendelian Randomization (MR) and discussed how genetic variants can be used as instrumental variables to investigate causal relationships between exposures and outcomes.

We learned that a valid Mendelian Randomization analysis depends on three assumptions:

1. The genetic variant is associated with the exposure.
2. The genetic variant is independent of confounders.
3. The genetic variant influences the outcome only through the exposure.

Now we are ready to answer an important question:

> How do we actually estimate a causal effect using a genetic variant?

The simplest Mendelian Randomization estimator is called the **Wald Ratio**.

### Learning Objectives

After completing this tutorial, you should be able to:

* Understand the logic behind the Wald Ratio.
* Estimate a causal effect using a single SNP.
* Interpret SNP-exposure and SNP-outcome effects.

* Recognize the strengths and limitations of single-SNP Mendelian Randomization.


### From Instrumental Variables to Causal Effects

Recall the basic Mendelian Randomization framework:


SNP → Exposure→Outcome


Suppose:

* The SNP influences the exposure.
* The exposure influences the outcome.
* The MR assumptions hold.

If these assumptions are satisfied, the SNP should also be associated with the outcome.

This gives us two measurable relationships:

### SNP → Exposure

$$
\beta_{GX}
$$

where:

* (G) represents the genetic variant.
* (X) represents the exposure.

### SNP → Outcome

$$\beta_{GY}$$

where:

* (Y) represents the outcome.

These two quantities form the foundation of the Wald Ratio.

### The Intuition Behind the Wald Ratio

Imagine a SNP increases BMI by:

[
0.5
]

units per allele.

Suppose the same SNP increases coronary heart disease risk by:

[
0.1
]

units per allele.

The natural question is:

> How much does coronary heart disease increase per unit increase in BMI?

We can estimate this by dividing the SNP-outcome effect by the SNP-exposure effect.

Conceptually:

$
\frac{Outcome\ Change}{Exposure\ Change}
$

This is exactly what the Wald Ratio does.

### The Wald Ratio Formula

The causal estimate is:

$$y=\frac{\beta_{GY}}{\beta_{GX}}$$

or more explicitly:

$$
\hat{\beta}_{MR}
=\frac{\beta_{GY}}
{\beta_{GX}}
$$

where:

* $\beta_{GY}$ is the SNP-outcome effect.
* $\beta_{GX}$ is the SNP-exposure effect.

This ratio estimates the causal effect of the exposure on the outcome.


#### Why Does the Wald Ratio Work?

Suppose the true causal effect is:

$
\beta_{XY}
$

If the SNP affects the outcome only through the exposure, then:

$
\beta_{GY}
=\beta_{GX}
\times
\beta_{XY}
$

Rearranging gives:

$
\beta_{XY}
=
\frac{\beta_{GY}}
{\beta_{GX}}
$

This is the key idea behind the Wald Ratio estimator.

The SNP acts as a naturally randomized perturbation of the exposure.

The resulting change in the outcome provides information about the causal effect.


### A Numerical Example

Suppose a GWAS reports:

### SNP → BMI

$
\beta_{GX}
=0.20
$

This means each additional effect allele increases BMI by 0.20 units.

#### SNP → Coronary Heart Disease

$
\beta_{GY}
=0.08
$

This means each additional effect allele increases coronary heart disease risk by 0.08 units.

The Wald estimate becomes:

$
\hat{\beta}_{MR}
=\frac{0.08}{0.20} =
0.40
$

Interpretation:

> A one-unit increase in BMI is associated with a 0.40-unit increase in coronary heart disease risk.

assuming the MR assumptions are valid.


### Visualizing the Logic

The SNP influences the exposure:

$
SNP
\rightarrow
BMI
$

The SNP also influences the outcome:

$
SNP
\rightarrow
CHD
$

The Wald Ratio asks:

> How much outcome change occurs for each unit of exposure change induced by the SNP?

The estimate therefore isolates the exposure-mediated pathway.

### Example: C-Reactive Protein and Blood Pressure

Consider the CRP example introduced earlier.

Researchers observed:

$
CRP
\leftrightarrow
Blood\ Pressure
$

Observationally, higher CRP levels appeared associated with higher blood pressure.

To investigate causality, a SNP associated with CRP was used as an instrument.

The MR estimate was very different from the observational estimate.

This illustrates an important lesson:

> Observational associations may not reflect causal effects.

The Wald Ratio provides a way to estimate causality under the MR assumptions.

#### Estimating Uncertainty

Every statistical estimate has uncertainty.

The Wald Ratio is no exception.

Suppose we have:

$
SE(\beta_{GX})
$

and

$
SE(\beta_{GY})
$

The uncertainty in the ratio can be approximated using the Delta Method.

A commonly used approximation is:

$
SE(\hat{\beta}_{MR})
=\sqrt{
\frac{
SE(\beta_{GY})^2
}
{
\beta_{GX}^2
}
}
$

More complete versions incorporate uncertainty in both numerator and denominator.

Most MR software performs these calculations automatically.



### Confidence Intervals

Once we obtain:

$
\hat{\beta}_{MR}
$

and

$
SE(\hat{\beta}_{MR})
$

we can calculate a 95% confidence interval:

$
\hat{\beta}_{MR}
\pm
1.96 \times SE
$

For example:

Estimate:

$
0.40
$

Standard Error:

$
0.10
$

Confidence Interval:

$
0.40 \pm 1.96(0.10)
$

$
0.40 \pm 0.196
$

$
(0.204,;0.596)
$

Because zero is not contained within the interval, the estimate would be statistically significant at the 5% level.

### The Importance of Instrument Strength

The Wald Ratio relies heavily on the denominator:

$
\beta_{GX}
$

If the SNP has only a very weak association with the exposure, the ratio becomes unstable.

For example:

$
\beta_{GY}
=0.02
$

$
\beta_{GX}
=0.001
$

Then:

$
\hat{\beta}_{MR}=20
$

A tiny denominator can produce extremely large and unreliable estimates.

This is known as the weak instrument problem.

Strong instruments are essential for reliable Mendelian Randomization analyses.

### Advantages of the Wald Ratio

The Wald Ratio has several advantages.

#### Simplicity

Only one SNP is required.

#### Easy Interpretation

The estimate directly represents a causal effect.

#### Historical Importance

The Wald Ratio forms the foundation of many more advanced Mendelian Randomization methods.

### Limitations of the Wald Ratio

Despite its simplicity, the Wald Ratio has important limitations.

#### Sensitive to Pleiotropy

If the SNP influences the outcome through pathways other than the exposure, the estimate becomes biased.

#### Weak Instrument Problems

Weak SNP-exposure associations can produce unstable estimates.

#### Limited Power

Using only one SNP often provides limited statistical power.

Modern MR studies therefore typically combine many SNPs.

### Why Multiple SNPs Are Better

Suppose we identify:

* 50 BMI-associated SNPs
* 100 cholesterol-associated SNPs
* 200 smoking-associated SNPs

Each SNP provides a separate Wald Ratio estimate.

Rather than relying on a single SNP, we can combine information across many instruments.

This increases:

* Precision
* Power
* Robustness

The most widely used approach for combining multiple instruments is called:

**Inverse Variance Weighted (IVW) Mendelian Randomization.**

However, modern MR studies rarely rely on a single SNP.

Instead, they typically combine information from many genetic instruments.

* It uses a single SNP as an instrumental variable.
* The causal estimate is obtained by dividing the SNP-outcome effect by the SNP-exposure effect.
* Strong instruments are essential for reliable estimates.
* Weak instruments can produce unstable results.
* The Wald Ratio provides the conceptual foundation for more advanced MR methods.
* Modern MR studies usually combine multiple SNPs using IVW and related approaches.


### Multiple-SNP Mendelian Randomization and the Inverse Variance Weighted (IVW) Method

In Part 3, we introduced the Wald Ratio, the simplest Mendelian Randomization (MR) estimator. The Wald Ratio uses a single genetic variant as an instrumental variable to estimate the causal effect of an exposure on an outcome.

While the Wald Ratio provides an intuitive introduction to Mendelian Randomization, modern MR studies rarely rely on a single SNP. Instead, researchers typically use dozens, hundreds, or even thousands of genetic variants associated with the exposure of interest.

This raises an important question:

> How can we combine information from multiple genetic instruments to obtain a more precise causal estimate?

The most widely used approach is called the **Inverse Variance Weighted (IVW) Method**.

#### Learning Objectives

After completing this tutorial, you should be able to:

* Understand why multiple SNPs are preferable to a single SNP.
* Explain how individual Wald Ratio estimates are combined.
* Understand the logic behind inverse variance weighting.
* Calculate an IVW estimate conceptually.
* Interpret IVW results.
* Understand the assumptions underlying IVW Mendelian Randomization.

#### Why One SNP Is Usually Not Enough

Suppose we identify a SNP associated with BMI.

Using the Wald Ratio, we can estimate:

[
BMI \rightarrow Coronary\ Heart\ Disease
]

However, relying on a single SNP creates several challenges:

#### Limited Statistical Power

A single SNP usually explains only a tiny proportion of variation in the exposure.

For example:

* One BMI-associated SNP may explain less than 0.1% of BMI variation.

This limits precision.

#### Sensitivity to Random Error

Any measurement error in the SNP-exposure or SNP-outcome estimate directly affects the causal estimate.

#### Vulnerability to Pleiotropy

If the SNP violates one of the MR assumptions, the entire analysis becomes biased.

For these reasons, modern MR studies typically use many genetic instruments.

### Multiple Instruments

Imagine a GWAS identifies five SNPs associated with BMI.

| SNP  | SNP → BMI ((\beta_{GX})) | SNP → CHD ((\beta_{GY})) |
| ---- | ------------------------ | ------------------------ |
| SNP1 | 0.20                     | 0.08                     |
| SNP2 | 0.15                     | 0.06                     |
| SNP3 | 0.10                     | 0.05                     |
| SNP4 | 0.25                     | 0.09                     |
| SNP5 | 0.18                     | 0.07                     |

Each SNP can produce its own Wald Ratio estimate.

For SNP1:

$
\frac{0.08}{0.20}
=
0.40
$

For SNP2:

$
\frac{0.06}{0.15}
=
0.40
$

For SNP3:

$
\frac{0.05}{0.10}
=0.50
$

For SNP4:

$
\frac{0.09}{0.25}
=
0.36
$

For SNP5:

$\frac{0.07}{0.18}
=
0.39
$

Now we have multiple estimates of the same causal effect.

The challenge is determining how to combine them.



### Why Not Simply Average Them?

A simple average would treat all SNPs equally.

However, not all SNPs provide equally precise information.

Consider:

* SNP A estimated from a GWAS with a very small standard error.
* SNP B estimated from a GWAS with a large standard error.

Clearly, SNP A provides more reliable information.

Therefore, more precise SNPs should contribute more heavily to the final estimate.

This motivates inverse variance weighting.

### The Principle of Inverse Variance Weighting

Inverse variance weighting is a standard statistical technique used in meta-analysis.

The basic idea is simple:

> More precise estimates should receive larger weights.

Precision is related to variance.

Smaller variance means greater precision.

Therefore:

$
Weight
=\frac{1}{Variance}
$

The smaller the variance, the larger the weight.

#### Understanding Inverse Variance Weighting

Suppose two SNPs produce causal estimates:

| SNP  | Estimate | Variance |
| ---- | -------- | -------- |
| SNP1 | 0.40     | 0.01     |
| SNP2 | 0.40     | 0.04     |

Weights become:

For SNP1:

$
w_1
=
 \frac{1}{0.01}
=
100
$

For SNP2:

$
w_2
= \frac{1}{0.04}
=
25
$

SNP1 receives four times more weight because it is estimated more precisely.

This principle forms the foundation of IVW Mendelian Randomization.

### The IVW Formula

Suppose each SNP produces a Wald Ratio estimate:

$
\hat{\beta}_1,
\hat{\beta}_2,
\hat{\beta}_3,
...
\hat{\beta}_k
$

with weights:

$
w_1,w_2,w_3,...,w_k
$

The IVW estimate is:

$
\hat{\beta}_{IVW}
=\frac{
\sum_{i=1}^{k}
w_i\hat{\beta}*i
}
{
\sum*{i=1}^{k}
w_i
}
$

This is simply a weighted average of the individual Wald Ratio estimates.

### Interpreting the Formula

The numerator:

$
\sum w_i\hat{\beta}_i
$

represents the weighted sum of SNP-specific causal estimates.

The denominator:

$
\sum w_i
$

represents the total weight.

The resulting estimate gives more influence to precise instruments and less influence to noisy instruments.

### A Numerical Example

Suppose three SNPs produce the following Wald estimates.

| SNP  | Estimate | Variance |
| ---- | -------- | -------- |
| SNP1 | 0.40     | 0.01     |
| SNP2 | 0.50     | 0.02     |
| SNP3 | 0.35     | 0.05     |

First calculate the weights.

SNP1:

$
100
$

SNP2:

$
50
$

SNP3:

$
20
$

Now calculate the weighted sum:

$
100(0.40)
+
50(0.50)
+
20(0.35)
=
40 + 25 + 7
=
72
$

Total weight:

$
100+50+20
=170
$

IVW estimate:

$
\hat{\beta}_{IVW}
=\frac{72}{170}
=
0.424
$

Interpretation:

> A one-unit increase in the exposure is associated with a 0.424-unit increase in the outcome.

assuming the MR assumptions are satisfied.

### An Alternative View of IVW

IVW can also be understood as a weighted regression.

For each SNP we have:

$
\beta_{GX}
$

and

$
\beta_{GY}
$

The IVW method fits a regression of:

$
\beta_{GY}
$

on

$
\beta_{GX}
$

weighted by inverse variance.

The slope of this regression corresponds to the causal estimate.

This interpretation becomes useful when we later discuss MR-Egger regression.

### Relationship Between Wald Ratio and IVW

The Wald Ratio and IVW are closely related.

#### Single SNP

With one SNP:

$
IVW = Wald\ Ratio
$

#### Multiple SNPs

With multiple SNPs:

$
IVW
=Weighted\ Average\ of\ Wald\ Ratios
$

Thus, IVW can be viewed as a natural extension of the Wald Ratio.

### Why IVW Became the Standard Method

IVW became the most widely used MR estimator because it offers several advantages.

#### Higher Statistical Power

Using many SNPs substantially increases power.

#### Greater Precision

Combining information reduces uncertainty.

#### Easy Interpretation

The estimate has a straightforward causal interpretation.

#### Computational Simplicity

IVW is easy to implement and computationally efficient.

### The Assumptions Behind IVW



IVW relies on the same three instrumental variable assumptions introduced in Part 2.

##### Assumption 1: Relevance

Each SNP must be associated with the exposure.

[
SNP
\rightarrow
Exposure
]

##### Assumption 2: Independence

Each SNP must be independent of confounders.

$
SNP
\not\leftrightarrow
Confounders
$

##### Assumption 3: Exclusion Restriction

Each SNP must affect the outcome only through the exposure.

$
SNP
\rightarrow
Exposure
\rightarrow
Outcome
$

and not through alternative pathways.


### The Problem of Horizontal Pleiotropy

IVW performs very well when all instruments are valid.

However, if some SNPs affect the outcome through pathways other than the exposure, bias may occur.

For example:

$
SNP
\rightarrow
Exposure
\rightarrow
Outcome
$

but also:

$
SNP
\rightarrow
Outcome
$

This phenomenon is called horizontal pleiotropy.

Horizontal pleiotropy is one of the major threats to Mendelian Randomization.

Later tutorials will introduce methods specifically designed to address this problem.

#### Fixed Effects and Random Effects IVW

In practice, two versions of IVW are commonly used.

##### Fixed Effects IVW

Assumes all SNPs estimate exactly the same causal effect.

Differences among SNP estimates are attributed solely to sampling variation.

#### Random Effects IVW

Allows some additional heterogeneity among SNP estimates.

Random effects models are often preferred when mild pleiotropy or heterogeneity is suspected.



### Example from Modern Genetic Epidemiology

Suppose researchers want to determine whether BMI causally influences coronary heart disease.

A GWAS identifies hundreds of SNPs associated with BMI.

For each SNP:

1. Obtain SNP → BMI effects.
2. Obtain SNP → CHD effects.
3. Calculate Wald ratios.
4. Combine them using IVW.

The resulting IVW estimate provides an overall causal estimate of BMI on coronary heart disease risk.

This workflow forms the foundation of many modern Mendelian Randomization studies.

#### Strengths of IVW

* High statistical power.
* Increased precision.
* Uses all available instruments.
* Straightforward interpretation.
* Widely accepted in genetic epidemiology.

#### Limitations of IVW

* Sensitive to pleiotropy.
* Requires valid instruments.
* Can be biased when assumptions are violated.
* Does not automatically detect directional pleiotropy.

Because of these limitations, IVW is usually accompanied by additional sensitivity analyses.


### Looking Ahead

The IVW method is often the starting point of a Mendelian Randomization analysis.

However, researchers must also investigate whether the IVW assumptions are reasonable.

One major concern is horizontal pleiotropy.

In the next tutorial, we will introduce:

* Horizontal pleiotropy
* Why it threatens causal inference
* Visualizing pleiotropic pathways
* Detecting violations of MR assumptions

This will prepare us for robust methods such as MR-Egger, Weighted Median, and Weighted Mode estimators.

### Key Takeaways

* Modern Mendelian Randomization studies usually use multiple SNPs.
* Each SNP produces a Wald Ratio estimate.
* IVW combines these estimates using inverse variance weighting.
* More precise instruments receive larger weights.
* IVW can be interpreted as a weighted average of Wald Ratios.
* IVW is the standard MR estimator used in many studies.
* Horizontal pleiotropy remains an important challenge.
* Sensitivity analyses are required to assess robustness.

and

$

for each genetic variant.

This naturally raises an important question:

In modern Mendelian Randomization studies, they often come from separate Genome-Wide Association Studies (GWAS).

This approach is known as **Two-Sample Mendelian Randomization** and has become the standard framework for contemporary MR analyses.

### Revisiting the MR Framework

Recall the Mendelian Randomization framework:

$
SNP
\rightarrow
Exposure
\rightarrow

$

To estimate a causal effect, we need:

1. SNP-exposure associations
2. SNP-outcome associations

The question is:

> Must these associations come from the same individuals?

The answer is:

> No.

This is one of the major strengths of modern Mendelian Randomization.


### One-Sample Mendelian Randomization

Historically, Mendelian Randomization was performed using a single dataset.

Researchers measured:

* Genotypes
* Exposure
* Outcome

for the same individuals.

For example:

| Person | Genotype | BMI | CHD |
| ------ | -------- | --- | --- |
| 1      | AA       | 24  | No  |
| 2      | AG       | 28  | Yes |
| 3      | GG       | 31  | Yes |

Using these data, researchers estimate:

$
SNP \rightarrow BMI
$

and

$
SNP \rightarrow CHD
$

within the same study population.

This approach is called **One-Sample Mendelian Randomization**.

### Advantages of One-Sample MR

#### Complete Data

All variables are measured in the same individuals.

#### Flexible Modeling

Researchers have access to individual-level data.

#### Direct Covariate Adjustment

Additional variables can be included in the analysis.

### Limitations of One-Sample MR

Despite its advantages, one-sample MR has important limitations.

#### Small Sample Sizes

Collecting genotypes, exposures, and outcomes in the same individuals is expensive.

#### Limited Statistical Power

Most individual studies are much smaller than modern GWAS consortia.

#### Data Access Restrictions

Individual-level genetic data are often difficult to obtain.

These limitations motivated the development of two-sample MR.


### Two-Sample Mendelian Randomization

Two-sample MR uses summary statistics from two separate GWAS.

The first GWAS provides:

$
SNP \rightarrow Exposure
$

associations.

The second GWAS provides:

$
SNP \rightarrow Outcome
$

associations.

The participants in these studies do not need to be the same individuals.

### Exposure GWAS

The exposure GWAS identifies SNPs associated with the exposure of interest.

Examples:

* BMI GWAS
* Smoking GWAS
* LDL cholesterol GWAS
* Blood pressure GWAS

Suppose we are interested in BMI.

The exposure GWAS provides:

| SNP | Beta |
| --- | ---- |
| rs1 | 0.12 |
| rs2 | 0.08 |
| rs3 | 0.15 |

These estimates represent:

$$
\beta_{GX}
$$

for each SNP.

### Outcome GWAS

The outcome GWAS measures associations between the same SNPs and the outcome.

For example:

| SNP | Beta |
| --- | ---- |
| rs1 | 0.05 |
| rs2 | 0.03 |
| rs3 | 0.06 |

These estimates represent:

$$
\beta_{GY}
$$

for each SNP.


### Combining the Two GWAS

After obtaining:

$$
\beta_{GX}
$$

from the exposure GWAS

and

$$
\beta_{GY}$$

from the outcome GWAS

we can calculate:

$$
\frac{\beta_{GY}}
{\beta_{GX}}
$$

for each SNP.

These are the individual Wald Ratio estimates.

The estimates are then combined using IVW.

This forms the basis of two-sample Mendelian Randomization.

### Why Two-Sample MR Changed the Field

The development of large GWAS consortia transformed Mendelian Randomization.

Researchers no longer needed access to:

* Individual-level genotype data
* Exposure measurements
* Outcome measurements

in the same participants.

Instead, publicly available GWAS summary statistics could be combined.

This dramatically increased statistical power.

Researchers routinely use these resources for MR studies.

A common question is:

> Can the same individuals appear in both GWAS?

The answer is:

### No Overlap

Exposure GWAS:

100% independent participants.

Outcome GWAS:

Different participants.

This is the ideal situation.

### Partial Overlap

Some participants appear in both studies.

This is common in modern biobanks.

Moderate overlap is often acceptable, especially when instruments are strong.

### Complete Overlap

Both GWAS use exactly the same participants.

In this case, the analysis begins to resemble one-sample MR.

Bias can become more problematic when instruments are weak.

### Weak Instrument Bias in Two-Sample MR

An important property of two-sample MR is that weak instrument bias behaves differently than in one-sample MR.

In one-sample MR:

Weak instruments tend to bias estimates toward the observational association.

In two-sample MR:

Weak instruments generally bias estimates toward the null.

This property makes two-sample MR particularly attractive.

For each SNP we only need:

* Standard error
* Effect allele
* Other allele

This information is routinely reported in GWAS summary statistics.

As a result, many MR studies can be conducted using publicly available data.

### Harmonization of Exposure and Outcome Data

In Part 5, we learned how Two-Sample Mendelian Randomization combines information from two separate GWAS:

1. An exposure GWAS
2. An outcome GWAS

To estimate a causal effect, we need SNP-exposure effects and SNP-outcome effects for the same genetic variants.

However, before we can combine these effects, we must ensure that the effect estimates refer to the same allele.

This process is known as **harmonization**.

Although harmonization may appear to be a simple data-cleaning step, it is one of the most important stages of a Mendelian Randomization analysis. A mistake during harmonization can completely reverse the interpretation of a causal effect.

### Learning Objectives

After completing this tutorial, you should be able to:

### Why Harmonization Is Necessary

Recall that Mendelian Randomization combines:

$
\beta_{GX}
$

and

$
\beta_{GY}
$

for the same SNP.

The Wald Ratio is:

$
\frac{\beta_{GY}}
{\beta_{GX}}
$

This calculation assumes that both effect estimates refer to exactly the same allele.

If they do not, the causal estimate becomes incorrect.

### Understanding Alleles

Every SNP has two alleles.

For example:

| SNP   | Alleles |
| ----- | ------- |
| rs123 | A / G   |
| rs456 | C / T   |
| rs789 | G / T   |

One allele is designated as the effect allele.

The other is designated as the non-effect allele.

### Effect Allele

The effect allele is the allele for which the GWAS effect size is reported.

For example:

| SNP   | Effect Allele | Other Allele | Beta |
| ----- | ------------- | ------------ | ---- |
| rs123 | A             | G            | 0.10 |

Interpretation:

Each additional copy of allele A increases the trait by 0.10.


units.

### Why Allele Alignment Matters

Suppose the exposure GWAS reports:

| SNP   | Effect Allele | Beta |
| ----- | ------------- | ---- |
| rs123 | A             | 0.10 |

The outcome GWAS reports:

| SNP   | Effect Allele | Beta  |
| ----- | ------------- | ----- |
| rs123 | G             | -0.05 |

At first glance, the effect estimates appear different.

However, the two studies are simply using different reference alleles.

The outcome estimate must be converted so that both studies refer to the same allele.

### Example of Correct Harmonization

Exposure GWAS:

| SNP   | EA | OA | Beta |
| ----- | -- | -- | ---- |
| rs123 | A  | G  | 0.10 |

Outcome GWAS:

| SNP   | EA | OA | Beta  |
| ----- | -- | -- | ----- |
| rs123 | G  | A  | -0.05 |

Notice:

Exposure effect allele A.


Outcome effect allele G.


These are opposite alleles.

To harmonize:

* Flip the sign of the outcome beta.
* Align the effect allele.

Result:

| SNP   | EA | Beta |
| ----- | -- | ---- |
| rs123 | A  | 0.05 |

Now both datasets refer to allele A.



### What Happens If We Fail to Harmonize?

Suppose:

Exposure effect +0.10.

Outcome effect  -0.05.


If we incorrectly assume both refer to the same allele:

$
\frac{-0.05}{0.10}
=-0.50
$

This suggests a negative causal effect.

After correct harmonization:

$
\frac{0.05}{0.10}=0.50
$

The direction completely changes.

A simple allele mismatch can reverse the biological interpretation of the study.

### DNA Strand Orientation

DNA consists of two complementary strands.

The nucleotide pairing rules are:

[
A \leftrightarrow T
]

[
C \leftrightarrow G
]

A SNP may therefore be reported on either strand.

For example:

Forward strand A/G


Reverse strand T/C


Both represent the same underlying genetic variant.

This creates additional harmonization challenges.

### Forward and Reverse Strands

Suppose one GWAS reports:

| SNP   | Alleles |
| ----- | ------- |
| rs123 | A/G     |

Another GWAS reports:

| SNP   | Alleles |
| ----- | ------- |
| rs123 | T/C     |

These may actually represent the same SNP.

The difference arises because the studies use opposite DNA strands.

The harmonization process must account for strand orientation.

### Strand Flipping

To align datasets, software may perform strand flipping.

Examples:


A → T



T → A



C → G



G → C


This ensures that exposure and outcome datasets refer to the same biological allele.

### Palindromic SNPs

Some SNPs create a special challenge.

Examples:


A/T


and


C/G


These SNPs are called **palindromic SNPs**.

### Why Are Palindromic SNPs Problematic?

Consider:


A/T


If we switch strands:


T/A


The alleles appear identical.

We cannot immediately determine which allele corresponds to which effect estimate.

This creates ambiguity.



### Example of a Palindromic SNP

Exposure GWAS:

| SNP   | EA | OA |
| ----- | -- | -- |
| rs100 | A  | T  |

Outcome GWAS:

| SNP   | EA | OA |
| ----- | -- | -- |
| rs100 | T  | A  |

Without additional information, it may be impossible to determine whether the alleles are correctly aligned.

### Using Allele Frequencies

Allele frequencies can help resolve ambiguity.

Suppose:

Exposure GWAS:


EAF = 0.10


Outcome GWAS:


EAF = 0.11


The similar frequencies suggest the same allele is being referenced.

In contrast:


0.10


versus


0.90


would indicate a mismatch.

Many MR software packages use allele frequencies during harmonization.

### Ambiguous Palindromic SNPs

Some palindromic SNPs cannot be aligned confidently.

Examples:


EAF = 0.50


Because both alleles occur at similar frequencies, strand orientation cannot be inferred reliably.

These SNPs are often removed from the analysis.

### Harmonization Workflow

A typical harmonization procedure involves:

#### Step 1

Match SNP identifiers.

Example:


rs123


must exist in both datasets.

#### Step 2

Compare effect alleles.

Determine whether the exposure and outcome datasets use the same allele.

#### Step 3

Flip effect sizes if necessary.

Reverse the sign of beta estimates when alleles differ.

#### Step 4

Check strand orientation.

Resolve forward/reverse strand differences.

#### Step 5

Remove ambiguous SNPs.

Exclude SNPs that cannot be aligned confidently.

Several mistakes frequently occur in MR studies.

# Horizontal Pleiotropy: The Biggest Threat to Mendelian Randomization

In the previous tutorials, we introduced Mendelian Randomization (MR) and discussed how genetic variants can be used as instrumental variables to estimate causal effects.

The validity of MR depends on three core assumptions:

1. Relevance
2. Independence
3. Exclusion Restriction

Among these assumptions, the exclusion restriction assumption is often the most difficult to satisfy.

A major reason for this difficulty is **pleiotropy**.

Pleiotropy is one of the most important concepts in statistical genetics and one of the biggest challenges in Mendelian Randomization.

Understanding pleiotropy is essential before learning robust MR methods such as MR-Egger, Weighted Median, and Weighted Mode estimators.

# Learning Objectives

After completing this tutorial, you should be able to:

* Define pleiotropy.
* Distinguish between vertical and horizontal pleiotropy.
* Understand why horizontal pleiotropy threatens MR.
* Identify valid and invalid instruments.
* Understand how pleiotropy creates bias.
* Appreciate why sensitivity analyses are necessary.

# What Is Pleiotropy?

Pleiotropy occurs when a single genetic variant influences multiple traits.

The term comes from Greek:

* "pleio" = many
* "tropy" = effects

In genetics, pleiotropy is extremely common.

Many SNPs affect more than one phenotype.

For example, a single variant may influence:

* BMI
* Type 2 Diabetes
* Blood Pressure
* Cholesterol

simultaneously.

# A Simple Example

Suppose a SNP influences body weight.

At the same time, the same SNP also influences:

* Appetite
* Insulin resistance
* Physical activity

The SNP therefore affects multiple biological processes.

This is an example of pleiotropy.

Importantly:

> Pleiotropy is normal biology.

The challenge is determining whether that pleiotropy violates MR assumptions.

### Revisiting the MR Framework

Recall the ideal MR model:

[
SNP
→
Exposure
→
Outcome
]

Example:

[
SNP
→
BMI
→
Coronary\ Heart\ Disease
]

In this framework:

* The SNP affects BMI.
* BMI affects coronary heart disease.
* There are no alternative pathways.

Under these conditions, MR can estimate a causal effect.

### The Exclusion Restriction Assumption

The exclusion restriction assumption states:

> The genetic variant influences the outcome only through the exposure.

Mathematically:

[
SNP
→
Exposure
→
Outcome
]

and not:

[
SNP
→
Outcome
]

through any other pathway.

This assumption is where pleiotropy becomes important.

### Two Types of Pleiotropy

Not all pleiotropy is problematic.

In Mendelian Randomization we distinguish between:

1. Vertical Pleiotropy
2. Horizontal Pleiotropy

Understanding this distinction is crucial.

### Vertical Pleiotropy

Vertical pleiotropy occurs when the SNP affects downstream traits through the exposure.

Example:

[
SNP
→
BMI
→
Blood\ Pressure
→
Heart\ Disease
]

The SNP affects heart disease through a biological chain of events.

This does not violate the exclusion restriction assumption.

The effect remains mediated by the exposure pathway.

### Why Vertical Pleiotropy Is Acceptable

Suppose:

[
SNP
→
LDL\ Cholesterol
]

and

[
LDL\ Cholesterol
→
Atherosclerosis
]

and

[
Atherosclerosis
→
Coronary\ Heart\ Disease
]

The SNP affects coronary heart disease through downstream consequences of LDL cholesterol.

This is exactly the type of causal pathway that MR is designed to capture.

Therefore:

> Vertical pleiotropy is generally not considered a problem for Mendelian Randomization.

#### Horizontal Pleiotropy

Horizontal pleiotropy occurs when a SNP influences the outcome through a pathway other than the exposure.

Example:


SNP
→
BMI


and simultaneously:


SNP
→
Blood\ Pressure


Suppose we are studying:


BMI
→Blood\ Pressure


The direct SNP effect on blood pressure creates an alternative pathway.

This violates the exclusion restriction assumption.

### Visualizing Horizontal Pleiotropy

Desired pathway:


SNP
→
Exposure
→
Outcome


Problematic pathway:


SNP
→
Outcome


The outcome is now influenced by two routes:

1. Through the exposure.
2. Through an independent pathway.

This makes it impossible to determine how much of the SNP-outcome association is truly caused by the exposure.

### Why Horizontal Pleiotropy Is Dangerous

The SNP-outcome effect now contains two components.

Exposure-mediated effect:


SNP
→
Exposure
→
Outcome


plus

Direct pleiotropic effect:


SNP
→
Outcome


The observed SNP-outcome association becomes:

$
Observed\ Effect
=Causal\ Effect
+
Pleiotropic\ Effect
$

As a result, the MR estimate becomes biased.

### A Numerical Example

Suppose:

[
SNP
\rightarrow
BMI
===

0.20
]

The true causal effect is:

[
BMI
\rightarrow
CHD
===

0.30
]

Expected SNP-outcome effect:

$
0.20 \times 0.30
=
0.06
$

Now imagine the SNP also directly affects CHD:

$
SNP
\rightarrow
CHD
=

0.04
$

Observed SNP-outcome effect becomes:

$
0.06 + 0.04
=0.10
$

Wald Ratio:

$
\frac{0.10}{0.20}
=0.50
$

The true causal effect was:

$
0.30
$

but the MR estimate becomes:

$
0.50
$

because of horizontal pleiotropy.

# Balanced vs Directional Pleiotropy

Not all pleiotropic effects behave the same way.

Two important patterns are:

## Balanced Pleiotropy

Some SNPs have positive pleiotropic effects.

Some SNPs have negative pleiotropic effects.

These effects may cancel out on average.

Example:

| SNP  | Pleiotropic Effect |
| ---- | ------------------ |
| SNP1 | +0.03              |
| SNP2 | -0.04              |
| SNP3 | +0.01              |
| SNP4 | -0.02              |

Average pleiotropic effect:

Approximately zero.

IVW estimates may remain relatively unbiased.

# Directional Pleiotropy

Directional pleiotropy occurs when pleiotropic effects consistently point in the same direction.

Example:

| SNP  | Pleiotropic Effect |
| ---- | ------------------ |
| SNP1 | +0.03              |
| SNP2 | +0.04              |

Positive.

### MR-Egger Regression: Detecting and Adjusting for Directional Pleiotropy

In Part 7, we learned that horizontal pleiotropy is one of the biggest threats to Mendelian Randomization (MR).

The Inverse Variance Weighted (IVW) method assumes that all genetic instruments are valid. However, if some SNPs influence the outcome through pathways other than the exposure, the IVW estimate may become biased.

This raises an important question:

> Can we detect whether pleiotropy is affecting our Mendelian Randomization analysis?

MR-Egger Regression was developed to address this problem.

Unlike IVW, MR-Egger allows genetic variants to have direct effects on the outcome and provides a statistical framework for detecting directional pleiotropy.

### Learning Objectives

After completing this tutorial, you should be able to:

* Understand the motivation behind MR-Egger Regression.
* Compare IVW and MR-Egger.
* Understand the MR-Egger regression model.
* Interpret the intercept and slope.
* Understand directional pleiotropy.
* Understand the InSIDE assumption.
* Recognize the strengths and limitations of MR-Egger.


### Why IVW Can Fail

Recall the IVW framework.

For each SNP we have:

$
\beta_{GX}
$

and

$
\beta_{GY}
$

The IVW method essentially fits a weighted regression through the origin:

$
\beta_{GY}
=\beta_{MR}\beta_{GX}
$

Notice:

The intercept is fixed at zero.

This means IVW assumes:

> There is no average pleiotropic effect across SNPs.

If this assumption is violated, the causal estimate may be biased.


### The Idea Behind MR-Egger

MR-Egger extends the IVW regression model by allowing a non-zero intercept.

Instead of:

$
\beta_{GY}
=\beta_{MR}\beta_{GX}
$

MR-Egger fits:

$
\beta_{GY}
=\alpha
+
\beta_{MR}\beta_{GX}
+
\varepsilon
$

where:

* (\alpha) is the intercept.
* (\beta_{MR}) is the causal estimate.
* (\varepsilon) is random error.

The key innovation is the intercept term.

### Why the Intercept Matters

The intercept represents the average pleiotropic effect across all instruments.

If:

$
\alpha = 0
$

there is no evidence of directional pleiotropy.

If:

$
\alpha \neq 0
$

there is evidence that SNPs are affecting the outcome through pathways other than the exposure.

Thus, the intercept becomes a diagnostic tool.



### Visualizing IVW and MR-Egger

Imagine plotting:

$
\beta_{GX}
$

on the x-axis

and

$
\beta_{GY}
$

on the y-axis.

Each SNP becomes a point on the graph.

#### IVW

IVW forces the regression line through:

$
(0,0)
$

The intercept is fixed at zero.

#### MR-Egger

MR-Egger allows the regression line to cross the y-axis at:

$
\alpha
$

This additional flexibility helps account for directional pleiotropy.

### The MR-Egger Slope

The slope of the MR-Egger regression:

[
\beta_{MR}
]

represents the estimated causal effect.

Interpretation is similar to IVW.

For example:

$
\beta_{MR}
=0.40
$

suggests:

> A one-unit increase in the exposure causes a 0.40-unit increase in the outcome.

assuming the MR-Egger assumptions hold.



### The MR-Egger Intercept

The intercept is often more interesting than the slope.

Suppose:

$
\alpha = 0.03
$

This suggests that, on average, SNPs have positive direct effects on the outcome that are not mediated through the exposure.

This provides evidence for directional pleiotropy.

Researchers often perform a hypothesis test:

$
H_0 : \alpha = 0
$

versus

$
H_A : \alpha \neq 0
$

A significant result suggests pleiotropic bias.

# Balanced vs Directional Pleiotropy Revisited

Recall from Part 7:

## Balanced Pleiotropy

Positive and negative pleiotropic effects cancel each other.

Average pleiotropic effect:

$
0
$

MR-Egger intercept:

Approximately zero.


## Directional Pleiotropy

Most pleiotropic effects point in the same direction.

Average pleiotropic effect:

Non-zero.

MR-Egger intercept:

Significantly different from zero.

This is exactly what MR-Egger is designed to detect.

### A Simple Example

Suppose five SNPs are associated with BMI.

If all SNPs influence coronary heart disease only through BMI, then:

$
\alpha \approx 0
$

The IVW and MR-Egger estimates should be very similar.

Now suppose many SNPs also directly influence coronary heart disease.

Then:

$
\alpha > 0
$

The MR-Egger intercept test may detect this bias.




## The InSIDE Assumption

MR-Egger introduces a new assumption.

This assumption is called:

**Instrument Strength Independent of Direct Effect**

or simply:

**InSIDE**.

### What Does InSIDE Mean?

The InSIDE assumption states:

> The strength of the SNP-exposure association is independent of the direct pleiotropic effect.

In other words:

Strong instruments should not systematically have larger pleiotropic effects than weak instruments.

Mathematically, the direct effects and instrument strengths should be uncorrelated.

#### Why Is InSIDE Important?

The MR-Egger slope can remain unbiased even when every SNP is pleiotropic.

This sounds remarkable.

However, it is only true if the InSIDE assumption holds.

If InSIDE is violated:

* The intercept may be misleading.
* The slope may become biased.

Unfortunately, InSIDE cannot usually be tested directly.

# Comparing IVW and MR-Egger

| Feature                        | IVW | MR-Egger |
| ------------------------------ | --- | -------- |
| Allows non-zero intercept      | No  | Yes      |
| Detects directional pleiotropy | No  | Yes      |
| Higher statistical power       | Yes | No       |
| More robust to pleiotropy      | No  | Yes      |
| Standard MR estimator          | Yes | No       |

IVW is usually the primary analysis.

MR-Egger is often used as a sensitivity analysis.

### Why MR-Egger Has Lower Power

MR-Egger estimates an additional parameter:

$
\alpha
$

Estimating the intercept introduces uncertainty.

As a result:

* Standard errors increase.
* Confidence intervals widen.
* Statistical power decreases.

This is the price paid for increased robustness.

MR-Egger is particularly useful when:

* Many instruments are available.
* Pleiotropy is suspected.
* Researchers want a sensitivity analysis.
* IVW assumptions may be questionable.

It has become one of the standard robustness checks in Mendelian Randomization.

MR-Egger was developed because researchers recognized that pleiotropy is widespread in genetics.

Rather than assuming pleiotropy does not exist, MR-Egger attempts to detect and adjust for it.

This makes MR-Egger one of the most important developments in modern Mendelian Randomization methodology.

Although MR-Egger is robust to some forms of pleiotropy, it often has low statistical power.

# Weighted Median Mendelian Randomization

In Part 8, we introduced MR-Egger Regression, a method that can detect and partially account for directional pleiotropy.

MR-Egger was an important advance because it relaxed some of the assumptions required by the standard Inverse Variance Weighted (IVW) estimator.

However, MR-Egger has limitations:

* Lower statistical power
* Wider confidence intervals
* Dependence on the InSIDE assumption
* Sensitivity to outlying SNPs

Researchers therefore sought alternative approaches that would remain robust when some genetic instruments are invalid while still retaining reasonable statistical power.

One of the most successful methods is the **Weighted Median Estimator**.

The weighted median approach has become one of the most widely used sensitivity analyses in Mendelian Randomization.

# Why Do We Need Another Method?

Recall the assumptions behind IVW.

IVW performs well when all genetic instruments are valid.

However, suppose we have 100 SNPs.

What happens if:

* 10 SNPs are invalid?
* 20 SNPs are invalid?
* 30 SNPs are invalid?

IVW may become biased because all SNPs contribute to the final estimate.

Even a relatively small number of problematic instruments can distort the result.

Researchers therefore wanted a method that could tolerate some invalid instruments.

# The Core Idea

The weighted median estimator is based on a simple observation:

> If most instruments are valid, the middle estimate should still be reliable.

Rather than averaging SNP estimates, the weighted median focuses on the center of the distribution.

This makes it less sensitive to extreme values and invalid instruments.

# Revisiting Wald Ratios

Suppose we have five SNPs.

Each produces a Wald Ratio estimate.

| SNP  | Wald Ratio |
| ---- | ---------- |
| SNP1 | 0.30       |
| SNP2 | 0.35       |
| SNP3 | 0.40       |
| SNP4 | 0.42       |
| SNP5 | 2.50       |

Notice SNP5.

Its estimate is dramatically larger than the others.

Possible explanations include:

* Horizontal pleiotropy
* Data errors
* Weak instruments
* Invalid instrument assumptions

IVW will be influenced by this outlier.

The weighted median is much less affected.

# What Is a Median?

A median is the middle value in an ordered list.

Example:

[
1,;2,;3,;4,;5
]

Median:

[
3
]

Unlike a mean, the median is resistant to extreme values.

Example:

[
1,;2,;3,;4,;100
]

Mean:

[
22
]

Median:

[
3
]

The median remains stable despite the outlier.

### Extending the Idea to Mendelian Randomization

In MR we do not simply calculate the median of SNP estimates.

Each SNP has a different level of precision.

More precise SNPs should contribute more information.

Therefore, we use a **weighted median**.

### The Weighted Median Concept

Suppose each SNP has:

* A causal estimate
* A statistical weight

The weighted median is defined as:

> The estimate where 50% of the total weight lies on either side.

The key idea is that SNPs with higher precision contribute more heavily.

#### A Simple Example

Suppose we have:

| SNP  | Estimate | Weight |
| ---- | -------- | ------ |
| SNP1 | 0.20     | 10     |
| SNP2 | 0.25     | 20     |
| SNP3 | 0.30     | 30     |
| SNP4 | 0.35     | 25     |
| SNP5 | 1.50     | 15     |

Total weight:


10+20+30+25+15
=100


The weighted median corresponds to the point where cumulative weight reaches:50


The estimate associated with that point becomes the weighted median estimate.

Notice that SNP5 has an extreme value.

Because it carries only 15% of the total weight, it has limited influence on the final estimate.

### The 50% Rule

The weighted median estimator has a remarkable property.

It produces a consistent causal estimate provided that:

> More than 50% of the total weight comes from valid instruments.

This is often called the:

**50% Valid Instrument Assumption**

### Why Is the 50% Rule Important?

Imagine:

* 40% of SNP weight comes from invalid instruments.
* 60% of SNP weight comes from valid instruments.

The weighted median can still recover the correct causal effect.

This is substantially more robust than IVW.

### Comparing IVW and Weighted Median

Suppose:

90 SNPs estimate:0.30


10 pleiotropic SNPs estimate:2.00


IVW will be pulled upward because all SNPs contribute to the average.

The weighted median remains close to:0.30


provided the majority of weight comes from valid instruments.

### Why Weighted Median Is Robust

The method does not rely on every instrument being valid.

Instead, it assumes:

> Most of the information comes from valid instruments.

This is often a much more realistic assumption in complex biological systems.

### Weighted Median and Outliers

The weighted median is particularly useful when a small number of SNPs produce extreme estimates.

Consider:

| SNP  | Estimate |
| ---- | -------- |

| SNP2 | 0.31     |

The weighted median remains much closer to:0.30

### When Does Weighted Median Fail?

The weighted median is not perfect.

Problems occur when:

* More than 50% of the weight comes from invalid instruments.
* Most SNPs share similar pleiotropic effects.
* Strong directional pleiotropy affects the majority of instruments.

In these situations, the weighted median may also become biased.

The weighted median method gained popularity because it occupies an attractive middle ground.

Compared with IVW:

* More robust.

* More precise.
* Greater statistical power.

Consistency strengthens confidence in causal conclusions.

### The Big Picture

The weighted median estimator was developed because real genetic datasets often contain some invalid instruments.

Rather than requiring perfection, the method asks:

> Does most of the information come from valid instruments?

If the answer is yes, the weighted median can provide reliable causal estimates even when some SNPs are problematic.

This robustness has made it one of the most important tools in modern Mendelian Randomization.

### Weighted Mode Mendelian Randomization

In Part 9, we introduced the Weighted Median estimator, a robust Mendelian Randomization (MR) method that can provide reliable causal estimates when more than 50% of the total instrument weight comes from valid instruments.

The weighted median estimator represented an important improvement over the standard Inverse Variance Weighted (IVW) approach because it can tolerate some invalid instruments.

However, researchers asked an important question:

> What if fewer than 50% of the instruments are valid?

In some situations, the majority of SNPs may not be valid instruments, yet there may still be a cluster of SNPs that identify the true causal effect.

This idea motivated the development of **Mode-Based Mendelian Randomization**, commonly known as the **Weighted Mode Estimator**.

### Learning Objectives

After completing this tutorial, you should be able to:

* Understand the motivation behind the weighted mode estimator.
* Define a modal causal estimate.
* Understand clustering of SNP-specific causal estimates.

### Why Another Robust Method?

Recall the assumptions behind previous methods.

#### IVW

Requires all instruments to be valid.

#### MR-Egger

Allows pleiotropy but relies on the InSIDE assumption.

#### Weighted Median

Requires more than 50% of total weight to come from valid instruments.

But what happens if:

* Only 30% of SNPs are valid?
* 40% are valid?
* The valid instruments do not constitute a majority?

Can we still recover the correct causal effect?

The weighted mode estimator attempts to answer yes.

### The Core Idea

The weighted mode estimator is based on a simple intuition:

> Valid instruments should estimate approximately the same causal effect.

If multiple SNPs are valid instruments, their Wald Ratio estimates should cluster around the true causal effect.

Invalid instruments may produce many different estimates.

Therefore:

> The largest cluster of SNP estimates is likely to represent the true causal effect.

This cluster defines the mode.

### What Is a Mode?

In statistics:

The mode is the most frequently occurring value.

Example:
1,;2,;2,;2,;3,;4


Mode:


2


because it appears most often.

The weighted mode estimator extends this concept to Mendelian Randomization.

### From Wald Ratios to Clusters

Suppose we calculate SNP-specific causal estimates.

| SNP  | Wald Ratio |
| ---- | ---------- |
| SNP1 | 0.30       |
| SNP2 | 0.31       |
| SNP3 | 0.29       |
| SNP4 | 0.32       |
| SNP5 | 1.20       |
| SNP6 | 1.35       |
| SNP7 | -0.40      |

Notice the pattern.

Cluster 1:

0.29,;0.30,;0.31,;0.32


Cluster 2:


1.20,;1.35


Cluster 3:


-0.40


The largest cluster is centered around:


0.30


The weighted mode estimator identifies this cluster as the most likely causal effect.

### Why Valid Instruments Cluster

Suppose the true causal effect is:


0.30


Every valid SNP estimates:


0.30


plus small sampling error.

Therefore valid instruments tend to produce similar estimates.

Invalid instruments affected by pleiotropy may generate very different values.

This creates clusters.

The largest cluster often corresponds to the valid instruments.

### Visualizing the Idea

Imagine plotting all SNP-specific causal estimates.

You might observe:

```text
Density

          *
         ***
        *****
       *******
      *********
------|---------|---------|-----
     0.3       1.2      -0.4
```

The highest peak represents the modal estimate.

The weighted mode estimator identifies this peak and uses it as the causal estimate.

### Weighted Mode Versus Simple Mode

Not all SNPs are equally informative.

Some SNPs have:

* Smaller standard errors
* Larger sample sizes
* More precise estimates

Therefore we use a weighted mode rather than a simple mode.

More precise SNPs contribute more heavily to the identification of the modal cluster.

### The Plurality Valid Assumption

The weighted mode estimator relies on a different assumption than the weighted median estimator.

Weighted Median:

> More than 50% of total weight must come from valid instruments.

Weighted Mode:

> The largest cluster of SNPs must consist of valid instruments.

This is known as the:

**Plurality Valid Assumption**

### Majority Versus Plurality

These terms are often confused.

#### Majority

More than 50%.

Example:

| Group        | Percentage |
| ------------ | ---------- |
| Valid SNPs   | 60%        |
| Invalid SNPs | 40%        |

Valid instruments form a majority.

#### Plurality

Largest group, but not necessarily more than 50%.

Example:

| Group          | Percentage |
| -------------- | ---------- |
| Valid SNPs     | 40%        |
| Invalid Type A | 35%        |
| Invalid Type B | 25%        |

Valid instruments do not form a majority.

However, they form the largest cluster.

The weighted mode estimator can still work.

### Why This Is Powerful

The weighted mode estimator can succeed in situations where:

* IVW fails
* Weighted Median fails

because it requires only the largest cluster to be valid.

This can be a much weaker and more realistic assumption.

Suppose we obtain:

| SNP2 | 0.31     |
| SNP3 | 0.30     |
| SNP4 | 0.32     |
| SNP5 | 1.50     |
| SNP6 | 1.40     |
| SNP7 | -0.60    |

Three distinct clusters emerge.

Cluster A:

[

]

[
1.40,;1.50
]

[
-0.60

| Weighted Median | 0.43     |
| Weighted Mode   | 0.42     |

All methods agree closely.

This provides strong support for a stable causal estimate.

### Why Weighted Mode Is Useful

The weighted mode estimator is particularly attractive because:

* It tolerates invalid instruments.
* It does not require the InSIDE assumption.
* It does not require a majority of valid SNPs.
* It can remain informative when other methods struggle.

### Limitations of Weighted Mode

Despite its advantages, the weighted mode estimator has limitations.

#### Sensitive to Cluster Identification

Different clustering choices may produce different results.

#### Lower Precision

Mode estimates often have larger standard errors than IVW.

#### Less Familiar

Many researchers are more comfortable interpreting IVW and MR-Egger results.

#### Requires Distinct Clustering

If SNP estimates form a broad continuous distribution rather than clear clusters, the method may perform poorly.

### Why Weighted Mode Is Usually a Sensitivity Analysis

Most MR studies use the weighted mode as part of a broader sensitivity analysis framework.

Researchers typically compare:

1. IVW
2. MR-Egger
3. Weighted Median
4. Weighted Mode

Agreement across methods increases confidence that the observed causal effect is not driven by violations of MR assumptions.

### The Robust MR Toolbox

At this stage, we have introduced four major Mendelian Randomization estimators:

#### IVW

Primary analysis.

#### MR-Egger

Detects directional pleiotropy.

#### Weighted Median

Robust if more than 50% of weight is valid.

#### Weighted Mode

Robust if the largest cluster is valid.

Together these methods form the foundation of modern MR sensitivity analyses.

### The Big Picture

The weighted mode estimator reflects an important principle in Mendelian Randomization:

> Even when some instruments are invalid, the true causal signal often emerges as the dominant pattern among the valid instruments.

Rather than averaging all SNPs, the weighted mode focuses on the most consistent cluster of evidence.

This makes it one of the most robust approaches available for causal inference using genetic instruments.

### Sensitivity Analyses in Mendelian Randomization

In Parts 4–10, we introduced the major Mendelian Randomization (MR) estimators:

* Inverse Variance Weighted (IVW)
* MR-Egger
* Weighted Median
* Weighted Mode

Each method attempts to estimate the causal effect of an exposure on an outcome, but each relies on different assumptions.

A natural question arises:

> Which estimate should we trust?

In practice, Mendelian Randomization rarely relies on a single estimator. Instead, researchers perform a series of sensitivity analyses to evaluate whether their conclusions are robust to violations of MR assumptions.

Sensitivity analyses are among the most important components of a modern MR study.

### Learning Objectives

After completing this tutorial, you should be able to:

### Why Sensitivity Analyses Are Necessary

Recall the three core MR assumptions:

1. Relevance
2. Independence
3. Exclusion Restriction

Unfortunately:

* We cannot directly verify all assumptions.
* Some SNPs may be invalid instruments.
* Horizontal pleiotropy may be present.
* Measurement errors may occur.

As a result:

> A single MR estimate is rarely sufficient.

Sensitivity analyses help determine whether conclusions remain stable under different assumptions.

#### The Goal of Sensitivity Analyses

The purpose is not necessarily to prove that an MR estimate is correct.

Instead, sensitivity analyses ask:

> How sensitive is the result to potential violations of MR assumptions?

If multiple methods produce similar conclusions, confidence increases.

If methods disagree substantially, caution is warranted.

### The Four Core MR Estimators

Most MR studies report:

#### IVW

Primary analysis.

#### MR-Egger

Tests for directional pleiotropy.

#### Weighted Median

Robust when more than 50% of weight comes from valid instruments.

#### Weighted Mode

Robust when the largest cluster of SNPs is valid.

Together these provide complementary perspectives on the causal effect.


### Scenario 1: Strong Agreement

Suppose the results are:

Interpretation:

* All methods point in the same direction.
* Effect sizes are similar.

Suppose:

| --------------- | -------- |

Interpretation:

* Large disagreement among methods.

#### Heterogeneity in Mendelian Randomization

Suppose every SNP is a valid instrument.

Then each SNP should estimate approximately the same causal effect.

Example:

| SNP  | Wald Ratio |
| ---- | ---------- |
| SNP1 | 0.31       |
| SNP2 | 0.29       |
| SNP3 | 0.33       |
| SNP4 | 0.30       |

These estimates are highly consistent.

Now consider:

| SNP  | Wald Ratio |
| ---- | ---------- |
| SNP1 | 0.31       |
| SNP2 | 0.29       |
| SNP3 | 1.75       |
| SNP4 | -0.50      |

These estimates differ dramatically.

This variability is called heterogeneity.

### Why Heterogeneity Matters

Large heterogeneity may indicate:

* Horizontal pleiotropy
* Invalid instruments
* Data quality problems
* Population differences
* Outlying SNPs

Heterogeneity is therefore an important warning signal.

### Cochran's Q Statistic

The most common heterogeneity test in MR is Cochran's Q statistic.

The idea is simple:

> Do SNP-specific causal estimates vary more than expected by chance?

Null hypothesis:


All\ SNPs\ estimate\ the\ same\ causal\ effect


Alternative hypothesis:

At\ least\ some\ SNPs\ differ


### Interpreting Cochran's Q

Suppose:

Q-test p-value: p = 0.75


Interpretation:

Little evidence of heterogeneity.

Now suppose:


p < 0.001


Interpretation:

Strong evidence of heterogeneity.

Researchers should investigate possible pleiotropy.

#$$ Heterogeneity Does Not Automatically Mean Failure

An important point:

> Heterogeneity is a warning sign, not proof of invalidity.

Possible explanations include:

* True biological complexity
* Measurement error
* Pleiotropy
* Population structure

Heterogeneity should trigger further investigation rather than immediate rejection of the analysis.



### The MR-Egger Intercept Test

Recall from Part 8:

MR-Egger estimates:

$
\beta_{GY}
=\alpha
+
\beta_{MR}\beta_{GX}
+
\varepsilon
$

The intercept:

$
\alpha
$

measures average directional pleiotropy.

### Interpreting the Intercept Test

Null hypothesis:

$
\alpha = 0
$

No directional pleiotropy.

Alternative hypothesis:

[
\alpha \neq 0
]

Directional pleiotropy present.

Example:

| Intercept | p-value |
| --------- | ------- |
| 0.001     | 0.72    |
| 0.035     | 0.002   |

The second result suggests pleiotropic bias.

### Leave-One-Out Analysis

Another popular sensitivity analysis is the leave-one-out procedure.

The idea is straightforward.

Suppose we have 20 SNPs.

Run IVW:

Using all 20 SNPs.

Then:

Remove SNP1.

Run IVW again.

Remove SNP2.

Run IVW again.

Continue until every SNP has been removed once.

### Why Leave-One-Out Is Useful

This procedure identifies influential SNPs.

Suppose:

Full analysis:

$
0.42
$

After removing SNP7:

$
0.05
$

This suggests SNP7 is driving the result.

Researchers should inspect that SNP carefully.

### Interpreting Leave-One-Out Results

Ideal pattern:

All leave-one-out estimates are similar.

Problematic pattern:

One or two SNPs dramatically alter the estimate.

This may indicate:

* Horizontal pleiotropy
* Data problems
* Invalid instruments
