Over the last two decades, Genome-Wide Association Studies (GWAS) have revolutionized our understanding of the genetic architecture of complex traits and diseases. Researchers have identified thousands of genetic variants associated with conditions such as schizophrenia, coronary artery disease, type 2 diabetes, obesity, depression, and many other common disorders.

Unlike Mendelian diseases, which are often caused by mutations in a single gene, most common diseases are influenced by a very large number of genetic variants, each contributing only a tiny effect. This phenomenon is known as **polygenicity**.

The realization that thousands of variants collectively influence disease risk led to the development of one of the most important tools in modern statistical genetics:

**Polygenic Risk Scores (PRS)**.

A Polygenic Risk Score combines information across many genetic variants into a single numerical value that represents an individual's inherited genetic predisposition to a trait or disease.

In this tutorial we will learn:

* What a Polygenic Risk Score is
* Why PRS works
* How PRS is calculated
* How GWAS provides the building blocks of PRS
* Why PRS can be useful despite tiny SNP effects
* Applications and limitations of PRS

### Why Do We Need Polygenic Risk Scores?

Historically, genetic studies attempted to identify single genes responsible for disease.
This strategy works remarkably well for Mendelian disorders such as:

* Huntington's disease
* Cystic fibrosis
* Sickle cell anemia

where mutations in a single gene can have a large biological effect.
However, common diseases behave very differently.

For example:

| Trait                   | Genetic Architecture  |
| ----------------------- | --------------------- |
| Height                  | Thousands of variants |
| BMI                     | Thousands of variants |
| Coronary Artery Disease | Thousands of variants |
| Schizophrenia           | Thousands of variants |
| Depression              | Thousands of variants |

No individual SNP is sufficiently informative to predict disease risk on its own.
Instead:  Risk = \text{Many Small Genetic Effects}.

The challenge becomes:
> How can we combine thousands of tiny genetic effects into a meaningful predictor?

Polygenic Risk Scores provide the answer.

### What Is a Polygenic Risk Score?

A Polygenic Risk Score (PRS), also known as a Polygenic Score (PGS), is a weighted sum of genetic variants across the genome.

Mathematically:$PRS_i = \sum_{j=1}^{m}\hat{\beta}*j x*{ij}$

where:

* $i$ represents an individual
* $j$ represents a SNP
* $m$ is the number of SNPs
* $x_{ij}$ is the genotype count
* $\hat{\beta}_j$ is the estimated SNP effect size

The score summarizes the combined contribution of many genetic variants to a phenotype.

#### Understanding the Components

##### Genotype Count

For a given SNP, an individual can carry zero, one, or two copies of the risk allele.

| Genotype | Risk Allele Count |
| -------- | ----------------- |
| AA       | 0                 |
| AG       | 1                 |
| GG       | 2                 |

Thus $x_{ij} \in {0,1,2}$.

The genotype count measures the number of risk alleles carried by the individual.

#### SNP Effect Size

The quantity $\hat{\beta}_j$ represents the estimated effect of SNP $j$ on the phenotype.

These effect sizes are usually obtained from a Genome-Wide Association Study (GWAS).

For example:

| SNP    | Effect Size |
| ------ | ----------- |
| rs1001 | 0.05        |
| rs1002 | 0.03        |
| rs1003 | -0.02       |

Positive values increase risk while negative values decrease risk.

#### A Simple Example

Suppose a GWAS produces the following SNP effects:

| SNP | Effect Size |
| --- | ----------- |
| rs1 | 0.05        |
| rs2 | 0.02        |
| rs3 | -0.03       |

A new individual has the following genotypes:

| SNP | Genotype Count |
| --- | -------------- |
| rs1 | 2              |
| rs2 | 1              |
| rs3 | 0              |

The Polygenic Risk Score becomes:

$PRS = (0.05)(2)+(0.02)(1)+(-0.03)(0)= 0.10 + 0.02 + 0 = 0.12$

This value represents the individual's overall genetic burden based on the selected SNPs.

### Why Does PRS Work?

At first glance, PRS may seem surprising. Many SNPs used in prediction are not actually causal. Yet they still improve prediction accuracy.

The reason is a phenomenon called **Linkage Disequilibrium (LD)**.

Linkage Disequilibrium describes the tendency of nearby SNPs to be inherited together.

Suppose:

```text
Causal SNP → Disease
```

We may not directly observe the causal SNP.

However:

```text
Observed SNP → Correlated with → Causal SNP
```

Because the observed SNP is correlated with the causal SNP, it can still carry useful predictive information.
This principle is fundamental to both GWAS and Polygenic Risk Scores.

### PRS as a Genetic Credit Score

A useful analogy is a financial credit score. Banks do not determine creditworthiness using a single transaction. Instead they combine information from:

* Income
* Existing loans
* Repayment history
* Credit utilization
* Financial behavior

Similarly, PRS combines information from thousands of SNPs across the genome.
Each SNP contributes only a tiny amount. Collectively, however, they can generate a meaningful predictor of genetic predisposition.

#### From GWAS to PRS

The construction of a PRS typically involves three datasets.

### Discovery Dataset

The discovery dataset is used to perform a GWAS.

The objective is to estimate SNP effect sizes $\hat{\beta}_1,\hat{\beta}_2,\ldots,\hat{\beta}_m$

These estimated effects become the weights used in PRS construction.

#### Target Dataset

The target dataset contains individuals whose genetic risk we wish to predict.

For each individual we require:

* Genotypes
* Phenotypes (for evaluation)

#### PRS Calculation

The GWAS effect sizes are applied to the target genotypes:

$PRS_i = \sum_j \hat{\beta}*j x*{ij} $

One PRS is generated for every individual.

### What Does a PRS Mean?

A Polygenic Risk Score is not:

* A diagnosis
* A certainty of disease
* A probability of disease

Instead, it represents:

> Relative genetic predisposition compared with other individuals.

For example:

| Percentile | Interpretation             |
| ---------- | -------------------------- |
| Bottom 10% | Lower genetic risk         |
| Middle 50% | Average genetic risk       |
| Top 10%    | Elevated genetic risk      |
| Top 1%     | Very elevated genetic risk |

Environmental factors continue to play a major role in determining actual disease outcomes.


#### Applications of Polygenic Risk Scores

Polygenic Risk Scores have become an important tool in biomedical research and are increasingly being used to study a wide range of complex diseases and traits. In cardiovascular medicine, PRS can help identify individuals with an elevated inherited risk of conditions such as coronary artery disease, potentially enabling earlier preventive interventions. In psychiatry, polygenic prediction is used to investigate genetic susceptibility to disorders including schizophrenia, bipolar disorder, and depression, helping researchers better understand the genetic architecture of mental health conditions. PRS also has important applications in metabolic disorders, where it can be used to assess genetic predisposition to obesity and type 2 diabetes. Beyond disease risk prediction, polygenic scores are widely applied to quantitative traits such as height, body mass index (BMI), and blood pressure, where they help explain individual differences in phenotype and contribute to our understanding of the genetic basis of human variation.


#### Limitations of Polygenic Risk Scores

Despite their utility, PRS are not perfect predictors.

Prediction accuracy is fundamentally limited by:

1. Trait heritability
2. SNP coverage
3. Discovery GWAS sample size
4. Statistical methodology

Even a theoretically perfect PRS cannot explain all variation in a phenotype.
Environmental influences remain critically important.

Examples include:

* Diet
* Exercise
* Smoking
* Stress
* Socioeconomic factors
* Random biological variation

Thus:

$Phenotype = Genetics+Environment+Random Effects$

### Key Takeaways

* Most common diseases are highly polygenic.
* Polygenic Risk Scores combine information across many SNPs.
* PRS is calculated as a weighted sum of risk alleles.
* GWAS provides the SNP effect sizes used as weights.
* Linkage Disequilibrium allows correlated SNPs to contribute predictive information.
* PRS measures relative genetic predisposition rather than certainty.
* Prediction accuracy is limited by heritability, sample size, SNP coverage, and methodology.


### Evaluating Polygenic Risk Scores

An important question is:

> How do we determine whether a PRS is actually useful?

A score that predicts disease poorly has little practical value, regardless of how sophisticated the methodology used to construct it may be. Now, we will explore the most commonly used evaluation metrics in polygenic prediction and learn how researchers assess the performance of PRS models.

### Why PRS Evaluation Matters

Suppose we build a PRS for schizophrenia.

After calculating scores for thousands of individuals, we obtain values such as:

| Individual | PRS   |
| ---------- | ----- |
| Person A   | 0.52  |
| Person B   | -0.13 |
| Person C   | 1.21  |

These numbers by themselves are meaningless.

The important question is:

> Do individuals with higher PRS actually have a higher probability of developing schizophrenia?

Evaluating prediction performance allows us to determine:

* Whether the PRS contains useful information
* How much variation in the phenotype can be explained
* Whether the PRS improves prediction beyond traditional risk factors
* Whether a more sophisticated method actually performs better

### Evaluating Quantitative Traits

For quantitative traits such as:

* Height
* BMI
* Blood pressure
* Cholesterol levels

the most common evaluation metric is $R^2$, which measures the proportion of phenotypic variance explained by the model.

The ISG workshop identifies prediction $R^2$ as the primary metric for quantitative traits.

#### Understanding Variance Explained

Consider human height.Suppose height varies substantially within a population. A useful PRS should explain some of this variation.

For example:

```text
Observed Height Variation
│
├── Genetics
├── Environment
└── Random Effects
```

If the PRS captures part of the genetic component, it should reduce unexplained variation.

The proportion of variance explained is quantified by $R^2$.

### The Null Model

Before including PRS, researchers usually fit a model containing only covariates.Typical covariates include:

* Sex
* Age
* Population principal components (PCs)
* Recruitment center
* Batch effects

The null model is: $y = Covariates + e$,

where:

* $y$ is the phenotype
* $e$ is residual error

### The Full Model

Next, we include the PRS.

The full model becomes $y = Covariates + PRS + e$.

If PRS contains predictive information, the model should explain more variance.

### Incremental R²

A crucial concept in polygenic prediction is **Incremental R²**.

Rather than reporting the total model R², researchers usually calculate: $Incremental\ R^2 = R^2_{Full}-R^2_{Null}$.

This quantity measures the additional variance explained by PRS after accounting for covariates.

For example:

| Model      | R²   |
| ---------- | ---- |
| Null Model | 0.12 |
| Full Model | 0.18 |

Then:

$Incremental\ R^2 = 0.18 - 0.12=0.06$.

Thus, the PRS explains an additional 6% of phenotypic variation.

### Why Incremental R² Is Important?

Suppose sex explains a large fraction of height variation.
Without adjustment:

```text
Height ← Sex
```

A high R² could simply reflect sex differences. Incremental R² isolates the contribution of genetic prediction.

Therefore:

> Incremental R² is generally more informative than total R².

### Evaluating Disease Prediction

Quantitative traits are relatively straightforward. Diseases introduce additional challenges because outcomes are binary.

For example:

| Individual | Disease Status |
| ---------- | -------------- |
| A          | Case           |
| B          | Control        |
| C          | Control        |
| D          | Case           |

Disease status takes values:

$$
0 = Control
$$

$$
1 = Case
$$

Standard linear regression is no longer appropriate. Instead, logistic regression is commonly used.

### Pseudo R²

For binary outcomes, researchers often report:

$$
Pseudo\ R^2
$$

Pseudo R² is analogous to ordinary R² but is derived from logistic regression.

Several versions exist:

* McFadden R²
* Nagelkerke R²
* Cox-Snell R²

Although interpretation differs from linear regression, larger values indicate better prediction.

### Area Under the ROC Curve (AUC)

One of the most widely used measures of disease prediction is $AUC$. The Area Under the Receiver Operating Characteristic Curve. AUC measures how well a model distinguishes cases from controls.

Imagine randomly selecting:

* One case
* One control

AUC represents the probability that the case receives a higher predicted risk than the control.
Interpretation:

| AUC  | Interpretation       |
| ---- | -------------------- |
| 0.50 | Random guessing      |
| 0.60 | Weak prediction      |
| 0.70 | Moderate prediction  |
| 0.80 | Strong prediction    |
| 0.90 | Excellent prediction |

Thus, $AUC = 0.5$ means no predictive value.

### Liability Scale R²

For disease traits, observed case-control samples do not reflect true population prevalence. Researchers therefore often transform prediction accuracy to the:

**Liability Scale**

The liability threshold model assumes:

```text
Underlying Disease Liability
│
├── Genetics
├── Environment
└── Random Effects
```

Disease occurs when liability exceeds a threshold.Liability-scale $R^²$ attempts to estimate the proportion of underlying liability explained by PRS.

This metric is particularly common in psychiatric genetics.

### Risk Stratification

Another useful evaluation approach is risk stratification.

Individuals are divided into PRS groups.

For example:

| Group      | PRS Percentile |
| ---------- | -------------- |
| Bottom 20% | Lowest risk    |
| Middle 60% | Average risk   |
| Top 20%    | Elevated risk  |

Researchers then compare disease prevalence across groups.

### Decile Analysis

A common strategy is to divide individuals into ten groups:

| Decile | PRS Ranking |
| ------ | ----------- |
| 1      | Lowest 10%  |
| 2      | Next 10%    |
| ...    | ...         |
| 10     | Highest 10% |

Disease prevalence is then plotted across deciles.

A useful PRS should show increasing risk with increasing PRS.

### Odds Ratios

Risk stratification is often summarized using odds ratios.

Example:

| Group         | Odds Ratio |
| ------------- | ---------- |
| Bottom Decile | 1.0        |
| Top Decile    | 3.5        |

Interpretation:

Individuals in the top PRS decile have 3.5 times higher disease odds compared with individuals in the bottom decile.

### Common Pitfall: Overfitting

One of the most important challenges in PRS evaluation is overfitting.

Suppose we:

1. Build PRS
2. Test multiple parameters
3. Select the best-performing model
4. Evaluate on the same individuals

This leads to inflated prediction estimates.

The model has effectively "seen" the answer.

### Discovery, Tuning, and Target Samples

To avoid overfitting, polygenic prediction studies typically use three datasets.

### Discovery Dataset

Used for GWAS.

Produces SNP effect estimates.

#### Tuning Dataset

Used to optimize parameters.

Examples:

* P-value thresholds
* Clumping parameters
* Hyperparameters

#### Target Dataset

Used only once.

Provides an unbiased estimate of prediction performance.

These datasets must remain independent.

### Why Independence Matters?

If individuals appear in both discovery and target datasets:

```text
Discovery Sample
      ↑
      │ overlap
      ↓
Target Sample
```

prediction accuracy becomes inflated.

This is one of the most common mistakes in genetic prediction studies.

The gold standard is:

> Out-of-sample prediction using completely independent datasets.

#### Key Takeaways

* PRS performance must be evaluated carefully.
* For quantitative traits, prediction accuracy is commonly measured using R².
* Incremental R² quantifies the additional variance explained by PRS.
* For diseases, common metrics include AUC, pseudo R², liability-scale R², and odds ratios.
* Risk stratification provides an intuitive interpretation of prediction performance.
* Discovery, tuning, and target datasets should remain independent.
* Overfitting can substantially inflate reported prediction accuracy.
* Out-of-sample validation is essential for reliable evaluation.


### Clumping and P-Value Thresholding (C+PT)

In the previous chapter, we learned how Polygenic Risk Scores (PRS) are evaluated using metrics such as Incremental $R^2$, AUC, liability-scale $R^2$, and risk stratification.

The next question is:

> How do we actually select which SNPs should be included in a PRS?

A naïve approach would be to include every SNP from a GWAS and use its estimated effect size directly. However, this creates several problems.

First, many SNPs are highly correlated because of Linkage Disequilibrium (LD).

Second, many SNPs have extremely weak or noisy effect estimates.

Third, including millions of SNPs can introduce substantial statistical noise.

To address these issues, one of the earliest and most influential methods for polygenic prediction was developed:

**Clumping and P-value Thresholding (C+PT)**.

Despite the emergence of sophisticated Bayesian approaches, C+PT remains one of the most widely used baseline methods in genetic prediction.

### Why We Cannot Simply Use Every SNP?

Suppose a GWAS identifies a strong association signal on chromosome 6. The region may contain hundreds of SNPs with similar p-values.

For example:

| SNP | P-value |
| --- | ------- |
| rs1 | 1e-12   |
| rs2 | 2e-12   |
| rs3 | 4e-12   |
| rs4 | 8e-11   |
| rs5 | 1e-10   |

At first glance, it may appear that all of these SNPs represent independent discoveries. In reality, many of these variants are inherited together. They may simply reflect the same underlying association signal. If we include all of them in a PRS, we effectively count the same genetic information multiple times.
This can lead to:

* Redundant information
* Inflated variance
* Overfitting
* Reduced prediction accuracy

### Review: Linkage Disequilibrium

Linkage Disequilibrium (LD) describes the non-random association between nearby genetic variants.
Imagine a causal SNP:

```text
Causal Variant
      ↓
 Disease
```

Nearby SNPs often travel together through generations.

```text
rs1 ----\
rs2 ----- Causal SNP
rs3 ----/
```

As a result:

* rs1 appears associated
* rs2 appears associated
* rs3 appears associated

even if only one variant is actually causal.
This creates redundancy.

### The Core Idea of C+PT

Clumping and P-value Thresholding addresses two questions:

1. Which SNPs should be kept?
2. Which SNPs should be discarded?

The procedure consists of two major steps:

```text
GWAS Summary Statistics
          ↓
     Clumping
          ↓
P-value Thresholding
          ↓
      Final SNP Set
          ↓
      Polygenic Score
```

### Step 1: Clumping

The first step removes redundant SNPs.
The objective is:

> Keep the most significant SNP in an LD region and remove nearby correlated SNPs.

Suppose we observe:

| SNP | P-value |
| --- | ------- |
| rs1 | 1e-12   |
| rs2 | 2e-11   |
| rs3 | 4e-10   |

and:

```text
rs1 ↔ rs2 ↔ rs3
```
have strong LD.

Clumping chooses:

```text
Keep:
rs1
```

and removes:

```text
Discard:
rs2
rs3
```

because they provide largely overlapping information.

### How Clumping Works

The algorithm generally follows these steps:

#### Step 1

Sort SNPs by significance. 
Most significant SNP first.

Example:

| SNP | P-value |
| --- | ------- |
| rs1 | 1e-12   |
| rs2 | 1e-10   |
| rs3 | 1e-8    |
| rs4 | 1e-6    |

#### Step 2

Select the most significant SNP.

```text
Keep rs1
```

#### Step 3

Remove nearby SNPs that are highly correlated.

For example:

```text
Window = 250 kb
LD threshold = 0.1
```

All SNPs satisfying:

$$
r^2 > 0.1
$$

within the specified window are removed.

#### Step 4

Move to the next remaining SNP.

Repeat until all SNPs have been processed.

### Understanding LD Thresholds

The most common LD measure is:

$$
r^2
$$

Interpretation:

| r²  | Meaning              |
| --- | -------------------- |
| 0   | No correlation       |
| 0.2 | Weak correlation     |
| 0.5 | Moderate correlation |
| 0.8 | Strong correlation   |
| 1.0 | Perfect correlation  |

Typical clumping thresholds:

| Threshold | Interpretation |
| --------- | -------------- |
| 0.1       | Strict         |
| 0.2       | Moderate       |
| 0.5       | Relaxed        |

A smaller threshold removes more SNPs.

### Visualizing Clumping

Before clumping:

```text
rs1 rs2 rs3 rs4 rs5 rs6 rs7
 |   |   |   |   |   |   |
 Strong LD Block
```

After clumping:

```text
rs1
```

Only the strongest representative remains.

### Step 2: P-Value Thresholding

After clumping, we still have many SNPs.

The next question becomes:

> How significant should a SNP be before we include it in the PRS?

P-value thresholding answers this question.

#### The Logic Behind Thresholding

Very small p-values usually indicate stronger evidence of association.

For example:

| SNP | P-value |
| --- | ------- |
| rs1 | 1e-12   |
| rs2 | 1e-8    |
| rs3 | 1e-5    |
| rs4 | 0.01    |
| rs5 | 0.5     |

Some SNPs likely contain true signal.

Others may simply reflect statistical noise.

Thresholding attempts to identify the optimal balance.

### Common P-Value Thresholds

Researchers often evaluate multiple thresholds:

$$
5\times10^{-8}
$$

$$
10^{-6}
$$

$$
10^{-4}
$$

$$
10^{-2}
$$

$$
0.05
$$

$$
0.1
$$

$$
0.5
$$

$$
1.0
$$

Each threshold produces a different PRS.

#### Example

Suppose we choose:

$$
P < 10^{-4}
$$

Then:

| SNP | P-value | Keep? |
| --- | ------- | ----- |
| rs1 | 1e-8    | Yes   |
| rs2 | 5e-5    | Yes   |
| rs3 | 0.002   | No    |
| rs4 | 0.20    | No    |

Only rs1 and rs2 contribute to the score.

### Why Not Use Only Genome-Wide Significant SNPs?

This seems intuitive.

After all:

$$
P < 5\times10^{-8}
$$

is considered statistically significant.

However, complex traits are highly polygenic.

Many true causal variants have effects too small to achieve genome-wide significance.

Consequently:

```text
Very Strict Threshold
↓
Too Few SNPs
↓
Missed Signal
```

Prediction accuracy often improves when less stringent thresholds are used.

### The Tuning Process

Since we do not know the optimal threshold beforehand, we try many thresholds.

Example:

| Threshold | Incremental R² |
| --------- | -------------- |
| 5e-8      | 0.03           |
| 1e-6      | 0.05           |
| 1e-4      | 0.08           |
| 0.01      | 0.11           |
| 0.05      | 0.09           |

The best threshold is:

```text
0.01
```

because it achieves the highest prediction accuracy.

### Why Prediction Accuracy Forms a Hump?

A common observation is:

```text
Too Few SNPs
↓
Missed Signal
↓
Low Accuracy
```

Then:

```text
Moderate Number of SNPs
↓
Optimal Signal
↓
Highest Accuracy
```

Then:

```text
Too Many SNPs
↓
Noise Dominates
↓
Accuracy Falls
```

This produces the characteristic hump-shaped curve observed in many PRS studies.

### Calculating the Polygenic Score

After selecting SNPs through clumping and thresholding, the final score is calculated as:

$PRS_i =\sum_{j=1}^{m}\hat{\beta}*j x*{ij}$

where:

* $m$ is the selected SNP set
* $\hat{\beta}_j$ is the GWAS effect estimate
* $x_{ij}$ is the genotype count

### Practical Example Using PLINK

A typical PLINK workflow looks like:

```bash
plink \
  --bfile target_data \
  --score selected_snps.txt 1 2 3 \
  --out prs_scores
```

where:

* Column 1 = SNP ID
* Column 2 = Effect allele
* Column 3 = Effect size

PLINK multiplies genotype counts by SNP effects and sums across SNPs.

### Advantages of C+PT

The popularity of C+PT comes from its simplicity.

Advantages include:

* Easy to understand
* Computationally efficient
* Works with GWAS summary statistics
* Requires modest computing resources
* Easy to reproduce

Because of these strengths, C+PT remains the standard baseline method for polygenic prediction.

#### Limitations of C+PT

Despite its success, C+PT has important limitations.

#### Throws Away Information

Many SNPs are discarded during clumping.

Potentially useful signals may be lost.

#### Ignores Joint LD Structure

LD is handled indirectly through pruning rather than explicit modeling.

#### Uses Marginal Effects

Each SNP is treated independently.

Joint effects are not modeled.

#### Requires Tuning

Multiple thresholds must be tested.

This introduces complexity and potential overfitting.

### Why New Methods Were Developed?

The limitations of C+PT motivated the development of:

* BLUP
* SBLUP
* LDpred
* BayesR
* BayesC
* SBayesR
* SBayesRC

These methods attempt to:

* Use all SNPs
* Model LD directly
* Reduce noise through shrinkage
* Improve prediction accuracy

In many modern datasets, Bayesian approaches outperform C+PT. However, understanding C+PT remains essential because it provides the conceptual foundation for all subsequent PRS methods.

#### Key Takeaways

* Clumping and P-value Thresholding (C+PT) is the traditional PRS method.
* Clumping removes redundant SNPs in linkage disequilibrium.
* P-value thresholding selects SNPs based on GWAS significance.
* Multiple thresholds are usually tested.
* Prediction accuracy often follows a hump-shaped curve.
* C+PT is computationally simple and widely used.
* The method discards information and ignores full LD structure.
* These limitations motivated the development of modern Bayesian approaches.

### Best Linear Unbiased Prediction (BLUP)

In the previous chapter, we explored **Clumping and P-value Thresholding (C+PT)**, one of the earliest and most widely used approaches for constructing Polygenic Risk Scores (PRS).

While C+PT is simple and effective, it has several limitations:

* It removes many SNPs during clumping.
* It ignores much of the information contained in the genome.
* It treats SNPs independently.
* It requires tuning of p-value thresholds.

These limitations motivated the development of a new class of methods known as **whole-genome regression methods**.

One of the most important of these methods is:

> **Best Linear Unbiased Prediction (BLUP)**

BLUP represents a major conceptual shift in polygenic prediction because it attempts to use **all SNPs simultaneously**, rather than selecting only a subset. Understanding BLUP is crucial because many modern Bayesian methods such as BayesR and SBayesR can be viewed as extensions of the BLUP framework.

#### Why Move Beyond C+PT?

Recall the basic philosophy of C+PT:

```text
Millions of SNPs
        ↓
   Clumping
        ↓
Keep a subset
        ↓
Calculate PRS
```

This approach assumes that many SNPs are not useful.

However, GWAS studies have repeatedly shown that complex traits are highly polygenic.

For example:

| Trait                  | Number of Associated Variants |
| ---------------------- | ----------------------------- |
| Height                 | Thousands                     |
| BMI                    | Thousands                     |
| Schizophrenia          | Thousands                     |
| Educational Attainment | Thousands                     |

If thousands of variants contribute to a trait, removing large numbers of SNPs may discard valuable information.

This observation led researchers to ask:

> What if we use every SNP?

#### The Infinitesimal Model

BLUP is based on one of the oldest ideas in quantitative genetics:

> The Infinitesimal Model

The infinitesimal model assumes:

* Every SNP contributes to the trait.
* Each SNP has a very small effect.
* Effects are normally distributed.
* No SNP has an exceptionally large influence.

Mathematically, $\beta_j \sim N(0,\sigma^2_\beta)$.

This means SNP effects are drawn from a normal distribution centered at zero. Most SNPs have effects close to zero. A few SNPs have somewhat larger effects. Very large effects are rare.

#### Visualizing the Assumption

Under the infinitesimal model:

```text
                ^
                |
                |
             *****
          ***********
        ***************
------|-------0-------|------
```

Most SNP effects cluster around zero.

The distribution is symmetric.

Positive and negative effects occur with similar frequency.

This assumption works surprisingly well for many highly polygenic traits.

### What Does "Best Linear Unbiased Prediction" Mean?

The name BLUP comes from classical statistics.

Each word has meaning.

#### Best

Produces the smallest prediction error variance among linear unbiased estimators.

#### Linear

Prediction is a linear combination of SNP effects.

$\hat{y}=X\hat{\beta}$

#### Unbiased

Predictions are not systematically too high or too low.

## Prediction

The ultimate goal is accurate prediction of phenotypes.

### Linear Regression Revisited

Consider ordinary linear regression:

$
y = X\beta + e
$

where:

* $y$ = phenotype
* $X$ = genotype matrix
* $\beta$ = SNP effects
* $e$ = residual error

In GWAS we often estimate one SNP at a time.

BLUP takes a different approach.

Instead of fitting millions of separate regressions:

```text
SNP1 → Trait
SNP2 → Trait
SNP3 → Trait
```

BLUP fits all SNPs simultaneously.

### The Mixed Model Framework

BLUP is usually formulated as a linear mixed model.

The model becomes:

$$
y = Xb + Zu + e
$$

where:

* $Xb$ represents fixed effects
* $Zu$ represents genetic effects
* $e$ represents residual error

#### Fixed Effects

Fixed effects include known covariates:

* Sex
* Age
* Principal Components
* Batch Effects

Example:

$Xb=Age + Sex + PC1 + PC2$.

These effects are estimated directly.

### Random Effects

The genetic component is treated as a random effect.

$u \sim N(0,G\sigma_g^2)$

where:

* $G$ is the Genetic Relationship Matrix (GRM)
* $\sigma_g^2$ is genetic variance

This assumption is central to BLUP.

Instead of estimating each SNP independently, BLUP models their collective contribution.

### The Genetic Relationship Matrix (GRM)

One of the most important concepts in BLUP is the Genetic Relationship Matrix.

The GRM measures genetic similarity between individuals.

Suppose we have three people:

| Person | Relationship |
| ------ | ------------ |
| A vs A | 1.0          |
| A vs B | 0.10         |
| A vs C | 0.02         |
| B vs C | 0.08         |

These values form a matrix:

$G =\begin{bmatrix}
1 & 0.10 & 0.02 \
0.10 & 1 & 0.08 \
0.02 & 0.08 & 1
\end{bmatrix}
$

The matrix summarizes how genetically similar individuals are.

### Why Use the GRM?

The intuition is simple.

If genetics influences a trait:

```text
More Genetically Similar
            ↓
More Phenotypically Similar
```

BLUP uses this relationship to estimate genetic effects.

#### Shrinkage

A key feature of BLUP is **shrinkage**.

Instead of trusting GWAS estimates completely, BLUP pulls effect sizes toward zero.

Suppose GWAS estimates:

| SNP | GWAS Effect |
| --- | ----------- |
| rs1 | 0.40        |
| rs2 | 0.05        |
| rs3 | 0.01        |

BLUP may produce:

| SNP | BLUP Effect |
| --- | ----------- |
| rs1 | 0.25        |
| rs2 | 0.03        |
| rs3 | 0.005       |

Large effects are reduced.

Small effects are reduced even more.

This process decreases noise and improves prediction.

### Why Shrinkage Helps

Imagine estimating SNP effects from a small sample.

Some estimates become large simply by chance.

Without shrinkage:

```text
Signal + Noise
```

With shrinkage:

```text
Mostly Signal
```

BLUP reduces overfitting by shrinking noisy estimates toward zero.

### BLUP Polygenic Scores

After estimating SNP effects, PRS calculation is unchanged.

We still compute:

$PRS_i=\sum_j\hat{\beta}*j x*{ij}$.

The difference is:

```text
C+PT
↓
Selected SNPs
```

versus

```text
BLUP
↓
All SNPs
```

Every SNP contributes to the final score.

### Advantages of BLUP

BLUP offers several improvements over C+PT.

#### Uses All SNPs

No information is discarded.

#### No Arbitrary P-value Thresholds

No need to choose:

```text
P < 0.01
P < 0.05
P < 0.1
```

#### Handles Polygenicity Naturally

Works well when many variants contribute small effects.

#### Reduced Overfitting

Shrinkage improves stability.

### Limitations of BLUP

Despite its strengths, BLUP has important limitations.

### Assumes All SNPs Have Effects

BLUP assumes $\beta_j \neq 0$ for every SNP.

Reality is often different.

Many SNPs likely have no effect at all.

### Assumes a Single Normal Distribution

All SNP effects come from one distribution $N(0,\sigma_\beta^2)$.

Complex traits may contain:

* Null SNPs
* Small-effect SNPs
* Medium-effect SNPs
* Large-effect SNPs

A single distribution may be unrealistic.

#### Limited Flexibility

BLUP cannot easily model:

* Sparse architectures
* Functional annotations
* Different effect-size classes

These limitations motivated Bayesian methods.

### BLUP vs C+PT

| Feature                     | C+PT      | BLUP     |
| --------------------------- | --------- | -------- |
| Uses all SNPs               | No        | Yes      |
| Requires tuning             | Yes       | No       |
| Models LD directly          | Partially | Better   |
| Shrinkage                   | No        | Yes      |
| Assumes infinitesimal model | No        | Yes      |
| Computational complexity    | Low       | Moderate |

BLUP was a major advance because it demonstrated that using all SNPs can improve prediction.

### From BLUP to Bayesian Methods

Researchers soon realized that the assumptions of BLUP were often too restrictive.

Questions emerged:

* What if many SNPs have zero effect?
* What if effect sizes come from multiple distributions?
* What if we want to include biological annotations?

Answering these questions led to Bayesian approaches such as:

* BayesA
* BayesB
* BayesC
* BayesR
* SBayesR
* SBayesRC

These methods generalize the BLUP framework while relaxing its assumptions.

### Key Takeaways

* BLUP is a whole-genome regression method.
* BLUP assumes all SNPs contribute small effects.
* SNP effects are assumed to follow a normal distribution.
* BLUP uses all SNPs simultaneously.
* The method relies on a Genetic Relationship Matrix (GRM).
* Shrinkage reduces overfitting and improves prediction.
* BLUP eliminates the need for clumping and p-value thresholding.
* The infinitesimal model is both BLUP's strength and its limitation.
* Modern Bayesian methods evolved directly from the BLUP framework.


### Bayesian Methods for Polygenic Prediction

In the previous chapter, we explored **Best Linear Unbiased Prediction (BLUP)**, one of the first whole-genome regression approaches for polygenic prediction.

BLUP represented a major advance over Clumping and P-value Thresholding (C+PT) because it:

* Uses all SNPs simultaneously
* Models genome-wide effects
* Applies shrinkage to reduce noise
* Eliminates arbitrary p-value thresholds

However, BLUP makes a strong assumption:

> Every SNP has a non-zero effect drawn from the same normal distribution.

Reality is often more complicated.

Some SNPs may have:

* No effect
* Tiny effects
* Moderate effects
* Large effects

This observation motivated the development of **Bayesian methods**, which provide a more flexible framework for modeling genetic architecture.

Modern approaches such as:

* BayesA
* BayesB
* BayesC
* BayesR
* SBayesR
* SBayesRC

all originate from Bayesian statistics.

Understanding the Bayesian framework is therefore essential for understanding modern polygenic prediction.

#### Why Bayesian Methods?

Suppose we analyze one million SNPs.

BLUP assumes $\beta_j \sim N(0,\sigma_\beta^2)$.

for every SNP.

This means:

```text id="k7ybj4"
SNP1 → Small Effect
SNP2 → Small Effect
SNP3 → Small Effect
...
SNP1,000,000 → Small Effect
```

While reasonable for highly polygenic traits, this assumption is often unrealistic.

In reality:

```text
Many SNPs → No Effect

Some SNPs → Tiny Effect

Few SNPs → Moderate Effect

Very Few SNPs → Large Effect
```

Bayesian models can represent this complexity.

### The Bayesian Philosophy

Classical statistics treats parameters as fixed but unknown quantities.

For example:

$$
\beta = 0.05
$$

We do not know its value, but we assume a single true value exists.

Bayesian statistics takes a different view.

It treats parameters as random variables.

Instead of asking:

> What is the value of β?

we ask:

> What is the probability distribution of β?

This seemingly small change has enormous consequences.

#### Prior Beliefs

Bayesian analysis begins with a **prior distribution**.

A prior represents our beliefs before seeing the data.

For example:

Suppose we believe most SNPs have effects close to zero.

We might write:

$$
\beta_j \sim N(0,\sigma^2)
$$

This prior expresses our belief that:

* Most effects are small
* Positive and negative effects are equally likely
* Large effects are rare

#### Data Provides Evidence

Next we observe data.

For example:

| SNP | GWAS Effect |
| --- | ----------- |
| rs1 | 0.05        |
| rs2 | 0.01        |
| rs3 | 0.00        |

The data contribute information through the likelihood function.

The likelihood answers:

> How probable is the observed data if a particular parameter value were true?

### Bayes' Theorem

Bayesian inference combines:

* Prior beliefs
* Observed data

using Bayes' theorem.

In words:

```text id="8bx8i2"
Posterior
=
Prior Information
+
Observed Data
```

The posterior distribution represents our updated beliefs after observing the data.

#### Understanding the Posterior Distribution

Suppose our prior believes:

```text 
β ≈ 0
```

but the GWAS strongly suggests:

```text 
β ≈ 0.3
```

The posterior combines both sources of information.

Result:

```text 
β ≈ 0.2
```

The final estimate lies somewhere between prior expectations and observed evidence.

#### Why This Is Useful

GWAS estimates are noisy.

Especially when sample sizes are limited.

Suppose a SNP has:

$$
\hat{\beta}=0.4
$$

This estimate may be inflated by chance.

Bayesian methods naturally shrink estimates toward plausible values.

Instead of:

```text
0.40
```

the posterior estimate may become:

```text id="7l8v6v"
0.22
```

This often improves prediction accuracy.

#### From BLUP to Bayesian Models

BLUP uses a single normal prior:

$$
\beta_j \sim N(0,\sigma_\beta^2)
$$

Bayesian methods allow more flexible priors.

For example:

```text
Many SNPs:
β = 0

Some SNPs:
Small Effects

Few SNPs:
Large Effects
```

These assumptions often better match biological reality.

#### Sparse Genetic Architecture

Consider a trait influenced by:

* 100 causal variants
* 1,000,000 measured SNPs

Most SNPs have $\beta = 0$.

BLUP struggles because it assumes every SNP contributes.

Bayesian models can explicitly represent sparsity.

This is one reason they often outperform BLUP.

#### BayesC: The First Important Bayesian Model

One of the most influential models is **BayesC**.

BayesC assumes:

Some SNPs have no effect:

$$
\beta_j = 0
$$

while others have non-zero effects:

$\beta_j \sim N(0,\sigma_\beta^2)$.

The model becomes:

$$
\beta_j=
\begin{cases}
0 & \text{with probability }\pi \
N(0,\sigma_\beta^2) & \text{with probability }1-\pi
\end{cases}
$$

This is known as a **spike-and-slab prior**.

##### Understanding the Spike-and-Slab Prior

Imagine a distribution consisting of two parts.

```text 
Spike
|
|
|
|
0
```

Most SNPs belong here.

They have exactly zero effect.

The second part:

```text 
Slab
```

contains SNPs with non-zero effects.

This simple idea dramatically improves flexibility.

### Why BayesC Improves Prediction?

BayesC recognizes that:

```text id="jsyepm"
Many SNPs
↓
No Effect
```

Rather than forcing every SNP to contribute.

This leads to:

* Less noise
* Better shrinkage
* Improved prediction

especially when the true genetic architecture is sparse.

#### The Challenge of Posterior Inference

Unfortunately, Bayesian models are mathematically complex.

The posterior distribution usually cannot be solved directly.

There is no closed-form solution.

Instead we must approximate the posterior.

This leads to one of the most important tools in Bayesian statistics:

**Markov Chain Monte Carlo (MCMC)**.

#### Why MCMC Is Needed?

Consider millions of SNPs.

Each SNP has:

* An unknown effect
* An unknown inclusion status

The number of possible parameter combinations becomes enormous.

Direct calculation becomes impossible.

Instead we generate samples from the posterior distribution.

#### Intuition Behind MCMC

Imagine exploring a landscape.

```text id="9jjlwm"
High Posterior Probability
         ^
         |
       Peak
      /   \
     /     \
```

The algorithm spends more time near regions of high posterior probability.

Over many iterations, sampled values reconstruct the posterior distribution.

#### The Basic BayesC Algorithm

Initialize:

```text id="hr8t53"
β
σ²β
σ²e
π
```

Then repeatedly:

1. Update SNP effects
2. Update genetic variance
3. Update residual variance
4. Update inclusion probability

Repeat thousands of times.

Eventually the chain converges.

### Trace Plots

One way to assess convergence is through trace plots.

A trace plot shows sampled values across iterations.

Good convergence looks like:

```text id="9zck4h"
Stable fluctuations
around a constant mean
```

Poor convergence looks like:

```text id="2xibqe"
Upward trend
or
downward trend
```

indicating the chain has not stabilized.

#### Posterior Means

After convergence:

We summarize the posterior using:

$$
E(\beta_j|Data)
$$

This posterior mean becomes the final SNP effect estimate.

These estimates are then used to calculate Polygenic Risk Scores.

### Why Bayesian Methods Often Outperform BLUP?

Bayesian models can:

* Represent sparse architectures
* Model multiple effect-size classes
* Perform adaptive shrinkage
* Better reflect biological reality

As a result they often produce:

```text id="7sv2lm"
Higher Prediction Accuracy
```

than BLUP.

#### The Evolution of Bayesian Models

BayesC was only the beginning.

Researchers later developed:

**BayesA**

All SNPs have effects.

Heavy-tailed prior.

**BayesB**

Many SNPs have zero effects.

Non-zero SNPs follow a separate distribution.

**BayesC**

Spike-and-slab prior.

Computationally simpler.

**BayesR**

Multiple effect-size distributions.

More realistic architecture.

**SBayesR**

Summary-statistics implementation of BayesR.

**SBayesRC**

Extends SBayesR by incorporating functional annotations.

### Summary

Bayesian methods transformed polygenic prediction by allowing more realistic assumptions about genetic architecture.

Instead of assuming:

```text id="v1ib1m"
All SNPs have small effects
```

Bayesian approaches allow:

```text id="l3fd3d"
Many SNPs → Zero Effect

Some SNPs → Small Effect

Few SNPs → Large Effect
```

This flexibility often leads to improved prediction accuracy.

Modern methods such as SBayesR and SBayesRC represent the current state of the art in polygenic prediction and are direct descendants of these Bayesian ideas.

#### Key Takeaways

* Bayesian methods treat SNP effects as random variables.
* Prior distributions encode biological assumptions.
* Bayes' theorem combines prior information and observed data.
* The posterior distribution represents updated beliefs.
* BayesC introduces the spike-and-slab prior.
* Bayesian methods naturally model sparse genetic architectures.
* MCMC sampling is used when posterior distributions cannot be solved analytically.
* Posterior means provide SNP effect estimates.
* Bayesian methods generally outperform BLUP when genetic architectures are complex.
* Modern methods such as SBayesR and SBayesRC are built on Bayesian foundations.



### Markov Chain Monte Carlo (MCMC): The Engine Behind Bayesian Polygenic Prediction

In the previous chapter, we introduced Bayesian methods for polygenic prediction and learned how models such as BayesC use prior distributions and Bayes' theorem to estimate SNP effects.

A major challenge immediately appeared:

> The posterior distribution cannot usually be calculated directly.

For simple statistical models, posterior distributions may have closed-form solutions.

Unfortunately, polygenic prediction is not a simple problem.

Modern datasets may contain:

* Millions of SNPs
* Thousands of individuals
* Multiple unknown parameters
* Complex dependency structures

The posterior distribution becomes extraordinarily complicated.

This is where one of the most powerful tools in modern statistics enters the picture:

**Markov Chain Monte Carlo (MCMC)**.

MCMC is the computational engine that makes Bayesian polygenic prediction possible.

Without MCMC, methods such as:

* BayesA
* BayesB
* BayesC
* BayesR

would be nearly impossible to implement.

#### Why Do We Need MCMC?

Suppose we want to estimate SNP effects:

$$
\beta_1,\beta_2,\ldots,\beta_m
$$

For one million SNPs:

$$
m = 1,000,000
$$

The posterior distribution becomes:

$$
P(\beta_1,\beta_2,\ldots,\beta_m|Data)
$$

This is a probability distribution in an extremely high-dimensional space.

Direct calculation is impossible.

Instead of computing the posterior exactly, we approximate it by generating samples from it.

This is the central idea behind MCMC.

##### What Is a Markov Chain?

The first part of MCMC is the **Markov Chain**.

A Markov chain is a sequence of states where:

> The next state depends only on the current state.

Not on the entire history.

For example:

```text id="yg1k1k"
State 1 → State 2 → State 3 → State 4
```

To determine State 4:

```text id="m9t6n6"
Need State 3 only
```

Earlier states do not matter directly.

This property is known as:

> The Markov Property

#### Everyday Example

Imagine walking through a city.

At each intersection:

* Turn left
* Turn right
* Go straight

Your next move depends only on your current location.

Not on the entire route you previously traveled.

This is a Markov process.

##### What Is Monte Carlo?

The second part of MCMC is:

**Monte Carlo**

Monte Carlo methods use randomness to approximate difficult calculations.

Suppose we want to estimate:

$$
\pi
$$

One approach:

* Randomly throw points into a square.
* Count how many land inside a circle.

As the number of points increases:

```text id="9crzz7"
Random Sampling
        ↓
Approximation
        ↓
True Value
```

Monte Carlo methods use the same philosophy.

Random sampling approximates difficult mathematical quantities.

#### Combining the Two Ideas

MCMC combines:

```text id="4n4vtp"
Markov Chain
      +
Monte Carlo
```

to generate samples from complicated probability distributions.

Rather than computing:

$$
P(\beta|Data)
$$

directly,

we generate many samples:

$$
\beta^{(1)}
$$

$$
\beta^{(2)}
$$

$$
\beta^{(3)}
$$

$$
\cdots
$$

These samples gradually reconstruct the posterior distribution.

#### Visualizing the Posterior

Imagine a posterior distribution:

```text id="lz2srl"
             *
           *****
         *********
       *************
------|------0------|------
```

The goal is to understand:

* Shape
* Mean
* Variance

Instead of calculating everything analytically, MCMC generates random samples.

Over time:

```text id="tjlwmh"
Samples
   ↓
Histogram
   ↓
Posterior Distribution
```

#### The Core Idea

Suppose the posterior has regions of:

* High probability
* Low probability

MCMC spends:

```text id="mvwskg"
More Time
↓
High Probability Regions
```

and

```text id="bz7l8v"
Less Time
↓
Low Probability Regions
```

As a result, the sample distribution gradually resembles the true posterior.

#### The Bayesian Parameters in Polygenic Prediction

In BayesC, several unknown quantities must be estimated.

These include:

#### SNP Effects

$$
\beta
$$

#### Genetic Variance

$$
\sigma_\beta^2
$$

#### Residual Variance

$$
\sigma_e^2
$$

#### Inclusion Probability

$$
\pi
$$

Each parameter depends on the others.

This dependency makes direct estimation difficult.

### Why Parameters Depend on Each Other

Consider one SNP.

Its estimated effect depends on:

* Residual variance
* Genetic variance
* Inclusion probability

But:

Residual variance depends on SNP effects.

Genetic variance depends on SNP effects.

Inclusion probability depends on SNP effects.

Thus:

```text id="29vkyy"
β depends on σ²

σ² depends on β
```

Everything is interconnected.

#### Gibbs Sampling

The most common MCMC algorithm in Bayesian genetics is:

**Gibbs Sampling**

The idea is surprisingly simple.

Instead of estimating all parameters simultaneously:

```text id="l4n6pw"
β
σ²β
σ²e
π
```

we update them one at a time.

##### Gibbs Sampling Workflow

Start with initial values.

Example:

```text id="1kzr0q"
β = 0

σ²β = 1

σ²e = 1

π = 0.95
```

Then repeatedly update:

```text id="z3m6s3"
Update β
↓
Update σ²β
↓
Update σ²e
↓
Update π
↓
Repeat
```

Thousands of times.

##### One Iteration of Gibbs Sampling

Suppose we are at iteration:

```text id="ivq8nq"
Iteration 500
```

Step 1:

Update SNP effects.

```text id="snn9kq"
β → new value
```

Step 2:

Update genetic variance.

```text id="ohvkqn"
σ²β → new value
```

Step 3:

Update residual variance.

```text id="qovldm"
σ²e → new value
```

Step 4:

Update inclusion probability.

```text id="9ewv1e"
π → new value
```

Move to iteration 501.

Repeat.

#### Burn-In Period

Early MCMC samples are often unreliable.

The chain has not yet reached the posterior distribution.

Example:

```text id="m2s0sj"
Iteration 1
Iteration 2
Iteration 3
```

These values may depend heavily on starting conditions.

Therefore we discard them.

This discarded portion is called:

> Burn-in

#### Example Burn-In

Suppose:

```text id="tt8olz"
Total Iterations = 50,000
```

Burn-in:

```text id="mycezu"
First 10,000 iterations
```

Retained:

```text id="0f9mxy"
Remaining 40,000 iterations
```

Only retained samples are used for inference.

#### Trace Plots

One of the most important diagnostic tools is the trace plot.

A trace plot shows:

```text id="h5w7pc"
Parameter Value
       vs
Iteration Number
```

##### Good Trace Plot

A well-behaved chain looks like:

```text id="4ehjlwm"
~~~~~~~~~~~
~~~~~~~~~~~
~~~~~~~~~~~
```

The chain fluctuates around a stable mean.

This suggests convergence.

#### Bad Trace Plot

A problematic chain may look like:

```text id="upfwkq"
/
 /
  /
   /
```

or

```text id="cy1q7i"
\/\/\/\/\/\/
```

with large trends.

Such patterns indicate poor convergence.

#### What Is Convergence?

Convergence means:

> The chain has reached the target posterior distribution.

After convergence:

```text id="vzv6hp"
Samples
≈
Posterior Distribution
```

Inference becomes reliable.

##### Posterior Mean

After collecting samples:

$$
\beta^{(1)},
\beta^{(2)},
\ldots,
\beta^{(N)}
$$

we calculate:

$\bar{\beta} =\frac{1}{N}
\sum_{i=1}^{N}
\beta^{(i)}
$

This posterior mean becomes the SNP effect estimate.

### Posterior Variance

We can also estimate uncertainty $Var(\beta)$.

Large variance:

```text id="sgnyjd"
High Uncertainty
```

Small variance:

```text id="7uq2kw"
High Confidence
```

This is one of the major advantages of Bayesian inference.

### Posterior Inclusion Probability (PIP)

In BayesC-type models, a SNP may be:

```text id="8ghn3u"
Included
```

or

```text id="awunvb"
Excluded
```

during each iteration.

Suppose:

```text id="qvgsfz"
50,000 iterations
```

A SNP is included:

```text id="7z7ifc"
45,000 times
```

Then:

$PIP=\frac{45000}{50000}= 0.90$

This value is called the:

**Posterior Inclusion Probability (PIP)**

### Interpreting PIP

| PIP  | Interpretation              |
| ---- | --------------------------- |
| 0.01 | Almost certainly irrelevant |
| 0.10 | Weak evidence               |
| 0.50 | Moderate evidence           |
| 0.90 | Strong evidence             |
| 0.99 | Very strong evidence        |

High-PIP SNPs are strong candidate causal variants.

#### Computational Cost

MCMC is powerful but expensive.

Example:

```text id="8zl2vq"
1 Million SNPs
50,000 Iterations
```

can require:

* Hours
* Days
* Large memory resources

This computational burden motivated later developments such as:

* SBayesR
* SBayesRC

which operate using summary statistics and efficient matrix methods.

**Strengths of MCMC**

MCMC provides:

* Flexible Bayesian inference
* Full posterior distributions
* Uncertainty estimation
* Posterior inclusion probabilities
* Realistic genetic architecture modeling

These strengths explain why Bayesian methods often outperform simpler approaches.

**Weaknesses of MCMC**

MCMC can be:

* Computationally expensive
* Slow to converge
* Difficult to diagnose
* Sensitive to poor mixing

These challenges motivated newer algorithms.

Nevertheless, MCMC remains one of the foundational tools of Bayesian genetics.

#### Summary

MCMC is the computational engine that makes Bayesian polygenic prediction possible.

Rather than solving complex posterior distributions analytically, MCMC approximates them through random sampling.

Methods such as BayesC repeatedly sample:

* SNP effects
* Genetic variance
* Residual variance
* Inclusion probabilities

Over many iterations, these samples reconstruct the posterior distribution.

Posterior means become SNP effect estimates, while posterior inclusion probabilities help identify important variants.

Understanding MCMC is essential for interpreting the output of modern Bayesian polygenic prediction methods.

#### Key Takeaways

* MCMC stands for Markov Chain Monte Carlo.
* MCMC approximates complex posterior distributions through sampling.
* A Markov chain depends only on the current state.
* Monte Carlo methods use random sampling for approximation.
* Gibbs sampling is the most common MCMC algorithm in Bayesian genetics.
* Burn-in samples are discarded.
* Trace plots help assess convergence.
* Posterior means provide SNP effect estimates.
* Posterior Inclusion Probabilities (PIP) quantify SNP importance.
* MCMC forms the computational foundation of BayesC, BayesR, and related methods.



### SBayesR: A Modern Bayesian Method for Polygenic Prediction

In the previous chapter, we learned how summary statistics and LD reference panels made it possible to perform polygenic prediction without access to individual-level genotype data.

This development led to one of the most influential methods in modern statistical genetics:

> **SBayesR (Summary Statistics BayesR)**

SBayesR combines:

* Bayesian modeling
* Summary statistics
* LD information
* Mixture priors

to estimate SNP effects across the entire genome.

Today, SBayesR is one of the most widely used methods for constructing Polygenic Risk Scores (PRS) from GWAS summary statistics.

It serves as the foundation for even more advanced methods such as SBayesRC.

### Why Was SBayesR Developed?

Earlier methods suffered from important limitations.

#### Clumping and P-value Thresholding (C+PT)

Problems:

* Discards SNPs
* Ignores joint LD structure
* Requires threshold tuning

*BLUP*

Problems:

* Assumes every SNP has an effect
* Uses a single normal distribution
* Cannot model sparse architectures

 *BayesC*

Advantages:

* Models sparsity
* Flexible priors

Problems:

* Requires individual-level data
* Computationally expensive

Researchers wanted:

```text id="qkzwm4"
Bayesian Flexibility
+
Summary Statistics
+
Scalability
```

SBayesR was the answer.

### The Main Idea Behind SBayesR

SBayesR starts from two inputs:

#### Input 1: GWAS Summary Statistics

For each SNP:

| SNP | Effect | SE   | P-value |
| --- | ------ | ---- | ------- |
| rs1 | 0.05   | 0.01 | 1e-8    |
| rs2 | 0.02   | 0.01 | 0.02    |
| rs3 | -0.01  | 0.02 | 0.40    |

#### Input 2: LD Reference Panel

A matrix describing correlations between SNPs.

```text id="r6zn6l"
GWAS Summary Statistics
            +
        LD Matrix
            ↓
         SBayesR
            ↓
     Posterior SNP Effects
            ↓
            PRS
```

#### The BayesR Philosophy

Recall BLUP:

$$
\beta_j
\sim
N(0,\sigma^2)
$$

Every SNP effect comes from the same distribution.

BayesR introduced a more realistic assumption.

Instead of one distribution:

```text id="5pjcbb"
Small Effects Only
```

BayesR assumes multiple effect-size classes.

#### The Five-Component Mixture Prior

SBayesR uses a mixture distribution.

Mathematically:

$
\beta_j
\sim
\sum_{k=1}^{K}
\pi_k
N(0,\gamma_k\sigma^2)
$

where:

* $\pi_k$ = probability of component k
* $\gamma_k$ = variance scaling factor

In practice, SBayesR typically uses five components.

#### The Five SNP Classes

A common configuration is:

| Component | Variance |
| --------- | -------- |
| Class 1   | 0        |
| Class 2   | 0.0001   |
| Class 3   | 0.001    |
| Class 4   | 0.01     |
| Class 5   | 1        |

Interpretation:

```text id="t45v9x"
Class 1 → No Effect

Class 2 → Tiny Effect

Class 3 → Small Effect

Class 4 → Moderate Effect

Class 5 → Large Effect
```

This architecture closely resembles real GWAS observations.

#### Why Multiple Classes Matter

Imagine a trait influenced by:

* Thousands of tiny-effect variants
* Hundreds of small-effect variants
* Dozens of moderate-effect variants
* A few large-effect variants

BLUP struggles because all SNPs share one distribution.

SBayesR can assign each SNP to the most appropriate class.

This flexibility improves prediction accuracy.

#### The Role of Mixing Proportions

The parameters:

$$
\pi_1,\pi_2,\pi_3,\pi_4,\pi_5
$$

represent the proportion of SNPs in each class.

Example:

| Component       | Proportion |
| --------------- | ---------- |
| Zero Effect     | 95%        |
| Tiny Effect     | 3%         |
| Small Effect    | 1.5%       |
| Moderate Effect | 0.4%       |
| Large Effect    | 0.1%       |

This means:

Most SNPs contribute little or nothing.

A small number carry most of the signal.

#### Learning the Genetic Architecture

One of the strengths of SBayesR is that it learns these proportions directly from the data.

Instead of specifying:

```text id="k6efjr"
95% Zero Effects
```

the algorithm estimates:

$$
\pi
$$

during MCMC sampling.

This provides insight into the genetic architecture of the trait.

#### Recovering Joint SNP Effects

GWAS summary statistics contain marginal effects.

Recall:

$
b=R\beta
+
\epsilon
$

where:

* $b$ = observed GWAS effects
* $R$ = LD matrix
* $\beta$ = joint SNP effects

SBayesR attempts to estimate $\beta$.

using both:

* GWAS effects
* LD structure

This converts marginal effects into joint effects.

#### Why LD Must Be Modeled

Suppose:

```text id="ct8rsv"
rs1 ↔ rs2 ↔ rs3
```

are strongly correlated.

A naïve analysis may assign signal to all three SNPs.

SBayesR evaluates them jointly.

Result:

```text id="22juzv"
One SNP may receive
most of the signal
```

while the others are shrunk toward zero.

This improves interpretability and prediction.

#### The Computational Challenge

A major obstacle is the LD matrix.

For:

```text id="8gmj55"
1 Million SNPs
```

the full LD matrix contains $10^{12}$.

entries.

This is enormous.

#### Sparse LD Matrices

Most SNPs are only correlated with nearby variants.

Therefore:

```text id="wjlwmu"
Most LD Values
≈ 0
```

SBayesR stores only important LD values.

This creates a sparse matrix representation.

Benefits:

* Reduced memory
* Faster computation
* Improved scalability

#### Eigen Decomposition

Another important optimization is matrix decomposition.

The LD matrix can be written as: $R=U \Lambda U^T$.

where:

* $U$ = eigenvectors
* $\Lambda$ = eigenvalues

This representation reduces computational burden during MCMC sampling.

Large biobank analyses would be nearly impossible without these optimizations.

#### MCMC in SBayesR

SBayesR still relies on Bayesian sampling.

The algorithm repeatedly updates:

```text id="ty9px8"
SNP Effects
↓
Mixture Membership
↓
Genetic Variance
↓
Residual Variance
```

This iterative process gradually reconstructs the posterior distribution.

##### Posterior SNP Effects

After convergence, each SNP receives:

$$
E(\beta_j|Data)
$$

the posterior mean effect.

Example:

| SNP | Posterior Effect |
| --- | ---------------- |
| rs1 | 0.03             |
| rs2 | 0.00             |
| rs3 | 0.002            |

These estimates are typically more stable than raw GWAS effects.

### Constructing the PRS

Once posterior effects are estimated:

$PRS_i=\sum_j
\hat{\beta}*j
x*{ij}$

This formula is identical to previous methods.

The difference lies in how the SNP effects were estimated.

#### Practical Workflow

The overall workflow looks like:

```text id="vgbknv"
GWAS Summary Statistics
            ↓
Summary Statistic QC
            ↓
LD Reference Matrix
            ↓
SBayesR
            ↓
Posterior SNP Effects
            ↓
PLINK --score
            ↓
Polygenic Risk Scores
```

A typical SBayesR analysis uses:

Example:

```text id="8oq2yj"

```

containing:

* SNP
* Alleles
* Effect size
* Standard error
* Sample size

#### LD Matrix

Example:

```text id="jq3rzc"
ukbEURu_hm3_sparse
```

generated from a reference panel.


The exact command varies depending on the LD matrix and reference panel.

#### SBayesR Output Files

Typical outputs include:

### SNP Effects

```text id="kp9g6x"
trait.snpRes
```

Contains posterior SNP effects.

### MCMC Summary

```text id="dnmvdr"

```

Contains parameter estimates.

## Convergence Information

Used to assess MCMC behavior.

## Advantages of SBayesR

SBayesR provides several important advantages.

#### Uses All SNPs

No arbitrary SNP selection.

#### Models LD Directly

Joint SNP effects are estimated.

#### Flexible Mixture Prior

Realistic effect-size distributions.

#### Summary Statistics Only

No individual-level data required.

#### Scalable

Suitable for biobank-scale analyses.

#### Limitations of SBayesR

Despite its success, SBayesR has limitations.

#### No Functional Annotation

Every SNP is treated equally.

Biological information is ignored.

#### LD Reference Dependence

Prediction quality depends on the LD reference panel.

#### Computational Cost

Still more expensive than simple methods such as C+PT.

These limitations motivated the development of SBayesRC.

#### SBayesR vs Earlier Methods

| Feature             | C+PT    | BLUP   | SBayesR |
| ------------------- | ------- | ------ | ------- |
| Uses All SNPs       | No      | Yes    | Yes     |
| Models LD           | Partial | Better | Yes     |
| Summary Statistics  | Yes     | No     | Yes     |
| Mixture Prior       | No      | No     | Yes     |
| Sparse Architecture | No      | No     | Yes     |
| Bayesian Framework  | No      | No     | Yes     |

SBayesR combines the strengths of earlier approaches while avoiding many of their limitations.

#### Key Takeaways

* SBayesR extends BayesR to GWAS summary statistics.
* It requires only summary statistics and an LD reference panel.
* SNP effects are modeled using a five-component mixture prior.
* Most SNPs are assigned near-zero effects.
* LD information allows estimation of joint SNP effects.
* Sparse matrix methods make large-scale analyses feasible.
* Posterior mean effects are used to construct PRS.
* SBayesR is one of the most successful modern polygenic prediction methods.


### SBayesRC: Incorporating Biological Knowledge into Polygenic Prediction

In the previous chapter, we explored SBayesR, one of the most successful Bayesian methods for constructing Polygenic Risk Scores (PRS) from GWAS summary statistics.

SBayesR improved prediction by:

* Using all SNPs simultaneously
* Modeling linkage disequilibrium (LD)
* Applying Bayesian shrinkage
* Using mixture distributions for SNP effects

However, SBayesR makes an important assumption:

> Every SNP is treated equally before observing the data.

From a biological perspective, this assumption is unlikely to be true.

Consider two SNPs:

| SNP | Location            |
| --- | ------------------- |
| rs1 | Protein-coding exon |
| rs2 | Intergenic region   |

Would we expect both SNPs to have the same probability of being causal?

Probably not.

Decades of molecular biology have shown that some genomic regions are much more likely to influence phenotype than others.

This observation motivated the development of:

**SBayesRC (Summary Statistics BayesR with Functional Annotations)**

### Why Biological Information Matters

The human genome contains roughly:

$$
3.2 \times 10^9
$$

base pairs.

Only a small fraction directly encodes proteins.

Many genomic regions differ dramatically in biological importance.

Examples include:

* Protein-coding exons
* Promoters
* Enhancers
* Transcription factor binding sites
* Conserved regions
* Regulatory elements

A SNP located in one of these regions may be more likely to influence phenotype than a randomly selected variant.

### The Central Question

Suppose we observe two SNPs with similar GWAS evidence.

| SNP | P-value |
| --- | ------- |
| rs1 | 1e-5    |
| rs2 | 1e-5    |

If rs1 lies inside a highly conserved coding region while rs2 lies in a biologically inactive region:

Should both SNPs receive the same prior probability of being causal?

SBayesRC says:

> No.

### What Are Functional Annotations?

Functional annotations describe biological properties of genomic regions.

Examples include:

| Annotation    | Meaning                  |
| ------------- | ------------------------ |
| Coding        | Protein sequence         |
| Promoter      | Gene regulation          |
| Enhancer      | Expression regulation    |
| Conserved     | Evolutionarily important |
| DHS           | Open chromatin           |
| Histone Marks | Regulatory activity      |

Each SNP may belong to one or more annotation categories.

### Example Annotation Matrix

Suppose we have:

| SNP | Coding | Promoter | Conserved |
| --- | ------ | -------- | --------- |
| rs1 | 1      | 0        | 1         |
| rs2 | 0      | 1        | 0         |
| rs3 | 0      | 0        | 0         |

This annotation matrix becomes an additional source of information for the model.

### The Key Idea Behind SBayesRC

Recall SBayesR:

$
\beta_j
\sim
\sum_k
\pi_k
N(0,\gamma_k\sigma^2)
$

The prior probabilities are identical for every SNP.

SBayesRC modifies this assumption.

Instead, $\pi_k=f(Annotations)$.

The probability that a SNP belongs to a particular effect-size class now depends on its biological annotation.

#### Intuition

Suppose:

```text
Coding SNP
```

and

```text
Intergenic SNP
```

have identical GWAS evidence.

SBayesRC may assign:

```text
Coding SNP
↓
Higher Prior Probability
```

of having a non-zero effect.

This allows biological knowledge to influence statistical inference.

#### Why This Can Improve Prediction

GWAS data are noisy.

Many SNPs show weak signals.

Biological annotations provide an additional source of evidence.

Instead of relying solely on:

```text
GWAS Signal
```

SBayesRC uses:

```text
GWAS Signal
+
Biological Knowledge
```

This often improves estimation of SNP effects.

#### Heritability Enrichment

A central concept in SBayesRC is:

**Heritability Enrichment**

The idea is simple.

Some annotation categories explain disproportionately large amounts of heritability.

For example:

| Annotation | SNP Fraction | Heritability Fraction |
| ---------- | ------------ | --------------------- |
| Coding     | 2%           | 10%                   |
| Conserved  | 5%           | 20%                   |
| Intergenic | 50%          | 20%                   |

Coding regions contribute more heritability than expected by chance.

#### Measuring Enrichment

Enrichment is calculated as:

$Enrichment = \frac{
\text{Heritability Proportion}
}{
\text{SNP Proportion}
}$

Example: $\frac{0.10}{0.02}=5$.

An enrichment value of 5 indicates that the annotation contributes five times more heritability than expected.

### Why Enrichment Matters

Annotations with high enrichment are more likely to contain causal variants.

SBayesRC uses this information to prioritize SNPs during Bayesian inference.

Consequently:

```text
Highly Enriched Regions
↓
Higher Prior Probability
```

of contributing to disease risk.

#### The SBayesRC Workflow

The workflow is:

```text
GWAS Summary Statistics
          +
LD Matrix
          +
Functional Annotations
          ↓
        SBayesRC
          ↓
 Posterior SNP Effects
          ↓
          PRS
```

Compared with SBayesR, the only additional input is:

```text
Functional Annotations
```

##### Types of Annotations Used

Large studies often use:

* BaselineLD annotations
* Coding regions
* Conserved regions
* Regulatory annotations
* Histone modifications
* Gene-expression annotations

Modern implementations may include dozens or even hundreds of annotation categories.

#### Why SBayesRC Often Outperforms SBayesR?

SBayesR learns from:

```text
Statistical Evidence
```

SBayesRC learns from:

```text
Statistical Evidence
+
Biological Evidence
```

This additional information frequently improves prediction accuracy.

The improvement is often modest but consistent across many traits.

#### Real-World Interpretation

Imagine two SNPs with identical GWAS significance.

Traditional models see:

```text
Equal Evidence
```

SBayesRC sees:

```text
Equal Statistical Evidence
+
Different Biological Plausibility
```

The model can therefore make more informed decisions.

#### Limitations

Despite its advantages, SBayesRC is not perfect.

#### Annotation Quality Matters

Poor annotations can reduce performance.

#### Computational Complexity

More parameters must be estimated.

#### Population Differences

Annotations may not behave identically across ancestries.

#### Incomplete Biology

Current annotations capture only a fraction of biological complexity.

#### SBayesRC vs SBayesR

| Feature                   | SBayesR | SBayesRC |
| ------------------------- | ------- | -------- |
| Summary Statistics        | Yes     | Yes      |
| LD Modeling               | Yes     | Yes      |
| Mixture Priors            | Yes     | Yes      |
| Functional Annotations    | No      | Yes      |
| Heritability Enrichment   | No      | Yes      |
| Biological Prioritization | No      | Yes      |

### Key Takeaways

* SBayesRC extends SBayesR by incorporating functional annotations.
* Not all SNPs are biologically equivalent.
* Annotation information modifies prior probabilities of SNP effects.
* Heritability enrichment identifies biologically important genomic regions.
* SBayesRC combines statistical and biological evidence.
* Functional annotations often improve prediction accuracy.
* SBayesRC represents one of the most advanced PRS methods currently available.

### Interpreting Polygenic Risk Scores: What Do They Really Mean?

After spending several chapters learning how Polygenic Risk Scores (PRS) are constructed, optimized, and evaluated, we now arrive at perhaps the most important question:

> What does a Polygenic Risk Score actually mean?

This question is surprisingly difficult.

Many people assume that a high PRS means:

```text 
You will get the disease.
```

or that a low PRS means:

```text 
You are protected.
```

Neither statement is correct.

A Polygenic Risk Score is fundamentally a **statistical predictor**, not a diagnosis.

Understanding what PRS can and cannot tell us is essential for responsible use in research, medicine, and public health.

### What a PRS Measures

Recall the standard formula:

$PRS_i=\sum_j
\hat{\beta}*j x*{ij}$

A PRS combines information across many genetic variants into a single number.

The resulting score represents:

> An individual's relative genetic predisposition compared with others in the same population.

Notice the word:

**relative**

A PRS does not directly measure disease.

Instead, it measures inherited genetic burden.

### An Analogy: Height Prediction

Imagine we create a PRS for height.

Suppose:

| Person | PRS  |
| ------ | ---- |
| A      | -2.1 |
| B      | 0.0  |
| C      | +2.4 |

Would we know their exact heights?

No.

However, we might predict:

```text 
Person C
↓
Likely Taller
```

than

```text 
Person A
```

The PRS provides information about relative position within the population.

The same logic applies to disease risk.

### Relative Risk Versus Absolute Risk

One of the most important distinctions in genetic prediction is:

#### Relative Risk

versus

#### Absolute Risk

These terms are often confused.

### Relative Risk

Suppose the average disease risk is:

$$
10%
$$

A PRS study reports:

```text id="iwjlwm"
Top PRS Decile
↓
3× Higher Risk
```

This means:

$10% \times 3=30%$

The individual's relative risk increased threefold.

### Absolute Risk

The actual probability becomes:

$$
30%
$$

Notice:

```text 
30%
≠
100%
```

Even high-risk individuals may never develop the disease.

This distinction is crucial when communicating PRS results.

#### PRS Percentiles

Raw PRS values are difficult to interpret.

Example:

| Individual | PRS   |
| ---------- | ----- |
| A          | -0.78 |
| B          | 0.13  |
| C          | 1.42  |

What does:

```text 
PRS = 1.42
```

actually mean?

By itself:

```text 
Almost Nothing
```

Interpretation becomes easier using percentiles.

#### Understanding Percentiles

Suppose a population contains:

```text 
100,000 individuals
```

A PRS percentile represents position within the distribution.

Examples:

| Percentile | Meaning                   |
| ---------- | ------------------------- |
| 1st        | Lower than 99% of people  |
| 50th       | Population average        |
| 90th       | Higher than 90% of people |
| 99th       | Higher than 99% of people |

Percentiles are often far more intuitive than raw scores.

#### Visualizing a PRS Distribution

Most PRS follow an approximately normal distribution.

```text 
                *
              *****
            *********
          *************
--------|-----50%-----|--------
```

Most individuals cluster near the middle.

Extreme scores occur less frequently.

#### Risk Stratification

A common strategy is to divide individuals into groups.

Example:

| Group      | PRS Range      |
| ---------- | -------------- |
| Bottom 10% | Lowest scores  |
| Middle 80% | Average scores |
| Top 10%    | Highest scores |

Researchers then compare disease prevalence.

#### Example: Coronary Artery Disease

Suppose:

| Group      | Disease Rate |
| ---------- | ------------ |
| Bottom 10% | 3%           |
| Average    | 7%           |
| Top 10%    | 18%          |

Interpretation:

```text id="0m4wjv"
Higher PRS
↓
Higher Disease Frequency
```

However:

```text 
Most High-PRS Individuals
Still Do Not Develop Disease
```

#### Odds Ratios

Risk differences are often summarized using odds ratios.

Suppose:

| Group         | Odds Ratio |
| ------------- | ---------- |
| Bottom Decile | 1.0        |
| Top Decile    | 3.5        |

Interpretation:

Individuals in the highest PRS group have:

$
3.5
$

times greater disease odds than individuals in the lowest group.

Notice:

```text 
Odds
≠
Probability
```

These concepts are related but not identical.

#### PRS Is Not Destiny

One of the biggest misconceptions about genetic prediction is:

> Genes determine everything.

This is false.

Most phenotypes arise from:

$Phenotype=

Genetics
+
Environment
+
Random Effects
$

Even highly heritable diseases are influenced by non-genetic factors.

### Example: Type 2 Diabetes

Genetic risk matters.

However:

* Diet
* Physical activity
* Obesity
* Smoking
* Sleep

also contribute substantially.

A person with:

```text 
High PRS
```

may never develop disease.

Conversely:

```text 
Low PRS
```

does not guarantee protection.

### Heritability Places an Upper Limit

Recall from earlier chapters:

Prediction accuracy is constrained by heritability.

Suppose:

$$
h^2 = 0.40
$$

Only 40% of phenotypic variation is genetic.

Even a perfect PRS cannot explain:

$$
100%
$$

of disease risk.

This limitation is fundamental.

### What Does a High PRS Actually Mean?

A high PRS generally suggests:

* Increased inherited susceptibility
* Elevated relative risk
* Greater likelihood of disease compared with the population average

It does not imply:

* Certainty of disease
* Immediate clinical diagnosis
* Presence of symptoms

These distinctions are essential.

### Clinical Interpretation

Researchers are increasingly interested in clinical applications.

Potential uses include:

#### Early Screening

Identify individuals at elevated risk before symptoms appear.

#### Preventive Medicine

Target interventions toward high-risk individuals.

#### Risk Communication

Combine genetic and environmental information.

### Personalized Medicine

Adjust treatment strategies according to risk profiles.

However, clinical implementation remains an active area of research.

### Population Differences

One of the most important limitations of PRS is portability.

Suppose a PRS is trained using:

```text 
European GWAS
```

and applied to:

```text 
African Population
```

Prediction accuracy often decreases substantially.

### Why Does This Happen?

Several factors contribute:

* Different allele frequencies
* Different LD patterns
* Different environmental exposures
* Different genetic architectures

As a result:

```text 
Same PRS
↓
Different Performance
Across Populations
```

This remains one of the major challenges in human genetics.

Clinical decisions should also consider:


rather than relying on genetics alone.

Such integrated models often outperform PRS-only approaches.

### Key Takeaways

* A PRS measures relative genetic predisposition.
* PRS is not a diagnosis.
* Relative risk and absolute risk are different concepts.
* Percentiles often provide the most intuitive interpretation.
* High PRS does not guarantee disease.
* Low PRS does not guarantee protection.
* Environmental factors remain critically important.
* Prediction accuracy is limited by heritability.
* PRS performance often varies across populations.
* Clinical use requires careful interpretation and communication.

#### Summary

Polygenic Risk Scores are powerful tools for summarizing inherited genetic risk.

However, interpretation requires caution.

A PRS is best viewed as a probabilistic indicator of susceptibility rather than a deterministic prediction of future outcomes.

Understanding this distinction is essential for researchers, clinicians, and individuals who may eventually use PRS in real-world settings.



### The Future of Polygenic Risk Scores: Multi-Ancestry Prediction, Functional Genomics, and Precision Medicine

Over the course of this tutorial series, we have journeyed from the basic concept of Polygenic Risk Scores (PRS) to some of the most advanced prediction methods currently available, including SBayesR and SBayesRC.

We have learned:

* How GWAS identifies genetic associations
* How Polygenic Risk Scores are constructed
* Why linkage disequilibrium matters
* How Bayesian methods improve prediction
* The role of functional annotations
* How PRS should be interpreted

The natural question now becomes:

> Where is the field heading next?

Despite remarkable progress, today's PRS methods are still far from perfect.

Several major challenges remain:

* Limited portability across ancestries
* Missing heritability
* Gene-environment interactions
* Rare variants
* Clinical implementation

These challenges are driving the next generation of statistical genetics research.

#### The Current State of Polygenic Prediction

Today's best-performing PRS methods can achieve impressive prediction accuracy for many traits.

Examples include:

| Trait                   | Typical Predictive Performance |
| ----------------------- | ------------------------------ |
| Height                  | High                           |
| BMI                     | Moderate                       |
| Coronary Artery Disease | Moderate                       |
| Type 2 Diabetes         | Moderate                       |
| Schizophrenia           | Moderate                       |
| Depression              | Lower                          |

However, prediction accuracy varies substantially across traits and populations.

The reasons behind these differences provide important clues about future developments.

#### The Multi-Ancestry Problem

One of the largest challenges in modern statistical genetics is ancestry portability.

Most large GWAS have been conducted in individuals of European ancestry.

Consequently:

```text id="e7hr0f"
European GWAS
        ↓
European PRS
```

often performs well.

However:

```text id="wjlwm8"
European GWAS
        ↓
African Population
```

usually results in reduced prediction accuracy.

#### Why Does Portability Decline?

Several biological factors contribute.

#### Allele Frequencies Differ

A SNP common in Europeans may be rare in Africans.

Example:

| SNP | European Frequency | African Frequency |
| --- | ------------------ | ----------------- |
| rs1 | 40%                | 5%                |

Prediction performance can therefore change dramatically.

#### Different LD Patterns

Linkage Disequilibrium structures vary between populations.

For example:

```text id="35j0t7"
European LD
rs1 ↔ rs2
```

may become:

```text id="jlwm24"
African LD
rs1   rs2
```

with much weaker correlation.

Methods relying on LD therefore lose predictive power.

#### Environmental Differences

Environmental exposures differ across populations.

Examples include:

* Diet
* Healthcare access
* Education
* Socioeconomic conditions

Consequently:

```text id="bd2v2j"
Same Genetics
+
Different Environment
=
Different Outcomes
```

#### Multi-Ancestry PRS

Researchers are increasingly developing:

**Multi-Ancestry PRS**

These methods combine data from multiple ancestral groups.

Goal:

```text id="qux0lw"
More Diverse GWAS
        ↓
Better Generalization
```

Modern approaches attempt to model:

* Shared genetic effects
* Population-specific effects
* Shared biological pathways

simultaneously.

#### Larger GWAS Sample Sizes

Prediction accuracy is strongly influenced by discovery sample size.

As GWAS datasets continue to grow:

```text id="3rl5zs"
More Samples
        ↓
More Accurate Effect Estimates
        ↓
Better PRS
```

Biobanks now contain:

| Resource                | Approximate Size   |
| ----------------------- | ------------------ |
| UK Biobank              | ~500,000           |
| All of Us               | ~1,000,000 planned |
| FinnGen                 | ~500,000           |
| Million Veteran Program | >1,000,000         |

Future datasets may include tens of millions of individuals.

#### Rare Variants

Most current PRS focus on common variants.

However:

```text id="jlwm4t"
Rare Variants
```

may have larger biological effects.

Traditional GWAS often struggles to identify them because:

* They are uncommon
* Statistical power is limited

Future PRS models will likely integrate:

* Common variants
* Rare variants
* Structural variants

into a unified framework.

#### Functional Genomics

Current PRS methods increasingly use biological annotations.

However, today's annotations remain incomplete.

Future prediction models may incorporate:

#### Gene Expression

eQTL information.

#### Chromatin Accessibility

Open chromatin regions.

#### Single-Cell Data

Cell-type specific effects.

#### Epigenetics

DNA methylation and histone modifications.

The future model may look like:

```text id="jlwm61"
GWAS
+
Expression
+
Epigenetics
+
Single Cell Data
        ↓
Enhanced PRS
```

#### Transcriptome-Wide Prediction

One exciting direction involves integrating genetically predicted gene expression.

Examples include:

* TWAS
* PrediXcan
* FUSION

Instead of predicting disease directly from SNPs:

```text id="jlwm72"
SNPs
↓
Gene Expression
↓
Disease
```

These methods may provide more biologically interpretable predictions.

#### Gene-Environment Interactions

Traditional PRS models assume:

```text id="jlwm83"
Genetics
+
Environment
```

contribute independently.

Reality is often more complex.

Example:

A genetic predisposition for obesity may be amplified by:

* High-calorie diets
* Sedentary lifestyle

but reduced by:

* Exercise
* Healthy nutrition

This phenomenon is known as:

**Gene-Environment Interaction (G×E)**

#### Future Prediction Models

Future risk prediction may incorporate:

```text id="jlwm94"
Genetics
+
Environment
+
Lifestyle
+
Biomarkers
+
Medical History
```

rather than relying on genetics alone.

Such integrated models may substantially outperform PRS-only approaches.

### Machine Learning and Deep Learning

Another rapidly growing area involves artificial intelligence.

Researchers are investigating:

* Neural networks
* Graph neural networks
* Transformer architectures
* Deep representation learning

for genomic prediction.

Potential advantages:

* Capture nonlinear effects
* Model complex interactions
* Integrate multiple data types

However:

Current evidence suggests that for many PRS applications:

```text id="jlwma5"
Better Data
>
More Complex Models
```

Thus, data quality often matters more than model complexity.
