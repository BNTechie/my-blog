‘awk’ is a text-processing language that is particularly well-suited for working with structured text files. It is commonly used to:

filter rows based on conditions
extract and manipulate specific columns
transform text into a desired format
perform simple calculations directly from the command line
It operates line by line, automatically splitting each line into fields, which makes it very effective for handling tabular data such as CSV or TSV files.

In practical terms, awk combines ideas from different tools: it allows conditional filtering similar to SQL queries, supports basic programming logic like Python, and can perform quick calculations and transformations often done in spreadsheet software. All of these can be done directly in the terminal without loading the data into a separate environment.


### Why awk is Essential?
It works directly on very large files (from gigabytes to terabytes)
It does not require loading data into memory
It is fast and efficient for line-by-line processing
It is well-suited for preprocessing data before machine learning pipeline.

### Basic Syntax

The basic structure of an awk command consists of a pattern and an action. The pattern acts as a condition that determines which lines to process, while the action defines what should be done with those lines. If no pattern is provided, the action is applied to every line in the file. In the simplest case, awk '{print}' file.txt prints each line as it is, which is equivalent to displaying the full contents of the file. This simple structure forms the foundation for all more advanced awk operations.


```bash
awk 'pattern { action }' file
pattern → condition (optional)
action → what to do
Simplest Example
awk '{print}' file.txt
Prints the entire file.
```


### Understanding Fields and Columns

By default, awk treats each line as a record and automatically splits it into fields based on whitespace (spaces or tabs). Each field can then be accessed using positional variables,

such as $1 for the first column, $2 for the second column, and $0 for the entire line. This makes it very convenient to work with structured text data. For example, if a file contains names and ages separated by spaces, using awk '{print $1}' file.txt will extract and print only the first column, which in this case corresponds to the names.

### Example File

Alice 25

Bob 30

Charlie 28


#### Print First Column


```bash
awk '{print $1}' file.txt
Output:

Alice
Bob
Charlie
```


#### Setting Delimiters
In real-world datasets, fields are not always separated by spaces. Instead, they are often structured using specific delimiters such as commas in CSV files or tabs in TSV files. In such cases, awk needs to be told how to split each line using the -F option. For example, -F "," is used for comma-separated values, while -F "\t" is used for tab-separated files, which are very common in genetics and bioinformatics. Setting the correct delimiter ensures that each column is interpreted properly and avoids incorrect data extraction.

##### CSV Example
    
awk -F "," '{print $1, $3}' data.csv

#### TSV Example

```bash
awk -F "\t" '{print $1, $2}' file.txt
```


##### Row Filtering
One of the most useful features of awk is its ability to filter rows based on conditions. These conditions are written directly inside the command and are applied to each line of the file. You can filter numeric values, match exact text, combine multiple conditions using logical operators, or even use pattern matching with regular expressions. This makes awk very flexible for selecting only the relevant subset of data without modifying the original file.

#### Filter Numeric Values

```bash
awk '$2 > 25' file.txt
```

#### Filter Exact Match
```bash
awk '$1 == "Alice"' file.txt
```

#### Multiple Conditions

```bash
awk '$2 > 25 && $2 < 35' file.txt
```

#### String Matching

```bash
awk '$1 ~ /A/' file.txt
```

##### Calculations with awk

In addition to filtering and extracting data, awk can also perform basic numerical calculations directly from the command line. Since it processes data line by line, it is straightforward to apply arithmetic operations to specific columns. For example, you can multiply values in a column, compute the sum of all entries, or calculate averages without needing any external tools.

##### Multiply Values

```bash
awk '{print $1, $2*2}' file.txt
```

##### Sum Column
```bash
awk '{sum += $2} END {print sum}' file.txt
```
##### Average

```bash
awk '{sum += $2} END {print sum/NR}' file.txt
```


The END block is useful for calculations that require processing the entire file, such as totals or averages. It runs after all lines have been read, allowing you to output final results based on accumulated values.

#### BEGIN and END Blocks


awk provides two special sections called BEGIN and END that allow you to run commands before and after processing the input file. The BEGIN block is executed once before any lines are read, making it useful for initializing variables or printing headers. The END block runs after all lines have been processed, which is helpful for summarizing results or printing final outputs. These blocks are especially useful when you need to combine line-by-line processing with overall file-level operations.




##### Example

```bash
awk '
BEGIN {print "Start"}
{print $1}
END {print "End"}
' file.txt
```

#### Formatting Output

While the default print statement is convenient, printf gives you more control over how the output is displayed. It allows you to define specific formats for text and numbers, similar to formatting in programming languages like C or Python. This is particularly useful when you want clean, structured output, such as aligned columns or descriptive text alongside values.

#### Using printf

```bash
awk '{printf "%s is %d years old\n", $1, $2}' file.txt
```

#### Working with Text (String Functions)

awk provides built-in functions for handling and transforming text. These functions are useful for cleaning and standardizing data during processing. For example, you can replace specific words or patterns using gsub, or change the case of text using functions like toupper and tolower. These operations are applied line by line, making them efficient even for large text files.



#### Replace Text
```bash
awk '{gsub("Alice", "Alicia"); print}' file.txt
```

#### Convert to Uppercase
```bash
awk '{print toupper($1)}' file.txt
```

#### Convert to Lowercase
```bash
awk '{print tolower($1)}' file.txt
```
#### Arrays in awk
awk supports associative arrays, which allow you to store and count values dynamically without predefining their size. These arrays are useful for counting frequencies or grouping data. In the example below, each unique value in the first column is used as a key, and its count is incremented as awk processes each line. The final result is printed in the END block, showing how many times each value appears in the file.

#### Count Frequency

```bash
awk '{count[$1]++} END {for (name in count) print name, count[name]}' file.txt
```

#### Real Use Case: Genetics Data

awk is useful in genetics and bioinformatics, where many file formats are plain text and tab-separated. Files such as VCF, annotation tables, or summary statistics often contain structured columns that can be filtered or extracted directly from the terminal. In these cases, awk helps with routine tasks such as selecting chromosome positions, filtering variants based on quality, or retrieving information from annotation fields. This is especially valuable when working with large genomics files, where quick command-line inspection is often more efficient than opening the data in another environment.



##### Example VCF Line

```bash
chr1 12345 rs123 A G . PASS AF=0.45
```

##### Extract Key Columns

```bash
awk -F "\t" '{print $1, $2, $3}' file.vcf
```

#### Filter PASS Variants
Some genetics file formats, especially VCF, contain additional information packed into a single column such as the INFO field. This field often stores multiple key-value pairs separated by semicolons. awk can be used to split this field further and extract specific values such as allele frequency.


```bash
awk '$7 == "PASS"' file.vcf
Parse INFO Field
awk -F "\t" '{
  split($8, info, ";");
  for (i in info) {
    if (info[i] ~ /^AF=/) {
      split(info[i], a, "=");
      print $1, $2, a[2];
    }
  }
}' file.vcf
```

#### Filter Rare Variants

A common genetics task is to identify rare variants based on allele frequency thresholds. Once a value such as allele frequency has been extracted from the INFO field, awk can be used to apply numerical filtering rules directly. This makes it possible to screen variants quickly without needing a separate script. For exploratory work or quick checks in a pipeline, this can be a very efficient way to focus only on the subset of variants relevant to the analysis.


```bash
awk -F "\t" '{
  split($8, info, ";");
  for (i in info) {
    if (info[i] ~ /^AF=/) {
      split(info[i], a, "=");
      if (a[2] < 0.01) print $0;
    }
  }
}' file.vcf
```



#### Combining awk with Other Commands

One of the strengths of awk is how easily it integrates with other command-line tools through pipelines. You can use tools like grep to filter specific lines and then pass the result to awk for further processing, such as extracting columns. Similarly, combining awk with commands like sort and uniq allows you to perform more advanced operations like counting and ranking values. This modular approach makes it possible to build powerful data-processing workflows using simple, reusable commands.



#### Example
```bash
grep "ERROR" logfile.txt | awk '{print $1, $5}'
Sort + Count
awk '{print $1}' file.txt | sort | uniq -c | sort -nr
```
