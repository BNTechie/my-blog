

# Part 1: Why Heritability? Understanding the Origins of Individual Differences

## Introduction

Why are people different from one another?

Why are some individuals taller, stronger, more intelligent, or more susceptible to diseases than others?

Why do some siblings resemble each other strongly, while others seem remarkably different?

These questions have fascinated scientists for over a century. Modern statistical genetics attempts to answer these questions by partitioning the observed variation in a trait into genetic and environmental components.

The central concept behind this endeavor is **heritability**.

In this tutorial, we will start from the basic concepts and gradually build toward Structural Equation Models (SEM), ACE models, and genomic heritability.

::: {.callout-note}
Heritability is one of the most frequently misunderstood concepts in genetics. It describes variation within a population, not the extent to which a trait is "genetic" for an individual.
:::

---

## What Is Heritability?

Heritability measures how much of the observed variation in a trait is attributable to genetic differences among individuals.

Importantly:

> Heritability is a property of a population, not an individual.

A common misunderstanding is that a heritability of 70% means that 70% of a person's trait is genetic.

This is incorrect.

Instead, a heritability estimate of 70% means:

> 70% of the variation observed across individuals in a population can be statistically attributed to genetic differences.

---

## Variance Is the Key

Heritability is fundamentally about explaining **variance**.

Suppose we measure height in 10,000 individuals.

Some individuals are:

- Very short
- Average height
- Very tall

The spread of these measurements is called the **phenotypic variance**.

We denote this as:

$$
V_P
$$

where:

- $V$ = variance
- $P$ = phenotype

---

## Partitioning Phenotypic Variance

One of the central ideas in quantitative genetics is that phenotypic variance can be decomposed into genetic and environmental components.

$$
V_P = V_G + V_E
$$

where:

| Component | Meaning |
|-----------|-----------|
| $V_P$ | Phenotypic variance |
| $V_G$ | Genetic variance |
| $V_E$ | Environmental variance |

This simple equation forms the foundation of quantitative genetics.

---

## A Simple Example

Imagine we observe the following heights:

```text
160 cm
165 cm
170 cm
175 cm
180 cm
```

The mean height is:

$$
170 \text{ cm}
$$

The variance measures how much individuals deviate from this average.

Now imagine:

- Genetics explains half of this variation
- Environment explains the other half

Then:

$$
V_G = 0.5
$$

$$
V_E = 0.5
$$

and therefore:

$$
V_P = 1.0
$$

---

## Defining Heritability

Heritability is defined as:

$$
h^2 = \frac{V_G}{V_P}
$$

This quantity ranges between:

$$
0 \le h^2 \le 1
$$

or equivalently:

$$
0\% \le h^2 \le 100\%
$$

---

## Interpretation

Suppose:

$$
h^2 = 0.80
$$

Interpretation:

> 80% of the observed variation in the trait is attributable to genetic differences among individuals.

The remaining:

$$
20\%
$$

is attributable to environmental factors.

---

## Important Caveat

High heritability does **not** imply immutability.

For example, human height has a heritability of approximately:

$$
h^2 \approx 0.80
$$

Yet average height increased substantially during the 20th century due to improvements in:

- Nutrition
- Healthcare
- Sanitation
- Living conditions

Environmental changes can strongly influence highly heritable traits.

::: {.callout-important}
High heritability does not mean environmental interventions are ineffective.
:::

---

## Why Heritability Depends on the Population

Consider two populations.

### Population A

Everyone has:

- Similar diet
- Similar healthcare
- Similar lifestyle

Environmental variation is small.

Therefore:

$$
V_E
$$

is relatively small.

Heritability tends to be higher.

### Population B

Individuals experience:

- Different diets
- Unequal healthcare
- Different socioeconomic conditions

Environmental variation is larger.

Therefore:

$$
V_E
$$

increases.

Heritability decreases.

---

## Heritability Is Context-Specific

This means:

> The same trait can have different heritability estimates in different populations.

Heritability depends on:

- Geography
- Historical period
- Age
- Ancestry
- Environmental exposure

This is why heritability estimates should always be interpreted in the context of the population being studied.

---

## Genetic vs Environmental Contributions

A useful conceptual diagram is:

```text
Phenotypic Variation
        │
        ├── Genetic
        │
        └── Environmental
```

As we will see later, both genetic and environmental influences can themselves be partitioned into multiple components.

---

## Why Study Heritability?

Estimating heritability helps answer important scientific questions.

### Medicine

How much of disease risk is genetic?

Examples:

- Schizophrenia
- Depression
- Alzheimer's disease
- Type 2 diabetes

### Psychology

How much variation in cognitive ability is genetic?

How important is family environment?

### Public Health

Can interventions reduce disease risk even for highly heritable traits?

### Agriculture

Which plant or animal traits should be selected in breeding programs?

---

## Historical Origins

Modern heritability theory originates from the work of Ronald A. Fisher.

In 1918, Fisher showed how Mendelian inheritance could explain continuously distributed traits such as height.

This work laid the foundation for modern quantitative genetics.

---

## Why Twins Are Special

Monozygotic (MZ) twins share:

```text
100% of their DNA
```

Dizygotic (DZ) twins share approximately:

```text
50% of their segregating DNA
```

Yet both typically grow up in similar family environments.

This creates a powerful framework for separating genetic and environmental influences.

---

## The Road Ahead

Throughout this tutorial we will build the following framework:

```text
Phenotypic Variance
        ↓
A, C, and E Components
        ↓
Twin Correlations
        ↓
Falconer's Formula
        ↓
Structural Equation Models
        ↓
ACE Models
        ↓
OpenMx
        ↓
Genetic Relatedness
        ↓
GRM
        ↓
SNP Heritability
```

---

## Key Takeaways

- Heritability measures the proportion of variance attributable to genetic differences.
- Heritability is a population-level quantity.
- Heritability does not describe how genetic a trait is for an individual.
- High heritability does not imply environmental interventions are ineffective.
- Phenotypic variance can be partitioned into genetic and environmental components.
- Twin studies provide one of the most powerful approaches for estimating heritability.
- Understanding variance decomposition is the foundation for ACE models and modern genomic analyses.

# Part 2: Sources of Variance — Understanding Additive Genetics (A), Shared Environment (C), Unique Environment (E), and Dominance (D)

## Introduction

In Part 1, we learned that phenotypic variance can be partitioned into genetic and environmental components:

$$
V_P = V_G + V_E
$$

This is an important starting point, but it is still too simplistic.

Neither genetic variance nor environmental variance is a single entity.

Instead, both can be decomposed into multiple components, each representing a different biological mechanism.

The classical twin design and Structural Equation Models (SEM) attempt to estimate these components separately.

Understanding these components is the foundation of:

- Twin studies
- ACE models
- ADE models
- OpenMx analyses
- SNP heritability
- Modern quantitative genetics

---

## Decomposing Phenotypic Variance

A more complete model is:

$$
V_P = V_A + V_D + V_C + V_E
$$

where:

| Component | Meaning |
|------------|----------|
| $V_A$ | Additive genetic variance |
| $V_D$ | Dominance genetic variance |
| $V_C$ | Shared environmental variance |
| $V_E$ | Unique environmental variance |

This decomposition appears repeatedly throughout twin modeling and quantitative genetics.

---

## A Useful Analogy

Imagine baking a cake.

The final cake represents:

$$
V_P
$$

The ingredients represent:

- Genetics
- Family environment
- Personal experiences

Different ingredients contribute differently to the final outcome.

Similarly, phenotypic variation arises from multiple sources.

---

# Additive Genetic Effects (A)

## What Does "Additive" Mean?

Additive genetics refers to the cumulative effect of individual alleles.

Each allele contributes a small amount to the phenotype.

These contributions simply add together.

Hence the name:

> Additive Genetic Effects

---

## Example

Suppose a height SNP has two alleles:

```text
A
a
```

Assume:

| Genotype | Effect |
|-----------|----------|
| aa | 0 cm |
| Aa | +2 cm |
| AA | +4 cm |

Notice:

$$
AA = 2 \times Aa
$$

Each allele contributes independently.

This is additive inheritance.

---

## Why Additive Effects Are Important

Most modern GWAS estimate additive effects.

When we fit:

$$
Y = \beta_0 + \beta_1 SNP + \epsilon
$$

the coefficient:

$$
\beta_1
$$

typically represents an additive effect.

---

## Polygenic Traits

Most complex traits are highly polygenic.

Examples include:

- Height
- BMI
- Intelligence
- Educational attainment
- Schizophrenia risk

Thousands of variants contribute tiny additive effects.

The sum of these effects forms:

$$
V_A
$$

---

## Visualizing Additive Effects

```text
SNP 1
   ↓
SNP 2
   ↓
SNP 3
   ↓
SNP 4
   ↓

Total Effect
```

Each SNP contributes a small amount.

The effects accumulate.

---

# Dominance Effects (D)

## What Is Dominance?

Dominance occurs when one allele masks the effect of another.

The heterozygote is no longer intermediate.

---

## Example

Consider:

| Genotype | Effect |
|-----------|----------|
| aa | 0 |
| Aa | 5 |
| AA | 5 |

The heterozygote behaves like:

```text
AA
```

rather than lying halfway between:

```text
aa and AA
```

This represents complete dominance.

---

## Why Dominance Matters

Dominance contributes to:

$$
V_D
$$

the dominance variance.

Unlike additive effects:

- dominance cannot be easily transmitted across generations
- dominance contributes less to long-term response to selection

---

## Additive vs Dominance

### Additive

```text
aa = 0
Aa = 2
AA = 4
```

Linear progression.

### Dominance

```text
aa = 0
Aa = 4
AA = 4
```

Non-linear progression.

---

## Visualization

### Additive

```text
0 ---- 2 ---- 4
```

### Dominance

```text
0 -------- 4
          /
         /
        4
```

The heterozygote deviates from the additive expectation.

---

# Shared Environment (C)

## What Is Shared Environment?

Shared environmental effects make family members more similar.

These are influences that individuals living in the same family experience together.

Examples:

- Household income
- Parenting style
- Neighborhood
- Diet
- Family culture
- School quality

---

## Examples

Suppose two siblings:

- Eat the same food
- Attend the same school
- Live in the same home

These common experiences increase similarity.

The variance attributable to these influences is:

$$
V_C
$$

---

## Why Twin Studies Care About C

Monozygotic twins and dizygotic twins often share similar environments.

If:

```text
MZ twins
```

and

```text
DZ twins
```

are similarly correlated,

then shared environment may explain much of the resemblance.

---

## Interpretation

Large:

$$
V_C
$$

suggests family environment strongly influences the trait.

Examples where C may matter:

- Childhood behavior
- Educational outcomes
- Language acquisition

---

# Unique Environment (E)

## What Is Unique Environment?

Unique environmental effects make family members different.

Even identical twins experience unique events.

Examples:

- Different friends
- Different illnesses
- Different teachers
- Different accidents
- Different life experiences

---

## Why E Is Always Present

Even monozygotic twins are never perfectly identical.

If:

$$
r_{MZ} < 1
$$

then:

$$
V_E > 0
$$

must exist.

---

## Measurement Error Belongs to E

This is one of the most important concepts in twin studies.

Measurement error is included in:

$$
V_E
$$

because errors are unique to each individual measurement.

Therefore:

$$
E =
Unique Environment
+
Measurement Error
$$

---

## Example

Suppose height is measured incorrectly.

One twin's height is recorded as:

```text
170 cm
```

instead of:

```text
172 cm
```

This contributes to:

$$
V_E
$$

not:

$$
V_A
$$

or

$$
V_C
$$

---

# Putting Everything Together

The most complete decomposition is:

$$
V_P =
V_A
+
V_D
+
V_C
+
V_E
$$

or equivalently:

$$
P = A + D + C + E
$$

---

# The ACE Model

Most classical twin studies estimate:

$$
P = A + C + E
$$

The corresponding variance decomposition is:

$$
V_P
=
V_A
+
V_C
+
V_E
$$

---

## Why No D?

Because:

- C and D cannot be estimated simultaneously using only MZ and DZ twins.

The classical twin design provides only limited information.

Researchers must choose:

### ACE

or

### ADE

but generally not both.

---

# The ADE Model

An alternative model is:

$$
V_P
=
V_A
+
V_D
+
V_E
$$

This is used when dominance appears important.

---

# Choosing Between ACE and ADE

The decision depends on twin correlations.

This is one of the most elegant ideas in classical twin analysis.

We compare:

$$
r_{MZ}
$$

and

$$
r_{DZ}
$$

to determine which components are likely important.

We will study this in detail in Part 3.

---

```python
import numpy as np
```


---

# Why These Components Matter

Everything in the rest of the tutorial depends on these ideas.

The ACE model is simply a formal statistical model that estimates:

$$
V_A
$$

$$
V_C
$$

$$
V_E
$$

from twin data.

The ADE model estimates:

$$
V_A
$$

$$
V_D
$$

$$
V_E
$$

instead.

---

## Key Takeaways

- Phenotypic variance can be partitioned into multiple components.
- Additive effects (A) represent cumulative allele effects.
- Dominance effects (D) represent non-additive allele interactions.
- Shared environment (C) makes family members more similar.
- Unique environment (E) makes family members different.
- Measurement error is included in E.
- ACE models estimate A, C, and E.
- ADE models estimate A, D, and E.
- Understanding these components is essential before studying twin correlations.

# Part 3: Why Twins Are a Natural Experiment — Understanding Twin Correlations and the Logic Behind ACE Models

## Introduction

In Part 2, we learned that phenotypic variance can be decomposed into:

$$
V_P = V_A + V_C + V_E
$$

or, in some situations,

$$
V_P = V_A + V_D + V_E
$$

The next question is:

> How can we estimate these components from real data?

The answer lies in one of the most elegant designs in all of genetics:

## The Classical Twin Design

Twin studies provide a natural experiment that allows us to separate genetic and environmental influences.

The key insight is simple:

> Monozygotic twins share more genetic material than dizygotic twins, while both typically share similar family environments.

By comparing how similar these two types of twins are, we can infer how much variation is due to genetics and how much is due to environment.

---

# Why Twins Are Special

Most studies face a major challenge:

```text
Genetics and environment are correlated.
```

Children inherit genes from their parents.

They also grow up in environments created by those parents.

This makes it difficult to determine whether similarities arise from:

- Genetics
- Environment
- Both

Twin studies provide a way to disentangle these effects.

---

# Monozygotic (MZ) Twins

Monozygotic twins originate from a single fertilized egg.

The embryo splits into two individuals.

As a result:

```text
MZ twins share nearly 100% of their DNA.
```

Expected genetic relatedness:

$$
R_{MZ}=1.0
$$

---

# Dizygotic (DZ) Twins

Dizygotic twins originate from two separate eggs.

They are genetically equivalent to ordinary siblings.

On average:

```text
DZ twins share approximately 50% of their segregating DNA.
```

Expected relatedness:

$$
R_{DZ}=0.5
$$

---

# Why This Difference Matters

Consider a trait such as height.

If genetics strongly influence height:

- MZ twins should be very similar
- DZ twins should be less similar

because MZ twins share more genetic material.

If genetics do not matter:

- MZ twins should be no more similar than DZ twins

because both groups share similar family environments.

---

# The Fundamental Logic of Twin Studies

The entire classical twin design is based on one idea:

> Greater similarity among MZ twins compared with DZ twins implies genetic influence.

This idea has been used for more than a century to estimate heritability.

---

# Measuring Similarity

Twin similarity is usually measured using a correlation coefficient.

For MZ twins:

$$
r_{MZ}
$$

For DZ twins:

$$
r_{DZ}
$$

These correlations become the foundation of all subsequent analyses.

---

# What Does a Twin Correlation Mean?

Suppose:

$$
r_{MZ}=0.80
$$

Interpretation:

If one twin has a high value, the co-twin is also likely to have a high value.

Strong correlation implies strong similarity.

---

# Correlation Scale

| Correlation | Interpretation |
|------------|---------------|
| 0 | No similarity |
| 0.25 | Weak similarity |
| 0.50 | Moderate similarity |
| 0.75 | Strong similarity |
| 1.00 | Perfect similarity |

---

# Scenario 1: Pure Shared Environment

Suppose:

$$
r_{MZ}=0.80
$$

and

$$
r_{DZ}=0.80
$$

MZ and DZ twins are equally similar.

What does this imply?

Since MZ twins share twice as much DNA as DZ twins, yet are no more similar, genetics cannot explain the resemblance.

The likely explanation is:

$$
C
$$

Shared environmental influences.

---

## Visualization

```text
MZ = 0.80

DZ = 0.80
```

Interpretation:

```text
Environment dominates.
```

---

# Scenario 2: Pure Additive Genetics

Suppose:

$$
r_{MZ}=0.80
$$

and

$$
r_{DZ}=0.40
$$

Notice:

$$
r_{DZ}
=
\frac12 r_{MZ}
$$

This is exactly what we expect under purely additive inheritance.

Interpretation:

```text
Strong additive genetic influence.
```

---

## Visualization

```text
MZ = 0.80

DZ = 0.40
```

---

# Scenario 3: Shared Environment Present

Suppose:

$$
r_{MZ}=0.80
$$

and

$$
r_{DZ}=0.65
$$

Notice:

$$
r_{DZ}
>
\frac12 r_{MZ}
$$

DZ twins are more similar than expected under pure additive genetics.

Interpretation:

Shared environmental influences are increasing similarity.

Therefore:

$$
C > 0
$$

---

## Visualization

```text
MZ = 0.80

DZ = 0.65
```

---

# Scenario 4: Dominance Effects Present

Suppose:

$$
r_{MZ}=0.80
$$

and

$$
r_{DZ}=0.25
$$

Notice:

$$
r_{DZ}
<
\frac12 r_{MZ}
$$

DZ twins are less similar than expected.

This pattern suggests:

$$
D > 0
$$

Dominance genetic effects.

---

## Visualization

```text
MZ = 0.80

DZ = 0.25
```

---

# Summary of Twin Correlation Logic

| Pattern | Interpretation |
|-----------|---------------|
| $r_{MZ}=r_{DZ}$ | Shared environment |
| $r_{DZ}=\frac12 r_{MZ}$ | Additive genetics |
| $r_{DZ}>\frac12 r_{MZ}$ | Shared environment present |
| $r_{DZ}<\frac12 r_{MZ}$ | Dominance effects present |

This table is one of the most important concepts in classical twin analysis.

---

# Simulating Twin Correlations

Let's create a simple simulation.

```python
import numpy as np
import matplotlib.pyplot as plt

mz_corr = [0.80,0.80,0.80,0.80]
dz_corr = [0.80,0.40,0.65,0.25]

labels = [
    "Shared Env",
    "Additive",
    "ACE",
    "ADE"
]

x = np.arange(len(labels))

width = 0.35

plt.figure(figsize=(10,6))

plt.bar(
    x-width/2,
    mz_corr,
    width,
    label="MZ"
)

plt.bar(
    x+width/2,
    dz_corr,
    width,
    label="DZ"
)

plt.xticks(
    x,
    labels
)

plt.ylabel("Twin Correlation")

plt.title(
    "Twin Correlation Patterns"
)

plt.legend()

plt.show()
```

---

# Why MZ Correlations Are Never 1

Many beginners expect:

```text
MZ twins share 100% of DNA
```

therefore:

```text
rMZ = 1
```

for every trait.

This is incorrect.

Even identical twins experience:

- Different illnesses
- Different teachers
- Different friends
- Random developmental variation

These contribute to:

$$
E
$$

the unique environment.

Consequently:

$$
r_{MZ}<1
$$

for almost all traits.

---

# A Real Example

Suppose we observe:

$$
r_{MZ}=0.67
$$

$$
r_{DZ}=0.41
$$

These values are very close to those used in the workshop practical.

What do they suggest?

First compute:

$$
\frac12 r_{MZ}
=
0.335
$$

Since:

$$
0.41 > 0.335
$$

the DZ correlation exceeds half the MZ correlation.

This suggests:

```text
Shared environmental influences exist.
```

In the next section we will use these correlations to estimate:

- A
- C
- E

directly.

---

# Why Twin Correlations Are Powerful

Remarkably, these two numbers:

$$
r_{MZ}
$$

and

$$
r_{DZ}
$$

contain enough information to produce initial estimates of heritability.

This approach is known as:

## Falconer's Method

Developed long before modern Structural Equation Modeling.

Despite its simplicity, it often produces surprisingly accurate estimates.

---

# Assumptions Behind Twin Correlations

The interpretation of twin correlations relies on several assumptions.

Most importantly:

### Equal Environment Assumption

MZ and DZ twins experience equally similar environments.

If MZ twins receive systematically more similar treatment, genetic effects may be overestimated.

We will discuss these assumptions in detail later when building formal ACE models.

---

# Key Takeaways

- Twin studies compare MZ and DZ similarity.
- MZ twins share approximately 100% of their DNA.
- DZ twins share approximately 50% of their DNA.
- Greater MZ similarity implies genetic influence.
- Twin correlations form the basis of heritability estimation.
- The relationship between $r_{MZ}$ and $r_{DZ}$ reveals whether A, C, or D are likely present.
- MZ correlations are never perfect because unique environmental influences always exist.
- Twin correlations provide the foundation for Falconer's formulas and ACE models.

# Part 4: Falconer's Heritability — Estimating A, C, and E from Twin Correlations

## Introduction

In Part 3, we learned that the comparison of monozygotic (MZ) and dizygotic (DZ) twin correlations provides information about the relative importance of:

- Additive genetics (A)
- Shared environment (C)
- Unique environment (E)

Before modern Structural Equation Models (SEM) existed, geneticists used a remarkably simple approach to estimate these components.

This method is known as:

## Falconer's Formula

Despite being developed decades ago, Falconer's method remains one of the most intuitive ways to understand twin studies.

More importantly, it helps explain where ACE models come from.

---

# Historical Background

Falconer's equations were developed by:

:contentReference[oaicite:0]{index=0}

as a simple way to estimate genetic and environmental variance components using only:

$$
r_{MZ}
$$

and

$$
r_{DZ}
$$

the twin correlations.

The beauty of the method is that it requires no optimization, no likelihood functions, and no structural equation modeling.

Just arithmetic.

---

# The Logic Behind Falconer's Method

Recall:

### MZ Twins

Expected relatedness:

$$
R_{MZ}=1
$$

### DZ Twins

Expected relatedness:

$$
R_{DZ}=0.5
$$

Assuming an ACE model:

$$
P=A+C+E
$$

we can write expected twin correlations.

---

# Expected MZ Correlation

MZ twins share:

- 100% of additive genetic effects
- 100% of shared environment

Therefore:

$$
r_{MZ}=A+C
$$

---

# Expected DZ Correlation

DZ twins share:

- 50% of additive genetic effects
- 100% of shared environment

=

-

=

$$

$$

=

$$

$$

$$

=
2r_{DZ}
-r_{MZ}
$$

=

$$

# Why Not Fit an ACE Model Immediately?

# Likelihood Ratio Tests

Model comparison uses:

## Likelihood Ratio Testing (LRT)

We compare:

### Full Model

### Constrained Model

$$

=
-2LL_{small}
+

$$

Small p-value:

```text
Constraint hurts model fit
```

Do not impose it.

---

Large p-value:

```text
Constraint acceptable
```

Use simpler model.

# Part 7: Twin Model Assumptions — Equal Environments, Random Mating, and Other Critical Assumptions

## Introduction

In Parts 1–6, we learned how twin studies estimate:

- Additive genetic effects (A)
- Shared environmental effects (C)
- Unique environmental effects (E)

using differences in similarity between monozygotic (MZ) and dizygotic (DZ) twins.

However, there is an important caveat:

> Twin studies do not directly observe A, C, and E.

Instead, they infer these components from observed patterns in the data.

This inference is only valid if several assumptions hold.

Every statistical model relies on assumptions.

Twin studies are no exception.

Before interpreting heritability estimates, we must understand:

- What assumptions are being made?
- Why are they necessary?
- What happens if they are violated?

---

# Why Assumptions Matter

Suppose a twin study estimates:

$$
h^2 = 0.70
$$

Does this automatically mean:

```text
70% of the trait is genetic?
```

Not necessarily.

The estimate is only as good as the assumptions underlying the model.

If the assumptions fail:

- Heritability may be overestimated
- Heritability may be underestimated
- A, C, and E may be biased

Understanding assumptions is therefore essential for interpreting results correctly.

---

# The Major Assumptions of Classical Twin Studies

The most important assumptions are:

1. Equal Environment Assumption (EEA)
2. Random Mating
3. No Gene–Environment Correlation
4. No Gene–Environment Interaction
5. Same Population Assumption
6. Representative Sampling
7. Accurate Zygosity Classification

We will examine each in detail.

---

# Assumption 1: Equal Environment Assumption (EEA)

## What Is the Equal Environment Assumption?

The Equal Environment Assumption states:

> Environmental similarity is approximately the same for MZ and DZ twins.

This is arguably the most important assumption in twin research.

---

## Why Is It Needed?

Recall:

MZ twins share:

$$
100\%
$$

of their DNA.

DZ twins share:

$$
50\%
$$

of their DNA.

The twin method assumes:

```text
The major difference between MZ and DZ twins is genetic similarity.
```

If MZ twins also experience substantially more similar environments, then increased similarity could be mistakenly attributed to genetics.

---

## Example

Suppose parents:

- Dress MZ twins identically
- Give them identical toys
- Encourage identical activities

but do not treat DZ twins similarly.

Then:

```text
MZ twins become more similar
```

for environmental reasons.

The twin model might incorrectly interpret this as genetic influence.

---

## Visual Illustration

### Ideal Situation

```text
MZ Twins
   ↑
Same Environment

DZ Twins
   ↑
Same Environment
```

Only genetics differs.

---

### Problematic Situation

```text
MZ Twins
   ↑↑↑
Much More Similar Environment

DZ Twins
   ↑
Less Similar Environment
```

Environmental effects become confounded with genetic effects.

---

## Does EEA Hold?

Many studies have tested the EEA.

Results generally suggest:

- MZ twins are treated somewhat more similarly
- However, this typically explains only a small portion of observed twin resemblance

The assumption appears reasonably robust for many traits.

Nevertheless, it should never be ignored.

---

# Assumption 2: Random Mating

## What Is Random Mating?

The twin model assumes that mate selection is random with respect to the trait.

For example:

```text
Tall individuals
```

are assumed not to systematically choose:

```text
Other tall individuals
```

---

## Why Does This Matter?

Suppose highly educated individuals preferentially marry other highly educated individuals.

This phenomenon is called:

## Assortative Mating

---

# Assortative Mating

Examples include:

- Height
- Intelligence
- Educational attainment
- Socioeconomic status

In these cases:

```text
Similar individuals tend to mate.
```

---

## Consequences

If assortative mating exists:

DZ twins may share more than the expected:

$$
50\%
$$

of trait-related genetic variation.

This increases:

$$
r_{DZ}
$$

and can inflate estimates of:

$$
C
$$

while reducing estimates of:

$$
A
$$

---

## Visual Example

Without assortative mating:

```text
DZ Relatedness ≈ 0.50
```

With assortative mating:

```text
DZ Relatedness > 0.50
```

The classical ACE model does not account for this automatically.

---

# Assumption 3: No Gene–Environment Correlation (rGE)

## What Is Gene–Environment Correlation?

Genes can influence the environments individuals experience.

This creates:

## Gene–Environment Correlation

often abbreviated:

```text
rGE
```

---

## Example

Suppose a child inherits musical talent.

That child may:

- Practice more
- Join music programs
- Receive encouragement from teachers

The environment becomes correlated with genetics.

---

## Three Types of rGE

### Passive rGE

Parents provide:

- Genes
- Environment

simultaneously.

Example:

Musical parents provide both musical genes and musical environments.

---

### Evocative rGE

Individuals evoke responses from others.

Example:

A naturally sociable child receives more social interactions.

---

### Active rGE

Individuals seek environments that match their genetic predispositions.

Example:

Athletic individuals seek sports activities.

---

## Why This Matters

The ACE model assumes:

```text
Genes
and
Environment
are independent.
```

Gene–environment correlation violates this assumption.

---

# Assumption 4: No Gene–Environment Interaction (G×E)

## What Is Gene–Environment Interaction?

A gene–environment interaction occurs when:

> The effect of a gene depends on the environment.

or

> The effect of an environment depends on genotype.

---

## Example

Consider a genetic predisposition for obesity.

Under:

```text
Healthy Diet
```

the genetic effect may be small.

Under:

```text
High-Calorie Diet
```

the same genetic predisposition may become much stronger.

---

## Mathematical View

Without interaction:

$$
Phenotype
=
Gene
+
Environment
$$

With interaction:

$$
Phenotype
=
Gene
+
Environment
+
Gene \times Environment
$$

---

## Why This Matters

Classical ACE models assume:

$$
G\times E=0
$$

If interactions exist:

variance components may be misestimated.

---

# Assumption 5: Same Population Assumption

Twin analyses assume:

- MZ twins
- DZ twins

come from the same underlying population.

---

## Example

Problematic design:

```text
MZ twins from one country

DZ twins from another country
```

Differences may reflect:

- Culture
- Environment
- Sampling

rather than genetics.

---

## Desired Situation

```text
MZ and DZ twins sampled from the same population.
```

---

# Assumption 6: Representative Sampling

Twin samples should represent the target population.

---

## Example

Suppose a study includes only:

```text
Highly educated families
```

Heritability estimates may not generalize to the broader population.

---

## Why This Matters

Remember:

Heritability depends on:

- Population
- Environment
- Historical context

Biased sampling can distort estimates.

---

# Assumption 7: Accurate Zygosity Classification

Twin studies require accurate classification.

Individuals must be correctly identified as:

- MZ
- DZ

---

## Consequences of Misclassification

Suppose some DZ twins are incorrectly labeled as MZ.

Observed:

$$
r_{MZ}
$$

decreases.

This may reduce estimated:

$$
A
$$

and inflate:

$$
E
$$

---

## Modern Approaches

Historically:

- Questionnaires
- Physical resemblance

were used.

Today:

- SNP genotyping
- DNA-based relatedness

provide highly accurate classification.

---

# Which Assumption Is Most Important?

Most twin researchers consider:

## Equal Environment Assumption

to be the most critical assumption.

Why?

Because violations directly affect the comparison between:

$$
r_{MZ}
$$

and

$$
r_{DZ}
$$

which forms the basis of heritability estimation.

---

# Are Twin Studies Useless If Assumptions Are Imperfect?

No.

All scientific models rely on assumptions.

The question is not:

```text
Are assumptions perfect?
```

but rather:

```text
Are violations large enough
to meaningfully bias results?
```

Many decades of research suggest that twin studies remain highly informative despite imperfect assumptions.

---

# Modern Validation

One fascinating result from recent genetics research is:

Twin-based heritability estimates often agree surprisingly well with:

- SNP heritability
- Family-based heritability
- Genomic relatedness analyses

This convergence increases confidence in twin-study conclusions.

---

# Simulation: Effect of Environmental Bias

Suppose true:

```python
r_MZ = 0.60
r_DZ = 0.30
```

Falconer:

```python
A = 2*(0.60-0.30)
print(A)
```

Output:

```text
0.60
```

Now imagine extra environmental similarity increases:

```python
r_MZ = 0.75
r_DZ = 0.30
```

Falconer becomes:

```python

print(A)
```

Output:

```text
0.90
```

The heritability estimate becomes substantially inflated.

This demonstrates why EEA is so important.

---

# Strengths and Weaknesses of Twin Studies

| Strengths | Weaknesses |
|------------|------------|
| Natural experiment | Relies on assumptions |
| Powerful for heritability estimation | Cannot prove causation |
| Relatively inexpensive | Vulnerable to some biases |
| Long history of successful use | May oversimplify complex biology |

---

# Key Takeaways

- Twin studies rely on several important assumptions.
- The Equal Environment Assumption is the most critical.
- Assortative mating can bias estimates of A and C.
- Gene–environment correlation violates independence assumptions.
- Gene–environment interaction can complicate variance decomposition.
- Accurate zygosity classification is essential.
- Most assumptions are reasonably supported by empirical research.
- Understanding assumptions is necessary before interpreting heritability estimates.

# Part 8: Building the ACE Model — From Twin Correlations to Covariance Structures

## Introduction

In Part 7, we examined the assumptions underlying classical twin studies.

We now arrive at one of the most important sections of the tutorial:

## Building the ACE Model

Up to this point we have relied on:

- Twin correlations
- Falconer's equations
- Conceptual reasoning

These approaches are valuable, but they do not fully exploit the information contained in twin data.

Structural Equation Modeling (SEM) provides a more rigorous framework.

To understand ACE models, we must first understand:

> How covariance between twins arises from genetics and environment.

This section bridges the gap between:

```text
Twin Correlations
        ↓
Falconer's Formula
        ↓
ACE Structural Equation Model
```

---

# The Core Idea

Recall:

Phenotypic variance can be decomposed as:

$$
V_P = V_A + V_C + V_E
$$

or simply:

$$
P=A+C+E
$$

The ACE model estimates these components by modeling the covariance between twins.

---

# Why Covariance?

Suppose we measure height in twins.

If one twin is tall:

- Does the co-twin also tend to be tall?

If yes:

the twins have positive covariance.

Twin covariance contains information about:

- Genetic similarity
- Environmental similarity

---

# Variance vs Covariance

Recall:

### Variance

Measures variability within individuals.

$$
Var(X)
$$

---

### Covariance

Measures similarity between individuals.

$$
Cov(X,Y)
$$

---

In twin studies:

```text
Variance
```

helps estimate:

$$
A+C+E
$$

while

```text
Covariance
```

helps separate:

$$
A
$$

from

$$
C
$$

---

# Phenotypic Variance

For a single individual:

$$
P=A+C+E
$$

Assuming independence:

$$
Var(P)
=
Var(A)+Var(C)+Var(E)
$$

Therefore:

$$
V_P
=
V_A+V_C+V_E
$$

---

# Path Coefficients

Recall from SEM:

We define path coefficients:

$$
a
$$

$$
c
$$

$$
e
$$

Then:

$$
V_A=a^2
$$

$$
V_C=c^2
$$

$$
V_E=e^2
$$

---

# Total Variance

Thus:

$$
V_P
=
a^2+c^2+e^2
$$

This equation defines the phenotypic variance.

---

# The Twin Covariance Problem

The key question becomes:

> How much covariance should we expect between twins?

The answer depends on:

- Genetic relatedness
- Shared environment

---

# Monozygotic Twins

MZ twins share:

$$
100\%
$$

of additive genetic effects.

They also share:

$$
100\%
$$

of shared environmental effects.

Therefore:

$$
Cov_{MZ}
=
V_A+V_C
$$

or equivalently:

$$
Cov_{MZ}
=
a^2+c^2
$$

---

# Why Is E Missing?

Notice:

$$
E
$$

does not appear.

Why?

Because unique environmental influences are not shared.

Examples:

- Different illnesses
- Different teachers
- Different friends
- Random developmental events

These do not contribute to covariance.

---

# Dizygotic Twins

DZ twins share:

$$
50\%
$$

of additive genetic effects.

They also share:

$$
100\%
$$

of shared environmental effects.

Therefore:

$$
Cov_{DZ}
=
0.5V_A
+
V_C
$$

or

$$
Cov_{DZ}
=
0.5a^2+c^2
$$

---

# The Entire ACE Model Emerges

Remarkably, everything follows from two equations:

### MZ Twins

$$
Cov_{MZ}
=
a^2+c^2
$$

### DZ Twins

$$
Cov_{DZ}
=
0.5a^2+c^2
$$

These are the foundation of ACE modeling.

---

# Visual Representation

```text
MZ Twins

Genetics 100%
Environment 100%

Covariance
=
A + C
```

---

```text
DZ Twins

Genetics 50%
Environment 100%

Covariance
=
0.5A + C
```

---

# Covariance Matrices

SEM works with covariance matrices.

For MZ twins:

$$
\Sigma_{MZ}
=
\begin{bmatrix}
V_P & V_A+V_C \\
V_A+V_C & V_P
\end{bmatrix}
$$

Substituting:

$$
V_P=A+C+E
$$

gives:

$$
\Sigma_{MZ}
=
\begin{bmatrix}
A+C+E & A+C \\
A+C & A+C+E
\end{bmatrix}
$$

---

# Interpretation

Diagonal elements:

```text
Variance
```

Off-diagonal elements:

```text
Covariance
```

---

# DZ Covariance Matrix

Similarly:

$$
\Sigma_{DZ}
=
\begin{bmatrix}
A+C+E & 0.5A+C \\
0.5A+C & A+C+E
\end{bmatrix}
$$

---

# Why Covariance Matrices Matter

The observed data produce:

### Observed MZ Matrix

$$
S_{MZ}
$$

### Observed DZ Matrix

$$
S_{DZ}
$$

The ACE model attempts to reproduce these matrices.

---

# Model Fitting

The workflow becomes:

```text
Observed Covariance Matrix
          ↓
ACE Model
          ↓
Estimate a,c,e
          ↓
Predicted Covariance Matrix
          ↓
Compare With Data
```

---

# Maximum Likelihood Estimation

OpenMx estimates:

$$
a
$$

$$
c
$$

$$
e
$$

using maximum likelihood.

The goal is:

```text
Minimize discrepancy
between

Observed Covariance Matrix

and

Predicted Covariance Matrix
```

---

# Example

Suppose:

$$
a=0.70
$$

$$
c=0.40
$$

$$
e=0.55
$$

Then:

$$
A=0.49
$$

$$
C=0.16
$$

$$
E=0.30
$$

---

# MZ Covariance

Expected covariance:

$$
0.49+0.16
=
0.65
$$

---

# DZ Covariance

Expected covariance:

$$
0.5(0.49)+0.16
=
0.405
$$

---

# Notice Something Interesting

These values correspond closely to:

$$
r_{MZ}\approx0.65
$$

and

$$
r_{DZ}\approx0.40
$$

which are very similar to the workshop example.

This demonstrates how Falconer's formulas and ACE models are closely related.

# Part 11: Model Comparison — ACE vs AE vs CE vs E Models

## Introduction

In Part 10, we learned how to fit an ACE model and interpret:

- Additive genetic effects (A)

- Unique environmental effects (E)

> Do we really need all three components?

- Can C be removed?

# Part 12: Genetic Relatedness — Understanding Expected Relationships in Families

## Introduction

Up to this point, our twin models have relied on a simple assumption:

### Monozygotic Twins

Share:

$$
100\%
$$

of their genetic material.

### Dizygotic Twins

Share:

$$
50\%
$$

of their genetic material.

These numbers may seem obvious, but where do they come from?

More importantly:

- Why do siblings share 50% of their DNA?
- Why do cousins share 12.5%?
- Why do grandparents share 25% with grandchildren?
- How is relatedness calculated mathematically?

Understanding genetic relatedness is essential because it forms the bridge between:

```text
Classical Twin Studies
           ↓
Family Studies
           ↓
Genetic Relationship Matrices (GRM)
           ↓
SNP Heritability
           ↓
Modern Genomic Analysis
```

This section introduces the mathematics of biological relatedness.

---

# What Is Genetic Relatedness?

Genetic relatedness measures:

> The probability that two individuals share a genetic variant inherited from a common ancestor.

It quantifies how genetically similar two individuals are expected to be.

---

# Why Relatedness Matters

Suppose we observe:

```text
Parent and Child
```

The child receives:

- Half of its DNA from the mother
- Half of its DNA from the father

Therefore:

```text
Parent and Child
```

must share a substantial fraction of their genome.

Relatedness provides a way to quantify this sharing.

---

# The Coefficient of Relationship

The standard measure is called:

## Coefficient of Relationship

Usually denoted:

$$
R
$$

or

$$
r
$$

The coefficient represents:

> The expected proportion of shared genetic material.

---

# Values of Relatedness

The coefficient ranges between:

$$
0
\le
R
\le
1
$$

where:

| Value | Interpretation |
|---------|---------------|
| 0 | No genetic relationship |
| 0.25 | Moderate relationship |
| 0.50 | First-degree relatives |
| 1.00 | Genetically identical |

---

# Why Relatives Share DNA

Humans possess:

```text
23 chromosome pairs
```

At reproduction:

piHat estimates relatedness between a pair of individuals.

```text

```

We need a matrix describing relationships among all pairs.

## Genetic Relationship Matrix (GRM)

---

# What Is a GRM?

A GRM is a matrix where:

Element:

$$
G_{ij}
$$

represents genetic similarity between:

```text
Individual i

and

Individual j
```

---

# Example

Five individuals:

$$
GRM=
\begin{bmatrix}
1.00 & 0.02 & -0.01 & 0.01 & 0.00\\
0.02 & 1.00 & 0.03 & -0.02 & 0.01\\
-0.01 & 0.03 & 1.00 & 0.01 & 0.02\\
0.01 & -0.02 & 0.01 & 1.00 & -0.01\\
0.00 & 0.01 & 0.02 & -0.01 & 1.00
\end{bmatrix}
$$

---

# Interpretation

Diagonal:

$$
1
$$

because individuals are perfectly related to themselves.

---

Off-diagonal:

represent pairwise genetic similarity.

---

# Why Are Some Entries Negative?

This confuses many beginners.

Negative GRM values do **not** mean:

```text
Negative biological relatedness
```

They simply indicate:

```text
Less sharing than average.
```

after standardization.

---

# Constructing the GRM

Suppose:

Genotype coding:

| Genotype | Value |
|----------|--------|
| AA | 0 |
| Aa | 1 |
| aa | 2 |

---

For SNP:

$j$

with allele frequency:

$$
p_j
$$

standardized genotype:

$$
z_{ij}
=
\frac{x_{ij}-2p_j}
{\sqrt{2p_j(1-p_j)}}
$$

---

# GRM Formula

The GRM is:



where:

- \(Z\) = standardized genotype matrix
- \(M\) = number of SNPs

This is one of the most important equations in statistical genetics.

---

# Intuition

The GRM asks:

> Across all SNPs, how similar are two individuals relative to population expectations?

---

# Example in Python

```python
import numpy as np

X = np.array([

[0,1,2],
[1,1,1],
[2,1,0]

])

p = X.mean(axis=0)/2

Z = (
    X - 2*p
) / np.sqrt(
    2*p*(1-p)
)

GRM = (
    Z @ Z.T
) / X.shape[1]

print(GRM)
```

# Part 14: SNP Heritability and GREML — Estimating Heritability from Unrelated Individuals

## Introduction

In Part 13, we introduced one of the most important objects in modern statistical genetics:

## The Genetic Relationship Matrix (GRM)

The GRM allows us to estimate genetic similarity directly from DNA rather than relying on:

- Twin status
- Family relationships
- Pedigrees

This raises a fascinating question:

> Can we estimate heritability using completely unrelated individuals?

Surprisingly, the answer is:

## Yes

This breakthrough was achieved through:

### GCTA

(Genome-wide Complex Trait Analysis)

and

### GREML

(Genomic Restricted Maximum Likelihood)

These methods revolutionized quantitative genetics because they made it possible to estimate heritability using:

```text
Large population cohorts
```

without requiring twins or family data.

Today, GREML is one of the most widely used methods for estimating:

## SNP Heritability

---

# Why Was GREML Needed?

Classical twin studies estimate:

$$
V_P=A+C+E
$$

using:

```text
Expected Relatedness
```

For example:

| Pair | Relatedness |
|--------|-------------|
| MZ Twins | 1.0 |
| DZ Twins | 0.5 |

---

However:

Twin studies require:

- Twin registries
- Family data
- Specific study designs

These resources are expensive and not always available.

---

# The Key Insight

Suppose we have:

```text
100,000 unrelated individuals
```

At first glance:

```text
Everyone appears unrelated.
```

But this is not entirely true.

Some individuals are:

```text
Slightly more genetically similar
```

than others.

---

# Example

Person A and Person B:

```text
GRM = 0.02
```

Person A and Person C:

```text
GRM = -0.01
```

These differences are tiny.

Yet across hundreds of thousands of individuals they contain information about genetic influence.

---

# The Core Idea of GREML

Suppose genetics influences height.

Then:

> Individuals who are slightly more genetically similar should also be slightly more phenotypically similar.

GREML quantifies this relationship.

---

# From Twin Logic to Population Logic

Twin studies:

```text
More genetic similarity
↓
More phenotypic similarity
```

GREML:

```text
Slightly more genetic similarity
↓
Slightly more phenotypic similarity
```

Same principle.

Different scale.

---

# SNP Heritability

GREML estimates:

## SNP Heritability

denoted:

$$
h^2_{SNP}
$$

---

# Definition

SNP heritability is:

> The proportion of phenotypic variance explained by the SNPs included in the analysis.

---

# Important Distinction

Twin heritability:

$$
h^2_{Twin}
$$

captures:

```text
All genetic effects
```

including:

- Rare variants
- Common variants
- Structural variants
- Untyped variants

---

GREML estimates:

$$
h^2_{SNP}
$$

which includes only:

```text
Observed SNPs
```

and variants tagged by those SNPs.

---

# Why SNP Heritability Is Usually Smaller

Typical example:

### Height

Twin heritability:

$$
80\%
$$

SNP heritability:

$$
50-60\%
$$

---

This difference led to the famous concept of:

## Missing Heritability

---

# The Missing Heritability Problem

Early GWAS identified:

```text
Only a small fraction
of expected heritability.
```

Researchers asked:

> Where is the missing heritability?

GREML helped demonstrate that many common variants collectively explain substantial variance even when individual SNPs fail to reach genome-wide significance.

---

# The GREML Model

Phenotype:

$$
y
=
X\beta
+
g
+
e
$$

where:

| Term | Meaning |
|--------|---------|
| $y$ | Phenotype |
| $X\beta$ | Fixed effects |
| $g$ | Genetic effect |
| $e$ | Residual error |

---

# Fixed Effects

Examples:

```text
Age
Sex
Batch
Principal Components
```

These are modeled explicitly.

---

# Random Genetic Effect

The key innovation is:

$$
g
\sim
N(0,G\sigma_g^2)
$$

where:

- \(G\) = GRM
- \(\sigma_g^2\) = genetic variance

---

# Residual Effect

Similarly:

$$
e
\sim
N(0,I\sigma_e^2)
$$

where:

- \(I\) = identity matrix
- \(\sigma_e^2\) = environmental variance

---

# Total Variance

Therefore:



This is the central equation of GREML.

---

# Interpretation

The model asks:

> Can phenotypic similarity be predicted from genetic similarity?

If yes:

genetic variance is large.

If no:

genetic variance is small.

---

# Estimating SNP Heritability

Once variance components are estimated:

$$
h^2_{SNP}
=
\frac{\sigma_g^2}
{\sigma_g^2+\sigma_e^2}
$$

---

# Example

Suppose:

$$
\sigma_g^2=0.45
$$

$$
\sigma_e^2=0.55
$$

Then:

$$
h^2_{SNP}
=
0.45
$$

Interpretation:

```text
45% of variance
is explained by SNPs.
```

---

# Why REML?

GREML uses:

## Restricted Maximum Likelihood

(REML)

instead of ordinary maximum likelihood.

---

# Why Not Ordinary ML?

Ordinary ML tends to underestimate variance components.

REML corrects for uncertainty introduced by estimating fixed effects.

---

# Intuition Behind REML

ML estimates:

```text
Mean
+
Variance
```

simultaneously.

REML first accounts for:

```text
Mean Structure
```

and then estimates:

```text
Variance Components
```

This produces less biased estimates.

---

# Relationship to Twin Models

Twin model:

$$
V_P=A+C+E
$$

GREML:

$$
V_P=V_G+V_E
$$

---

# Key Difference

Twin studies estimate:

```text
A
C
E
```

GREML estimates:

```text
VG
VE
```

without separating:

```text
A
and
C
```

because family information is unavailable.

---

```text

explained by SNPs
```

---

# Standard Error

```text

```

Interpretation:


````python
# Part 15: REML in Depth — The Mathematics Behind Variance Component Estimation

## Introduction

In Part 14, we learned that GCTA estimates SNP heritability using:

## GREML

(Genomic Restricted Maximum Likelihood)

and that the fundamental model is:

$$
y = X\beta + g + e
$$

where:

- \(y\) = phenotype
- \(X\beta\) = fixed effects
- \(g\) = genetic effects
- \(e\) = residual effects

We also saw that the total variance is:

$$
V = G\sigma_g^2 + I\sigma_e^2
$$

where:

- \(G\) = Genetic Relationship Matrix (GRM)
- \(I\) = Identity Matrix

But an important question remains:

> How does GCTA actually estimate \(\sigma_g^2\) and \(\sigma_e^2\)?

The answer lies in one of the most important statistical methods in quantitative genetics:

## Restricted Maximum Likelihood (REML)

REML is the mathematical engine behind:

- GREML
- Animal breeding
- Mixed models
- Genomic prediction
- Random effects models
- Linear mixed models (LMM)

Understanding REML is one of the biggest steps toward mastering modern statistical genetics.

---

# Why We Need REML

Suppose we have:

```text
Height
```

for 100,000 individuals.

We want to estimate:

$$
V_G
$$

and

$$
V_E
$$

---

## The Problem

Estimating variances is surprisingly difficult.

Why?

Because we first need to estimate:

$$
\mu
$$

the population mean.

Any uncertainty in the mean affects variance estimates.

---

# A Simple Example

Suppose observations are:

```text
10
12
14
16
18
```

Mean:

$$
\bar{x}=14
$$

Variance:

$$
\frac{\sum(x_i-\bar{x})^2}{n-1}
$$

Notice:

The variance calculation depends on the estimated mean.

---

# The Bias Problem

Imagine we estimate:

```text
Mean
```

and

```text
Variance
```

simultaneously.

Ordinary Maximum Likelihood (ML) tends to:

```text
Underestimate variance.
```

This is because some variation is "used up" estimating the mean.

---

# Classical Example

Population variance:

$$
\sigma^2
$$

Maximum likelihood estimate:

$$
\hat{\sigma}_{ML}^2
=
\frac1n
\sum(x_i-\bar{x})^2
$$

---

However, statistics textbooks teach:

$$
s^2
=
\frac1{n-1}
\sum(x_i-\bar{x})^2
$$

Why?

Because:

```text
One degree of freedom
was spent estimating the mean.
```

---

# REML Generalizes This Idea

REML extends this correction to:

- Multiple fixed effects
- Mixed models
- Genomic relationship matrices
- High-dimensional genetics

---

# Maximum Likelihood Refresher

Before understanding REML, let's briefly revisit ML.

---

## Likelihood

The likelihood answers:

> How probable are the observed data given specific parameter values?

Suppose:

$$
y
\sim
N(\mu,\sigma^2)
$$

Likelihood:

$$
L(\mu,\sigma^2)
=
P(y|\mu,\sigma^2)
$$

---

# Maximum Likelihood Estimation

ML finds:

$$
\hat{\mu}
$$

and

$$
\hat{\sigma}^2
$$

that maximize:

$$
L
$$

or equivalently:

$$
\log L
$$

---

# Why ML Is Not Ideal for Variance Components

ML treats:

```text
Fixed effects
```

and

```text
Variance components
```

simultaneously.

This creates bias.

Especially when:

- many covariates exist
- sample sizes are limited

---

# The Core Idea of REML

REML asks:

> What if we remove the fixed effects first?

Then estimate variance components using only the remaining variation.

---

# A Useful Analogy

Imagine trying to measure:

```text
Noise
```

in a room.

If people are talking:

```text
Speech
```

acts like a fixed effect.

Before measuring noise:

you first remove the speech.

REML follows the same philosophy.

---

# The Mixed Model

Recall:

$$
y
=
X\beta
+
Zu
+
e
$$

---

# Components

### Fixed Effects

$$
X\beta
$$

Examples:

- Sex
- Age
- Principal components

---

### Random Effects

$$
Zu
$$

Examples:

- Polygenic effects
- Family effects

---

### Residual

$$
e
$$

Unexplained variation.

---

# Variance Structure

Random effects:

$$
u
\sim
N(0,G)
$$

Residuals:

$$
e
\sim
N(0,R)
$$

Therefore:

$$
Var(y)
=
ZGZ^T
+
R
$$

---

# In GREML

Specifically:



This is the covariance structure that GCTA fits.

---

# How REML Removes Fixed Effects

The clever trick:

Construct linear combinations of observations that eliminate:

$$
X\beta
$$

entirely.

---

# Simple Example

Suppose:

$$
y_1
=
\mu+e_1
$$

$$
y_2
=
\mu+e_2
$$

Subtract:

$$
y_1-y_2
=
e_1-e_2
$$

Notice:

$$
\mu
$$

disappears.

---

# This Is the Essence of REML

Instead of modeling:

```text
Raw observations
```

REML models:

```text
Residual variation
```

after removing fixed effects.

---

# Degrees of Freedom

Remember:

Variance estimation loses degrees of freedom.

If:

$$
p
$$

fixed effects are estimated,

effective sample size becomes:

$$
n-p
$$

REML automatically accounts for this.

---

# REML Likelihood

Ordinary ML maximizes:

$$
L(y)
$$

REML maximizes:

$$
L(K^Ty)
$$

where:

$$
K^TX=0
$$

meaning:

the transformed observations contain no fixed-effect information.

---

# Don't Panic About the Math

Many genetics papers write:

$$
L_R=
-\frac12
\left[
\log|V|
+
\log|X^TV^{-1}X|
+
y^TPy
\right]
$$

This looks intimidating.

Conceptually it simply means:

```text
Estimate variance
after accounting for
fixed effects.
```

---

# Geometric Interpretation

Imagine phenotype space:

```text
Total Variation
```

consists of:

```text
Fixed Effects
+
Residual Variation
```

REML projects data into:

```text
Residual Space
```

and estimates variance there.

---

# Why REML Works Well in Genetics

Genetic studies often include:

- Age
- Sex
- PCs
- Batch effects
- Study site

If ML is used:

variance estimates become biased.

REML properly accounts for these covariates.

---

# REML in GCTA

Command:

```bash
gcta64 \
--grm myGRM \
--pheno phenotype.txt \
--reml \
--out results
```

The flag:

```bash
--reml
```

triggers REML optimization.

---

# What GCTA Estimates

GCTA estimates:

$$
\sigma_g^2
$$

and

$$
\sigma_e^2
$$

directly.

---

# Example Output

```text
V(G) = 0.42

SE = 0.04

V(e) = 0.58

SE = 0.04
```

---

# SNP Heritability

Computed as:

$$
h^2_{SNP}
=
\frac{0.42}
{0.42+0.58}
=
0.42
$$

Interpretation:

```text
42% of phenotypic variance
is explained by SNPs.
```

---

# Why REML Became So Important

Before REML:

estimating variance components in large genetic datasets was difficult.

REML enabled:

- Animal breeding
- Twin studies
- GREML
- Polygenic prediction
- Genomic selection

---

# REML vs Ordinary Least Squares

| Feature | OLS | REML |
|----------|------|------|
| Fixed effects | ✓ | ✓ |
| Random effects | ✗ | ✓ |
| Variance components | ✗ | ✓ |
| Heritability estimation | ✗ | ✓ |
| GRM usage | ✗ | ✓ |

---

# REML vs ML

| Feature | ML | REML |
|----------|------|------|
| Estimates fixed effects | ✓ | ✓ |
| Estimates variance components | ✓ | ✓ |
| Variance bias correction | ✗ | ✓ |
| Preferred in quantitative genetics | ✗ | ✓ |

---

# Simulation Example

Generate phenotype:

```python
import numpy as np

np.random.seed(42)

n = 1000

genetic = np.random.normal(
    0,
    np.sqrt(0.4),
    n
)

environment = np.random.normal(
    0,
    np.sqrt(0.6),
    n
)

phenotype = (
    genetic +
    environment
)
```

---

# Variance Components

```python
print(
    np.var(genetic)
)

print(
    np.var(environment)
)

print(
    np.var(phenotype)
)
```

Expected:

```text
Genetic ≈ 0.4

Environmental ≈ 0.6

Total ≈ 1.0
```

---

# Conceptual Connection to ACE Models

ACE:

$$
V_P=A+C+E
$$

GREML:

$$
V_P=V_G+V_E
$$

Both require:

```text
Variance Component Estimation
```

REML is the statistical machinery that makes this possible.

---

# Why Every Statistical Geneticist Learns REML

REML appears repeatedly in:

- GCTA
- BOLT-LMM
- LDAK
- Genomic prediction
- Animal breeding
- Mixed models
- GWAS correction

Understanding REML provides a foundation for many advanced methods.

---

# Key Takeaways

- REML stands for Restricted Maximum Likelihood.
- REML estimates variance components after accounting for fixed effects.
- Ordinary ML tends to underestimate variance.
- REML corrects this bias.
- GREML relies on REML to estimate SNP heritability.
- The central variance model is:

$$
Var(y)=G\sigma_g^2+I\sigma_e^2
$$

- REML is fundamental to modern statistical genetics.
- Many GWAS and genomic prediction methods are built on REML principles.

---

````

## Introduction

- BMI and Type 2 Diabetes
- Educational attainment and IQ
- Height and BMI

Observationally, these traits may appear correlated.

But are they genetically correlated?

To answer this question, statistical geneticists developed:

# Genetic Correlation

Genetic correlation has become one of the most important concepts in modern human genetics because it allows us to quantify:

> How much of the genetic architecture is shared between two traits.

---

# What Is Genetic Correlation?

Genetic correlation measures:

> The extent to which genetic effects influencing one trait also influence another trait.

Usually denoted:

$$
r_g
$$

---

Suppose:



```text

```

decrease:

```text
ADHD Risk
```

Then:

$$
r_g < 0
$$

---

These are not the same.

---

## Phenotypic Correlation

Observed relationship between traits.

Example:

$$

=
0.6
$$

between:

- Height
- Weight

Relationship between genetic influences.

Example:

$$
r_g
=
0.3
$$

A high phenotypic correlation does not necessarily imply a high genetic correlation.

# Part 19: Mendelian Randomization (MR) — Using Genetics to Infer Causality

## Introduction

In Part 18, we learned about:

## Genetic Correlation

which tells us whether two traits share genetic influences.

For example:

| Trait Pair | Genetic Correlation |
|------------|--------------------|
| Depression ↔ Anxiety | High |
| BMI ↔ Type 2 Diabetes | Moderate |
| Schizophrenia ↔ Bipolar Disorder | High |

However, genetic correlation cannot answer an important question:

> Does Trait A cause Trait B?

For example:

- Does obesity cause depression?
- Does smoking cause lung cancer?
- Does education influence Alzheimer's risk?
- Does vitamin D reduce disease risk?

Observational studies often struggle with these questions because of:

- Confounding
- Reverse causation
- Selection bias

Mendelian Randomization (MR) was developed to address this problem.

MR is one of the most important causal inference tools in modern genetics and epidemiology.

---

# The Fundamental Problem

Suppose we observe:

```text
Higher BMI
↓
Higher risk of Type 2 Diabetes
```

Does this mean:

```text
BMI causes diabetes?
```

Not necessarily.

---

# Possible Explanations

### Explanation 1

BMI causes diabetes.

```text
BMI
 ↓
Diabetes
```

---

### Explanation 2

Diabetes causes BMI changes.

```text
Diabetes
 ↓
BMI
```

---

### Explanation 3

A third factor affects both.

```text
Lifestyle
 ↙      ↘

BMI   Diabetes
```

This is called:

## Confounding

---

# Why Observational Studies Struggle

Suppose we study:

```text
Coffee Consumption
```

and

```text
Heart Disease
```

People who drink coffee may differ in:

- Smoking habits
- Income
- Education
- Diet
- Stress levels

These factors may influence the results.

---

# The Idea Behind MR

MR uses genetic variants as:

## Instrumental Variables (IVs)

to infer causality.

---

# Why Genetics?

Genetic variants are assigned at conception.

Therefore:

```text
Genes occur before disease.
```

This helps avoid:

- Reverse causation
- Many confounding effects

---

# The Natural Randomization

Gregor Mendel's laws imply:

```text
Parental alleles
↓
Randomly transmitted
↓
Offspring genotype
```

This resembles a randomized clinical trial.

---

```text

↓

↓
Outcome
```

---

```text

↓
Exposure
↓
Outcome
```

---

# Core Components of MR

MR involves three variables:

### Instrument

Genetic variant

$$
G
$$

---

### Exposure

Risk factor

$$
X
$$

Examples:

- BMI
- Smoking
- Cholesterol

---

### Outcome

Disease or trait

$$
Y
$$

Examples:

- Diabetes
- Cancer
- Depression

---

# Causal Diagram

```text
G
↓
X
↓
Y
```

The goal:

Determine whether:

$$
X
\rightarrow
Y
$$

is causal.

---

# Example

Suppose:

### Exposure

BMI

---

### Outcome

Type 2 Diabetes

---

### Instrument

FTO SNP

Known to influence BMI.

---

The logic:

```text
FTO
↓
BMI
↓
Diabetes
```

If BMI truly causes diabetes:

individuals carrying BMI-increasing alleles should also have higher diabetes risk.

---

# The Three Core Assumptions

MR relies on three assumptions.

These assumptions are absolutely critical.

---

# Assumption 1: Relevance

The genetic variant must influence the exposure.

Diagram:

```text
G
↓
X
```

Mathematically:

$$
G \rightarrow X
$$

must exist.

---

# Example

BMI-associated SNP:

✓ Valid

---

Random SNP unrelated to BMI:

✗ Invalid

---

# Assumption 2: Independence

The instrument must not be associated with confounders.

Diagram:

```text
Confounder

↙      ↘

G      X
```

must not occur.

---

Examples of confounders:

- Smoking
- Socioeconomic status
- Diet

---

# Assumption 3: Exclusion Restriction

The instrument affects the outcome only through the exposure.

Diagram:

Valid:

```text
G
↓
X
↓
Y
```

Invalid:

```text
G
↘
  Y
```

direct pathway.

---

# Why This Is Important

Suppose:

SNP affects:

- BMI
- Diabetes directly

Then:

```text
MR becomes biased.
```

---

# Horizontal Pleiotropy

One of the biggest challenges in MR.

---

# Definition

A genetic variant influences multiple traits.

Example:

```text
SNP
 ↙   ↘

BMI  Diabetes
```

independently.

---

This violates:

## Exclusion Restriction

---

# Vertical Pleiotropy

Valid situation:

```text
SNP
↓
BMI
↓
Diabetes
```

This is precisely what MR wants.

---

$$

=
\frac{\beta_{GY}}
{\beta_{GX}}

---

Denominator:

Effect of SNP on exposure.

$$

$$

units.

---

$$
0.1
$$

units.

---

MR estimate:

$$
0.1/0.2
=
0.5
$$

Interpretation:

```text

```

# Part 21: Fine-Mapping and Credible Sets — Identifying the Most Likely Causal Variants

## Introduction

In Part 20, we learned how:

- eQTL analysis
- Colocalization
- TWAS

help connect GWAS signals to genes and biological mechanisms.

However, another major challenge remains.

Suppose a GWAS identifies:

```text
Chromosome 3

Lead SNP:

rs12345

p = 2 × 10⁻¹⁵
```

Many beginners assume:

> rs12345 is the causal variant.

Unfortunately, this is usually not true.

Because of:

## Linkage Disequilibrium (LD)

many nearby SNPs are highly correlated.

As a result:

```text
The lead SNP
may not be causal.
```

Instead, it may simply tag the true causal variant.

This leads to one of the most important topics in post-GWAS analysis:

# Fine-Mapping

---

# The GWAS Resolution Problem

A significant GWAS locus often contains:

```text
10
50
100
500+
SNPs
```

in strong LD.

---

# Example

```text
GWAS Peak

rs100
rs101
rs102
rs103
rs104
rs105
```

All show:

```text
p < 5 × 10⁻⁸
```

Which one is causal?

GWAS alone cannot tell us.

---

# Why This Happens

Suppose:

```text
rs103
```

is genuinely causal.

Nearby SNPs:

```text
rs100
rs101
rs102
rs104
rs105
```

are inherited together.

Consequently:

all SNPs appear associated.

---

# Visual Illustration

```text
True Causal Variant

          rs103
            ▲

LD Block

rs100 rs101 rs102 rs103 rs104 rs105
```

All SNPs inherit the signal.

---

# The Goal of Fine-Mapping

Fine-mapping attempts to answer:

> Which SNPs are most likely to be causal?

Instead of:

```text
Hundreds of SNPs
```

we aim for:

```text
A handful of SNPs
```

or ideally:

```text
One SNP
```

---

# Fine-Mapping vs GWAS

GWAS asks:

```text
Which regions are associated?
```

Fine-mapping asks:

```text
Which variants are causal?
```

---

# Why Fine-Mapping Matters

Understanding causal variants helps:

- Functional experiments
- Drug development
- Mechanistic biology
- Gene prioritization
- Precision medicine

---

# A Simple Example

Suppose:

| SNP | p-value |
|------|---------|
| rs1 | 1e-10 |
| rs2 | 2e-10 |
| rs3 | 4e-10 |
| rs4 | 8e-10 |

All SNPs are highly significant.

Without fine-mapping:

we cannot distinguish among them.

---

# The Bayesian Perspective

Modern fine-mapping is primarily Bayesian.

Instead of asking:

```text
Significant?
```

we ask:

> What is the probability this SNP is causal?

---

# Posterior Probability of Causality

For each SNP:

$$
P(Causal \mid Data)
$$

This quantity is often called:

## Posterior Inclusion Probability

or

## PIP

---

# Interpretation

### SNP A

```text
PIP = 0.80
```

Interpretation:

```text
80% probability
of being causal.
```

---

### SNP B

```text
PIP = 0.02
```

Interpretation:

Very unlikely.

---

# Example Fine-Mapping Result

| SNP | PIP |
|------|------|
| rs101 | 0.02 |
| rs102 | 0.05 |
| rs103 | 0.82 |
| rs104 | 0.08 |
| rs105 | 0.03 |

Strong evidence points toward:

```text
rs103
```

---

# What Is a Credible Set?

Rather than identifying a single SNP,

Bayesian methods often produce:

## Credible Sets

---

# Definition

A credible set is:

> The smallest set of variants containing the causal variant with a specified probability.

Typically:

```text
95%
```

or

```text
99%
```

---

# Example

Suppose:

| SNP | PIP |
|------|------|
| rs103 | 0.50 |
| rs104 | 0.25 |
| rs105 | 0.15 |
| rs106 | 0.05 |
| rs107 | 0.05 |

---

Cumulative probability:

```text
rs103 = 0.50

rs103 + rs104 = 0.75

rs103 + rs104 + rs105 = 0.90

rs103 + rs104 + rs105 + rs106 = 0.95
```

---

# 95% Credible Set

Contains:

```text
rs103
rs104
rs105
rs106
```

Interpretation:

> There is a 95% probability that the true causal variant lies within this set.

---

# Why Credible Sets Are Useful

Instead of:

```text
500 SNPs
```

we may reduce uncertainty to:

```text
5 SNPs
```

This dramatically improves biological follow-up.

---

# Bayes Factors

Fine-mapping often begins by computing:

## Bayes Factors

For SNP:

$$
BF_j
=
\frac{
P(Data|Causal)
}
{
P(Data|Not\ Causal)
}
$$

---

# Interpretation

Large BF:

```text
Strong support for causality.
```

Small BF:

```text
Weak support.
```

---

# Approximate Bayes Factor

A commonly used approximation:

$$
ABF
=
\sqrt{
\frac{V}
{V+W}
}
\exp
\left(
\frac{Z^2W}
{2(V+W)}
\right)
$$

where:

- \(V\) = variance of effect estimate
- \(W\) = prior variance
- \(Z\) = GWAS Z-score

Many fine-mapping tools use variants of this approach.

---

# Single Causal Variant Assumption

Early methods assumed:

```text
One causal variant
per locus.
```

---

# Reality

Many loci contain:

```text

```

Example:

```text

Signal 2

Signal 3
```

within the same region.

---

```python
n_snps = 1000
```

# Key Takeaways

- Genetic architecture describes how genetic variation influences traits.
